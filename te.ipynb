      <h3>Why "Floating-Point"?</h3>
      <p>The term refers to the decimal point's ability to "float" among different positions in a number depending on magnitude.</p>
    </div>
    <div class="card red">
      <h3>Precision Issues</h3>
      <p>Floating-point numbers are internally represented as binary fractions. Hence, many decimal fractions cannot be exactly represented, leading to approximation.</p>
    </div>
  </div>
</div>

```python
# Note how converting from decimal to binary representation is not exact
1.1 + 2.2 == 3.3
```
