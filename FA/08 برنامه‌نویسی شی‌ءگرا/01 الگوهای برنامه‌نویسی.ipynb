  <h2>۱. مروری بر الگوهای برنامه‌نویسی</h2>
  <div class="container">

      <h3>دستوری (Imperative) در برابر اعلانی (Declarative)</h3>
      <p>الگوها به طور کلی به دو دسته اصلی تقسیم می‌شوند:</p>

      <h3>رویکرد «چگونه؟»</h3>
      <p>مثل دستور پخت غذا، شما یک سری دستورالعمل پشت سر هم به کامپیوتر می‌دهید. این روش از حلقه‌ها، شروط و تغییر متغیرها استفاده می‌کند.</p>

      <p>وظایف را به توابع (رویه‌ها) تقسیم می‌کند. برای کارهای خطی و ساده بهترین انتخاب است.</p>

```python
def sum_of_evens(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:
            total += num
    return total
```

```python
numbers = [1, 2, 3, 4, 5, 6]
result = sum_of_evens(numbers)
result
```

      <p>کد را در قالب اشیاء (Objects) سازماندهی می‌کند که شامل داده‌ها و متدها هستند. مفاهیم اصلی آن <strong>کپسوله‌سازی (Encapsulation)</strong>، <strong>وراثت (Inheritance)</strong> و <strong>چندریختی (Polymorphism)</strong> است.</p>

```python
class Circle:
    pi = 3.14159

    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return self.pi * (self.radius ** 2)
```

```python
circle1 = Circle(5)
circle2 = Circle(10)

print(f"مساحت دایره اول با شعاع {circle1.radius} برابر است با {circle1.calculate_area()}.")
print(f"مساحت دایره دوم با شعاع {circle2.radius} برابر است با {circle2.calculate_area()}.")
```

  <h2>۳. الگو اعلانی (Declarative)</h2>
  <div class="container">
    <div class="card purple">
      <h3>رویکرد «چه چیزی؟»</h3>
      <p>روی توصیف نتیجه‌ی دلخواه تمرکز دارد و مراحل کنترل و اجرای دقیق (Control flow) را از دید شما پنهان می‌کند.</p>
      <img src="../../images/imperative-declarative.gif" alt="Imperative vs Declarative">
    </div>
    <div class="card purple">
      <h3>برنامه‌نویسی تابعی (Functional Programming)</h3>
      <p>بر اساس توابع خالص (Pure Functions) و تغییرناپذیری (Immutability) استوار است. پایتون از طریق ابزارهایی مثل <code>lambda</code>، <code>map</code> و <code>filter</code> از این سبک پشتیبانی می‌کند.</p>

```python
numbers = [1, 2, 3, 4, 5, 6]
result = sum(filter(lambda x: x % 2 == 0, numbers))
result
```
