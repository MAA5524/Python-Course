<div class="section-card" id="toc3_">
  <h2>۳. اعضای محافظت‌شده (Protected)</h2>
  <div class="container">
    <div class="card purple">
      <h3>تک زیرخط (<code>_</code>)</h3>
      <p>وقتی نام یک متغیر را با یک <code>_</code> شروع می‌کنیم، به برنامه‌نویسان دیگر می‌گوییم: «این متغیر برای استفاده داخلی کلاس طراحی شده است». پایتون جلوی دسترسی به آن را نمی‌گیرد، اما توسعه‌دهندگان به این قرارداد احترام می‌گذارند.</p>
    </div>
  </div>
</div>

```python
class Vehicle:
    def __init__(self, make, fuel_capacity):
        self.make = make                        # عمومی (Public)
        self._fuel_capacity = fuel_capacity     # محافظت‌شده (Protected)
        self._fuel_level = fuel_capacity        # محافظت‌شده

    def _consume_fuel(self, amount):            # متد محافظت‌شده
        self._fuel_level -= amount
        if self._fuel_level < 0:
            self._fuel_level = 0
        print(f"Fuel left in {self.make}: {self._fuel_level} liters")
```

```python
class Sedan(Vehicle):
    def drive(self, distance):
        print(f"Driving the sedan for {distance} miles.")
        fuel_consumption_rate = 0.04
        self._consume_fuel(distance * fuel_consumption_rate) # استفاده در زیرکلاس
```

```python
class SUV(Vehicle):
    def drive(self, distance):
        print(f"Driving the SUV for {distance} miles.")
        fuel_consumption_rate = 0.07
        self._consume_fuel(distance * fuel_consumption_rate)
```

```python
# ساخت نمونه‌ها
toyota_camry = Sedan("Toyota Camry", 60)
ford_explorer = SUV("Ford Explorer", 70)
```

```python
# حرکت کردن با ماشین‌ها
toyota_camry.drive(200)
```

```python
ford_explorer.drive(200)
```

<div class="section-card" id="toc4_">
  <h2>۴. اعضای خصوصی (Private)</h2>
  <div class="container">
    <div class="card red">
      <h3>جفت زیرخط (<code>__</code>)</h3>
      <p>این کپسوله‌سازی واقعی در پایتون است. با قرار دادن <code>__</code> در ابتدای نام، پایتون فرآیندی به نام <strong>Name Mangling</strong> را انجام می‌دهد و نام متغیر را در پس‌زمینه تغییر می‌دهد تا از بیرون کلاس به هیچ وجه قابل دسترسی نباشد.</p>
    </div>
  </div>
</div>

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner             # ویژگی عمومی
        self.__balance = balance       # ویژگی خصوصی (Private)

    def __update_balance(self, amount):  # متد خصوصی
        self.__balance += amount
        return self.__balance

    def deposit(self, amount):         # متد عمومی به عنوان رابط
        if amount > 0:
            new_balance = self.__update_balance(amount)
            print(f"Deposit successful. New balance: {new_balance}")
        else:
            print("Deposit amount must be positive.")
```

```python
# ساخت حساب کاربری
my_account = Account("John Doe", 1000)
```

```python
# تلاش برای دسترسی به ویژگی‌های خصوصی از بیرون (باعث ارور می‌شود)
# my_account.__balance  # این خط خطای AttributeError تولید می‌کند
```

```python
my_account.deposit(500)  # این متد به درستی کار می‌کند
```

<div class="section-card" id="toc6_">
  <h2>۵. مثال کاربردی: سیستم پست وبلاگ</h2>
  <div class="container">
    <div class="card blue">
      <h3>سناریو</h3>
      <p>کلاسی برای مدیریت مقالات وبلاگ (<code>BlogPost</code>) می‌نویسیم که: عنوان آن <strong>عمومی</strong>، متن محتوای آن <strong>محافظت‌شده</strong>، و وضعیت انتشار آن <strong>خصوصی</strong> است.</p>
    </div>
  </div>
</div>

```python
class BlogPost:
    def __init__(self, title, content):
        self.title = title       # عمومی
        self._content = content  # محافظت‌شده
        self.__status = "پیش‌نویس" # خصوصی

    def edit_content(self, new_content):  # متد عمومی برای ویرایش
        self._content = new_content
        print("محتوای پست به‌روزرسانی شد.")

    def __publish(self):  # متد خصوصی برای انتشار نهایی
        self.__status = "منتشر شده"
        print("پست با موفقیت منتشر شد.")

    def make_public(self):  # متد عمومی (رابط)
        if self.__can_publish():
            self.__publish()
        else:
            print("محتوا کوتاه است؛ امکان انتشار پست وجود ندارد.")

    def __can_publish(self):  # متد خصوصی چک کردن شرایط
        return len(self._content) > 100  # شرط فرضی انتشار
```

```python
# ساخت یک پست وبلاگ
my_post = BlogPost("اولین پست من", "این محتوای اولیه پست وبلاگ من است که خیلی کوتاه است.")
```

```python
my_post.make_public()
```
