  <div class="container">
    <div class="card green">
      <h3>چگونه ارث‌بری کنیم؟</h3>
      <p>کافیست در زمان تعریف کلاس جدید، نام کلاس والد را درون پرانتز بنویسید: <code>class Child(Parent):</code></p>
    </div>

```python
class Vehicle:  # این کلاس والد (پایه) است
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.make} {self.model}")
```

```python
# ارث‌بری از Vehicle
class Car(Vehicle):  # کلاس فرزند که از Vehicle ارث‌بری می‌کند
    pass
```

```python
# ساخت نمونه‌ای از کلاس Car
my_car = Car("Toyota", "Corolla", 2021)
```

```python
my_car.display_info()  # خروجی: 2021 Toyota Corolla
```

      <h3>افزودن و بازنویسی (Overriding) متدها</h3>
      <p>زیرکلاس‌ها (Subclasses) می‌توانند متدهای اختصاصی خود را تعریف کنند یا متدهایی را که از والد به ارث برده‌اند برای تغییر رفتار، بازنویسی (Override) نمایند.</p>

```python
class Car(Vehicle):
    def display_car_info(self):  # متد جدید مخصوص کلاس Car
        print(f"{self.year} {self.make}")

    # بازنویسی متد ارث‌برده شده
    def display_info(self):
        print(f"This is a {self.year} {self.make} {self.model}.")
```

```python
# ساخت نمونه و استفاده از متدهای آن
my_car = Car("Honda", "Civic", 2020)
```

```python
my_car.display_car_info()  # خروجی: Honda Civic 2020 با ظرفیت صندوق: 450 لیتر
```

```python
my_car.display_info()  # خروجی: این یک ماشین Honda Civic مدل 2020 است.
```

<div class="section-card" id="toc2_">
  <h2>۲. تابع <code>super()</code> و بازنویسی متدها</h2>
  <div class="container">
    <div class="card purple">
      <h3>تابع <code>super()</code> چیست؟</h3>
      <p>تابع <code>super()</code> به شما اجازه می‌دهد تا به متدهای کلاس والد دسترسی پیدا کنید. این تابع زمانی بسیار حیاتی است که می‌خواهید قابلیتی را توسعه (Extend) دهید، به جای اینکه آن را کاملاً با کد جدید جایگزین کنید.</p>
    </div>
    <div class="card purple">
      <h3>بازنویسی متد (Method Overriding)</h3>
      <p>زمانی رخ می‌دهد که یک زیرکلاس (فرزند) پیاده‌سازی خاص و جدیدی برای متدی ارائه می‌دهد که قبلاً در کلاس والد آن تعریف شده است.</p>
    </div>
  </div>
</div>

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

class Car(Vehicle):
    def __init__(self, make, model, trunk_capacity):
        super().__init__(make, model)  # استفاده مجدد از منطق کلاس والد
        self.trunk_capacity = trunk_capacity
```

```python
class Vehicle:
    def display_info(self):
        print("This is a Vehicle")

class Car(Vehicle):
    def display_info(self):
        super().display_info()  # صدا زدن متد از کلاس Vehicle
        print("and more specifically, this is a Car")
```
