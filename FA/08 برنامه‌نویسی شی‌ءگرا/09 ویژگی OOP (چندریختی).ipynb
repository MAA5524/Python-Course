<div class="section-card">
  <h2>چندریختی (Polymorphism) چیست؟</h2>
  <div class="container">
    <div class="card blue">
      <h3>مفهوم</h3>
      <p>کلمه Polymorphism از ریشه یونانی به معنای «چندشکلی» می‌آید. این قابلیت اجازه می‌دهد که اشیاء کلاس‌های مختلف، به عنوان اشیاء یک سوپرکلاس (کلاس والد) مشترک در نظر گرفته شوند.</p>
    </div>
    <div class="card purple">
      <h3>دو جنبه کلیدی</h3>
      <ul>
        <li><strong>بازنویسی متد (Method Overriding):</strong> زیرکلاس یک پیاده‌سازی اختصاصی برای متدی که در کلاس والد تعریف شده، ارائه می‌دهد.</li>
        <li><strong>داک تایپینگ (Duck Typing):</strong> «اگر مثل اردک راه می‌رود و مثل اردک صدا درمی‌آورد، پس اردک است.» پایتون به <em>رفتار</em> شی‌ء توجه می‌کند، نه نوع (Type) آن.</li>
      </ul>
    </div>
    <div class="card blue">
      <h3>مثال: وسایل نقلیه</h3>
      <p>تصور کنید یک کلاس پایه <code>Vehicle</code> با متد <code>display_info()</code> داریم. زیرکلاس‌های <code>Car</code> و <code>Motorcycle</code> می‌توانند این متد را به روش‌های متفاوتی پیاده‌سازی کنند.</p>
    </div>
  </div>
</div>

```python
class Vehicle:
    def display_info(self):
        print("این یک وسیله نقلیه است")

class Car(Vehicle):
    def display_info(self):
        print("این یک ماشین است")

class Motorcycle(Vehicle):
    def display_info(self):
        print("این یک موتور سیکلت است")
```

```python
# تابعی که هر وسیله نقلیه‌ای را می‌پذیرد و متد display_info آن را صدا می‌زند
def vehicle_info(vehicle):
    vehicle.display_info()
```

```python
# ساخت نمونه‌ها
car = Car()
motorcycle = Motorcycle()
```

```python
# نمایش عملکرد چندریختی
vehicle_info(car)
```

```python
vehicle_info(motorcycle)
```

  <h2>۱. پیاده‌سازی چندریختی از طریق ارث‌بری</h2>
  <div class="container">
    <div class="card green">
      <h3>بازنویسی متد (Method Overriding)</h3>
      <p>زیرکلاس‌ها می‌توانند پیاده‌سازی اختصاصی خودشان را برای متدی که در کلاس والد تعریف شده است، ارائه دهند.</p>
    </div>
    <div class="card green">
      <h3>مثال: صدای حیوانات</h3>
      <p>یک کلاس پایه <code>Animal</code> با یک متد عمومی <code>speak()</code> که توسط کلاس‌های <code>Dog</code> و <code>Cat</code> بازنویسی می‌شود.</p>
    </div>
  </div>
</div>

```python
class Animal:
    def speak(self):
        return "این حیوان صدایی ندارد"

class Dog(Animal):
    def speak(self):
        return "هاپ!"

class Cat(Animal):
    def speak(self):
        return "میو"
```

```python
# ساخت نمونه‌ها
dog = Dog()
cat = Cat()
```

```python
# استفاده از چندریختی
dog.speak()
```

```python
cat.speak()
```

<div class="section-card" id="toc2_">
  <h2>۲. تایپ پویا (Dynamic Typing) و رفتار چندریختی</h2>
  <div class="container">
    <div class="card orange">
      <h3>تایپ پویا</h3>
      <p>پایتون نیازی به تعیین صریح نوع متغیر ندارد. یک متغیر می‌تواند اکنون یک عدد صحیح و بعداً یک رشته باشد. این ویژگی چندریختی انعطاف‌پذیری را ممکن می‌سازد.</p>
    </div>
    <div class="card orange">
      <h3>مثال</h3>
      <p>یک تابع ساده <code>fetch_item</code> می‌تواند روی لیست‌ها، دیکشنری‌ها و رشته‌ها کار کند، چون همگی از مفهوم دسترسی با ایندکس (Index) پشتیبانی می‌کنند.</p>
    </div>
  </div>
</div>

```python
def fetch_item(collection, position):
    return collection[position]
```

```python
# به لطف تایپ پویا، این تابع می‌تواند روی انواع داده مختلف کار کند
fetch_item([1, 2, 3], 1)
```

```python
fetch_item({'a': 1, 'b': 2}, 'b')
```

```python
fetch_item("Python", 3)
```

```python
# کلاس والد
class Document:
    def display(self):
        print("نمایش یک سند.")

# کلاس‌های فرزند
class PDF(Document):
    def display(self):
        print("نمایش سند PDF.")

class Word(Document):
    def display(self):
        print("نمایش سند Word.")

class Spreadsheet(Document):
    def display(self):
        print("نمایش سند صفحه گسترده (Spreadsheet).")

# تسک ۲: نمایش چندریختی
def display_document(document):
    document.display()

# ساخت نمونه‌هایی از انواع اسناد
pdf = PDF()
word = Word()
spreadsheet = Spreadsheet()

# ذخیره اسناد در یک لیست
documents = [pdf, word, spreadsheet]

# حلقه زدن روی لیست و نمایش هر سند
for document in documents:
    display_document(document)

# تسک ۳: بررسی تایپ پویا
def change_document_type(document, new_type):
    # ساخت یک نمونه جدید از نوع مشخص شده
    return new_type()

# در ابتدا، یک سند Word می‌سازیم
current_document = Word()
current_document.display()  # باید متن "نمایش سند Word." را چاپ کند

# تغییر نوع سند به PDF
current_document = change_document_type(current_document, PDF)
current_document.display()  # باید متن "نمایش سند PDF." را چاپ کند
```
