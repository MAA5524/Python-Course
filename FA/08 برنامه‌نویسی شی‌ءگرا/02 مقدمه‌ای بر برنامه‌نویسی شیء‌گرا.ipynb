<div class="section-card" id="toc_intro">
  <h2>۱. شی‌ءگرایی چیست؟ (در برابر رویه‌ای)</h2>
  <div class="container">
    <div class="card orange">
      <h3>الگو برنامه‌نویسی شی‌ءگرا (OOP)</h3>
      <p>برنامه‌نویسی شی‌ءگرا (Object-Oriented Programming) یک الگو برنامه‌نویسی است که بر اساس مفهوم <strong>«شی‌ء» (Object)</strong> بنا شده است. اشیاء می‌توانند شامل داده‌ها (به عنوان ویژگی‌ها/Attributes) و کدها (به عنوان متدها/Methods) باشند.</p>
      <p>هدف اصلی OOP، مدل‌سازی موجودیت‌های دنیای واقعی (مانند ارث‌بری و چندریختی) در داخل کد است. در این الگو، تمرکز ما روی «اشیائی» است که می‌خواهیم دستکاری کنیم، نه صرفاً روی «منطقی» که قرار است این کار را انجام دهد.</p>
      <p><em>تاریخچه:</em> ریشه‌های OOP به دهه ۱۹۶۰ (زبان Simula) و ۱۹۷۰ (زبان Smalltalk) برمی‌گردد. امروزه این الگو در زبان‌های قدرتمندی مثل پایتون، جاوا و C++ کاملاً مسلط است.</p>
    </div>
    <div class="card orange">
      <h3>شی‌ءگرایی در برابر برنامه‌نویسی رویه‌ای (Procedural)</h3>
      <p><strong>برنامه‌نویسی رویه‌ای (Procedural)</strong> بر روی فراخوانی رویه‌ها یا توابع (Functions) تمرکز دارد و معمولاً داده‌ها را از توابع جدا نگه می‌دارد.</p>
      <ul>
        <li><strong>رویکرد:</strong> شی‌ءگرایی، داده‌ها و متدها را با هم در یک بسته (کلاس) قرار می‌دهد. رویه‌ای، آن‌ها را به صورت دنباله‌ای از دستورات مجزا می‌بیند.</li>
        <li><strong>سازمان‌دهی:</strong> در OOP کدها بر اساس اشیاء سازمان‌دهی می‌شوند، اما در روش رویه‌ای کدها بر اساس توابع و منطق سازمان‌دهی می‌شوند.</li>
        <li><strong>استفاده مجدد:</strong> OOP از طریق ارث‌بری، استفاده مجدد از کد را بسیار راحت می‌کند. کدهای رویه‌ای برای نگهداری سخت‌تر هستند زیرا تغییر در یک بخش می‌تواند کل برنامه را تحت تاثیر قرار دهد.</li>
      </ul>
    </div>
    <div class="card orange">
      <h3>برتری‌های OOP نسبت به روش رویه‌ای</h3>
      <ul>
        <li><strong>ماژولار بودن:</strong> چون همه چیز یک شیء است، درک و نگهداری کد بسیار آسان‌تر می‌شود.</li>
        <li><strong>جلوگیری از تکرار:</strong> مکانیزم‌هایی مثل ارث‌بری باعث می‌شوند کدهای تکراری از بین بروند.</li>
        <li><strong>مقیاس‌پذیری:</strong> بهترین انتخاب برای پروژه‌های بزرگ و نرم‌افزارهای پیچیده است.</li>
        <li><strong>امنیت:</strong> با کپسوله‌سازی، داده‌های حساس پنهان می‌شوند و دسترسی به آن‌ها محدود می‌گردد.</li>
      </ul>
    </div>

<div class="section-card" id="toc1_">
  <h2>۲. مفاهیم اصلی شی‌ءگرایی (Core Concepts)</h2>
  <div class="container">
    <div class="card purple" id="toc1_1_">
      <h3>کلاس‌ها و اشیاء (Classes and Objects)</h3>
      <ul>
        <li><strong>کلاس (Class):</strong> یک الگو، نقشه یا Blueprint برای ساخت اشیاء است. کلاس تعریف می‌کند که شیء چه داده‌ها و رفتارهایی داشته باشد.</li>
        <li><strong>شی‌ء (Object):</strong> یک نمونه (Instance) ساخته شده از روی کلاس است. شیء تجسم واقعیِ داده‌ها و متدهای کلاس در حافظه است.</li>
      </ul>
      <p><strong>مثال:</strong> یک کلاس به نام <code>Dog</code> (سگ) ویژگی‌هایی مثل (سن، نژاد) و رفتارهایی مثل (پارس کردن) را تعریف می‌کند. حالا «سگ همسایه شما که ۵ سال دارد و نژادش لابرادور است»، یک <strong>شی‌ء</strong> از این کلاس است.</p>
      <p><strong>استعاره در دنیای واقعی:</strong> کلاس مانند یک «قالبِ کلوچه‌زنی» است و اشیاء، همان «کلوچه‌هایی» هستند که با آن قالب ساخته می‌شوند.</p>
      <div style="display: flex; gap: 10px; justify-content: center; flex-wrap: wrap;">
          <img src="../../images/class-objects.png" alt="کلاس و شیء">
      </div>
    </div>
    <div class="card purple" id="toc1_2_">
      <h3>ویژگی‌ها و متدها (Attributes and Methods)</h3>
      <ul>
        <li><strong>ویژگی‌ها (Attributes/Properties):</strong> خصوصیاتی هستند که وضعیت یا داده‌های یک شیء را ذخیره می‌کنند (همان متغیرهای درون کلاس).</li>
        <li><strong>متدها (Methods):</strong> رفتارها یا کارهایی هستند که یک شیء می‌تواند انجام دهد (همان توابع درون کلاس).</li>
      </ul>
      <p><strong>مثال:</strong> در کلاس ماشین (<code>Car</code>)، ویژگی‌ها می‌توانند (رنگ، مدل) باشند و متدها می‌توانند (رانندگی کردن، بوق زدن) باشند.</p>
      <img src="../../images/object.png" alt="ویژگی‌ها و متدها">
    </div>
    <div class="card purple" id="toc1_3_">
      <h3>کپسوله‌سازی (Encapsulation)</h3>
      <p>کپسوله‌سازی یعنی بسته‌بندی کردن داده‌ها (ویژگی‌ها) و کدها (متدها) در یک واحد مشخص (به نام کلاس) و محدود کردن دسترسی مستقیمِ دنیای بیرون به برخی از اجزای داخلی آن.</p>
      <ul>
        <li><strong>مثال:</strong> یک کلاس حساب بانکی (<code>BankAccount</code>) موجودی حساب را مخفی می‌کند (Private). شما برای تغییر موجودی نمی‌توانید مستقیماً دستکاری‌اش کنید، بلکه باید از متدهای ایمنی مثل <code>()deposit</code> (واریز) یا <code>()withdraw</code> (برداشت) استفاده کنید.</li>
        <li><strong>اهمیت:</strong> امنیت داده‌ها را تضمین می‌کند و از تغییرات ناخواسته جلوگیری می‌کند.</li>
      </ul>
    </div>
    <div class="card purple" id="toc1_4_">
      <h3>ارث‌بری (Inheritance)</h3>
      <p>ارث‌بری به یک کلاس جدید (فرزند/Child) اجازه می‌دهد تا تمام ویژگی‌ها و متدهای یک کلاس از قبل موجود (والد/Parent) را به ارث ببرد. این کار باعث می‌شود کدهای تکراری به شدت کاهش یابد.</p>
      <ul>
        <li><strong>ارث‌بری یگانه (Single):</strong> یک فرزند از یک والد ارث می‌برد.</li>
        <li><strong>ارث‌بری چندگانه (Multiple):</strong> یک فرزند همزمان از چند والد ارث می‌برد (پایتون از این حالت پشتیبانی می‌کند).</li>
        <li><strong>ارث‌بری چندسطحی (Multilevel):</strong> یک کلاس از کلاسی ارث می‌برد که خودش فرزند کلاس دیگری است (پدربزرگ -> پدر -> فرزند).</li>
      </ul>
    </div>
    <div class="card purple" id="toc1_5_">
      <h3>چندریختی (Polymorphism)</h3>
      <p>چندریختی به معنای «داشتن اَشکال مختلف» است. این مفهوم اجازه می‌دهد که اشیائی از کلاس‌های مختلف، بتوانند از طریق یک رابط (نام متد) یکسان فراخوانی شوند، اما هر کدام رفتار مخصوص به خودشان را نشان دهند.</p>
      <ul>
        <li><strong>تغییر رفتار (Overriding):</strong> کلاس فرزند می‌تواند متدی را که از والد به ارث برده، مطابق با نیاز خودش بازنویسی کند.</li>
        <li><strong>مثال:</strong> فرض کنید یک متد به نام <code>()draw</code> (رسم کردن) داریم. اگر آن را روی کلاس <code  style='font-family: "Vazirmatn"'>دایره</code> صدا بزنیم، یک دایره می‌کشد و اگر روی کلاس <code style='font-family: "Vazirmatn"'>مربع</code> صدا بزنیم، یک مربع رسم می‌کند. نام متد یکی است، اما خروجی متفاوت است.</li>
      </ul>
    </div>
    <div class="card purple" id="toc1_6_">
      <h3>انتزاع یا تجرید (Abstraction)</h3>
      <p>انتزاع یعنی پنهان کردن جزئیات پیچیده پیاده‌سازی و نمایش دادن تنها بخش‌های ضروری به کاربر. تمرکز انتزاع روی این است که یک شیء <em>«چه کاری انجام می‌دهد»</em>، نه اینکه <em>«چگونه آن را انجام می‌دهد»</em>.</p>
      <ul>
        <li><strong>تفاوت با کپسوله‌سازی:</strong> انتزاع پیچیدگی‌ها را در سطح طراحی مخفی می‌کند، اما کپسوله‌سازی وضعیت (داده‌های) داخلی را در سطح اجرا مخفی می‌کند.</li>
        <li><strong>پیاده‌سازی:</strong> معمولاً از طریق کلاس‌های انتزاعی (Abstract Classes) انجام می‌شود؛ کلاس‌هایی که نمی‌توان مستقیماً از روی آن‌ها شیء ساخت و فقط نقش راهنما را دارند.</li>
      </ul>
    </div>
  </div>
</div>

  <h2>۳. مزایا و معایب شی‌ءگرایی</h2>
  <div class="container">
    <div class="card green">
      <h3>✅ مزایا (Advantages)</h3>
      <ul>
        <li><strong>عیب‌یابی آسان:</strong> ماژولار بودن اجازه می‌دهد وقتی باگی رخ داد، دقیقاً روی همان شیء یا کلاس تمرکز کنید.</li>
        <li><strong>استفاده مجدد:</strong> به کمک ارث‌بری (Inheritance) از نوشتن کدهای تکراری جلوگیری می‌شود.</li>
        <li><strong>انعطاف‌پذیری:</strong> به لطف چندریختی (Polymorphism) سیستم به راحتی قابل توسعه و تغییر است.</li>
        <li><strong>امنیت:</strong> کپسوله‌سازی جلوی دسترسی تصادفی و خراب شدن داده‌ها را می‌گیرد.</li>
      </ul>
    </div>
    <div class="card red">
      <h3>❌ معایب (Disadvantages)</h3>
      <ul>
        <li><strong>منحنی یادگیری تند:</strong> درک اولیه آن نسبت به کدهای خطی و رویه‌ای سخت‌تر است.</li>
        <li><strong>اجرای کُندتر:</strong> ساخت اشیاء و مدیریت حافظه آن‌ها نیازمند پردازش بیشتری است.</li>
        <li><strong>حجم بیشتر کد:</strong> معمولاً نیازمند نوشتن خطوط کد و معماری بیشتری است.</li>
        <li><strong>اضافی بودن در کارهای کوچک:</strong> برای نوشتن یک اسکریپت ساده ریاضی، استفاده از کلاس کاملاً اضافی (Overkill) است.</li>
      </ul>
    </div>
    <div class="card blue">
      <h3>چه زمانی از شی‌ءگرایی استفاده کنیم؟</h3>

            <li>برنامه‌های دارای رابط کاربری گرافیکی (GUI).</li>
            <li>پروژه‌های تیمی که نیاز به توسعه و آپدیت مداوم دارند.</li>

          </ul>

<div class="section-card" id="toc3_">
  <h2>۴. جمع‌بندی و قدم بعدی</h2>
  <div class="container">
    <div class="card pink">
      <h3>ذهنیت شی‌ءگرا (The OOP Mindset)</h3>
      <p>همه چیز در دنیای اطراف ما (مثل گوشی موبایل، ماشین، یا دستگاه قهوه‌ساز) را می‌توان در برنامه‌نویسی مدل‌سازی کرد. هر کدام از این‌ها دارای <strong>ویژگی‌هایی</strong> (مثل رنگ و سرعت) و <strong>رفتارهایی</strong> (مثل تماس گرفتن یا قهوه ساختن) هستند. فکر کردن به این روش باعث می‌شود برنامه‌نویسی برای انسان ملموس‌تر شود.</p>
    </div>
    <div class="card pink">
      <h3>قدم بعدی چیست؟</h3>
      <p>در درس بعدی، ما این مفاهیم تئوری را وارد دنیای واقعی <strong>کدهای پایتون</strong> می‌کنیم. با هم یاد می‌گیریم که پایتون چطور با سینتکس تمیز و ساده‌ی خود، کلاس‌ها، ارث‌بری و کپسوله‌سازی را در قالب تمرین‌های عملی پیاده‌سازی می‌کند.</p>
    </div>
