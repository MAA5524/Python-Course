```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} میگه هاپ هاپ!"
```

```python
# فراخوانی متد نمونه
my_dog.bark()
```

  <h2>۳. متدهای کلاس (Class Methods)</h2>
  <div class="container">
    <div class="card purple">
      <h3>مدیریت وضعیت کلاس</h3>
      <p>متدهای کلاس پارامتر <code>cls</code> را به عنوان اولین ورودی می‌گیرند و با <code>@classmethod</code> مشخص می‌شوند. آن‌ها می‌توانند متغیرهای کلاسی را تغییر دهند که روی <strong>تمام</strong> نمونه‌ها تأثیر می‌گذارد.</p>
      <p><strong>کاربرد رایج:</strong> ساخت متدهای کارخانه‌ای (سازنده‌های جایگزین) و پیاده‌سازی الگوی Singleton.</p>
    </div>
  </div>
</div>

```python
class Singleton:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            print("در حال ایجاد تنها نمونه‌ی موجود (Singleton)...")
            cls._instance = super(Singleton, cls).__new__(cls)
        return cls._instance

    @classmethod
    def get_instance(cls):
        if not cls._instance:
            cls()
        return cls._instance
```

```python
obj1 = Singleton()
```

```python
obj2 = Singleton()
```

```python
obj1 is obj2
```

    <div class="card purple">
      <h3>مثال: تغییر وضعیت کلاس</h3>
      <p>بیایید از یک متد کلاس استفاده کنیم تا نرخ تخفیف را برای <strong>تمام</strong> آیتم‌های یک انبار تغییر دهیم.</p>

```python
class Inventory:
    discount_rate = 10  # درصد

    def __init__(self, product_name, product_price):
        self.product_name = product_name
        self.product_price = product_price

    @classmethod
    def set_discount_rate(cls, new_rate):
        cls.discount_rate = new_rate
```

```python
# نرخ تخفیف اولیه
print("نرخ تخفیف اولیه:", Inventory.discount_rate)
```

```python
# تغییر نرخ تخفیف با متد کلاس
Inventory.set_discount_rate(15)
print("نرخ تخفیف جدید:", Inventory.discount_rate)
```

<div class="section-card" id="toc4_">
  <h2>۴. مقایسه و جمع‌بندی</h2>
  <div class="container">
    <div class="card green">
      <h3>جدول خلاصه</h3>
      <p><strong>متد نمونه:</strong> از <code>self</code> استفاده می‌کند. وضعیت شیء را تغییر می‌دهد. (حالت پیش‌فرض)</p>
      <p><strong>متد استاتیک:</strong> بدون <code>self</code> یا <code>cls</code>. منطق مستقل دارد. (توابع کمکی یا Helper)</p>
      <p><strong>متد کلاس:</strong> از <code>cls</code> استفاده می‌کند. وضعیت کلاس را تغییر می‌دهد. (سازنده‌های جایگزین)</p>
    </div>
    <div class="card green">
      <h3>مثال‌های دنیای واقعی</h3>
      <ul>
        <li><strong>حساب بانکی (نمونه):</strong> <code>deposit(self, amount)</code> - موجودی یک نفر را آپدیت می‌کند.</li>
        <li><strong>کمکی ریاضی (استاتیک):</strong> <code>area(radius)</code> - صرفاً یک محاسبه ریاضی انجام می‌دهد.</li>
        <li><strong>کارمند (کلاس):</strong> <code>from_string(cls, str)</code> - از روی یک متن ورودی، شیء کارمند می‌سازد.</li>
      </ul>
    </div>
  </div>
</div>

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
```

```python
class MathHelper:
    @staticmethod
    def area_of_circle(radius):
        return 3.14 * radius * radius
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def from_string(cls, employee_str):
        name, salary = employee_str.split('-')
        return cls(name, int(salary))
```

```python
emp = Employee.from_string("Ali-5000")
```

```python
emp.name, emp.salary
```
