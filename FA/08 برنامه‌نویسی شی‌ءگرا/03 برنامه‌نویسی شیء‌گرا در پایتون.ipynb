      <h3>شی‌ءگرایی (OOP) چیست؟</h3>
      <p>برنامه‌نویسی شی‌ءگرا (Object-Oriented Programming) یکی از قدرتمندترین رویکردها در توسعه نرم‌افزار است که حول محور استفاده از کلاس‌ها و اشیاء (Objects) می‌چرخد. در این روش، داده‌ها و رفتارها در قالب واحدهای عملیاتیِ منسجم گروه‌بندی می‌شوند.</p>
      <p><strong>پایتون، بهترین بستر برای یادگیری:</strong> سینتکس ساده و خوانایی بالای پایتون، آن را به بستری ایده‌آل برای یادگیری مفاهیم شی‌ءگرایی (بدون گیج شدن در پیچیدگی‌های کدنویسی) تبدیل کرده است.</p>
    </div>
    <div class="card blue">
      <h3>مزایای یادگیری شی‌ءگرایی در پایتون</h3>
      <ul>
        <li><strong>بهبود مهارت حل مسئله:</strong> مسائل پیچیده را به واحدهای کوچک‌تر و قابل مدیریت (کلاس‌ها و اشیاء) می‌شکند.</li>
        <li><strong>استفاده مجدد از کد:</strong> با نوشتن کدهای ماژولار، از تکرار جلوگیری شده و در زمان صرفه‌جویی می‌شود.</li>
        <li><strong>همکاری تیمی بهتر:</strong> این رویکرد استاندارد باعث می‌شود کدهای شما برای دیگران خواناتر و دیباگ آن‌ها راحت‌تر باشد.</li>
        <li><strong>فرصت‌های شغلی متنوع:</strong> یادگیری این مفاهیم برای توسعه وب، علم داده و مهندسی نرم‌افزار کاملاً ضروری است.</li>
      </ul>

<div class="section-card" id="toc1_">
  <h2>۱. مفاهیم پایه شی‌ءگرایی در پایتون</h2>
  <div class="container">
    <div class="card purple">
      <h3>کلاس‌ها و اشیاء (Classes & Objects)</h3>
      <ul>
        <li><strong>کلاس (Class):</strong> یک نقشه، الگو یا Blueprint است. کلاس داده‌ها (ویژگی‌ها) و متدهایی را که با آن داده‌ها کار می‌کنند در خود کپسوله (بسته‌بندی) می‌کند.</li>
        <li><strong>شی‌ء (Object):</strong> یک نمونه (Instance) ساخته شده از روی یک کلاس است. شیء در واقع پیاده‌سازیِ واقعیِ داده‌ها و عملکردهای تعریف‌شده در آن کلاس است.</li>
      </ul>
      <img src="../../images/class-objects.png" alt="کلاس‌ها و اشیاء">
    </div>
    <div class="card purple">
      <h3>مثال ساخت یک کلاس ساده</h3>
      <p>بیایید یک کلاس ساده به نام <code>Dog</code> (سگ) بسازیم که دو ویژگی (<code>name</code>، <code>age</code>) و یک متد برای پارس کردن (<code>bark</code>) دارد.</p>
    </div>
  </div>
</div>

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} says Woof!"
```

<div class="section-card" id="toc2_">
  <h2>۲. ویژگی‌های کلیدی شی‌ءگرایی (چهار ستون اصلی)</h2>
  <div class="container">
    <div class="card orange">
      <h3>۱. کپسوله‌سازی (Encapsulation)</h3>
      <p><strong>مفهوم:</strong> پنهان کردن جزئیات داخلی و نمایش دادن تنها بخش‌های ضروری. کپسوله‌سازی داده‌ها و متدها را در یک واحدِ بسته نگه می‌دارد و با محدود کردن دسترسی مستقیم به داده‌ها، از تغییرات ناخواسته و تصادفی جلوگیری می‌کند.</p>
      <img src="../../images/oop-encapsulation.png" alt="کپسوله سازی">
      <p><strong>مثال:</strong> یک حساب بانکی که به جای تغییر مستقیم موجودی، باید از متدهای ایمنی مثل واریز (Deposit) استفاده کنید.</p>
    </div>
    <div class="card orange">
      <p>در مثال زیر، <code>__balance</code> یک ویژگی خصوصی (Private) است (که با دو خط تیره مشخص شده) و در داخل کلاس <code>Account</code> کپسوله شده است. در نتیجه این متغیر از بیرون کلاس به صورت مستقیم قابل دسترسی و دستکاری نیست.</p>
    </div>
  </div>
</div>

```python
class Account:
    def __init__(self, owner, amount=0):
        self.owner = owner
        self.__balance = amount  # ویژگی خصوصی (Private)

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("واریز با موفقیت انجام شد.")

    def show_balance(self):
        print(f"موجودی حساب: {self.__balance}")
```

```python
acc = Account("John")
```

```python
acc.deposit(100)
```

```python
acc.show_balance()  # خروجی: موجودی حساب: 100
```

      <h3>۲. ارث‌بری (Inheritance)</h3>
      <p><strong>مفهوم:</strong> به یک کلاس فرزند (Child) اجازه می‌دهد تا تمام ویژگی‌ها و متدها را از یک کلاس والد (Parent) به ارث ببرد. این کار باعث استفاده مجدد از کدها شده و یک رابطه سلسله‌مراتبی منطقی ایجاد می‌کند.</p>

```python
class Animal:
    def speak(self):
        print("Animal speaks")
```

```python
class Dog(Animal):
    def speak(self):
        print("Dog barks")
```

```python
# ساخت نمونه‌ای از سگ
dog = Dog()
```

```python
dog.speak()  # خروجی: Dog barks
```

<div class="section-card">
  <div class="container">
    <div class="card blue">
      <h3>۳. چندریختی (Polymorphism)</h3>
      <p><strong>مفهوم:</strong> کلمه Polymorphism به معنای «داشتن اَشکال مختلف» است. این قابلیت اجازه می‌دهد که با اشیائی از کلاس‌های مختلف، مشابه با اشیاء یک کلاس والد مشترک رفتار شود. در نتیجه، یک متد یکسان می‌تواند بر اساس نوع شیء، رفتارهای متفاوتی نشان دهد.</p>
      <img src="../../images/oop-polymorphism.png" alt="چندریختی">
      <p><strong>مثال:</strong> یک دستور <code>()fly</code> (پرواز کردن) روی یک طوطی (Parrot) و یک پنگوئن (Penguin) رفتارهای کاملاً متفاوتی تولید می‌کند.</p>
    </div>
    <div class="card blue">
      <p>در مثال زیر، متد <code>fly</code> از کلاس <code>Bird</code> در هر دو کلاس فرزند بازنویسی (Override) شده است. وقتی تابع <code>test_flying</code> دستور <code>()bird.fly</code> را صدا می‌زند، پایتون به صورت خودکار تشخیص می‌دهد که بر اساس نوع شیء پاس داده شده، کدام نسخه را اجرا کند.</p>
    </div>
  </div>
</div>

```python
class Bird:
    def fly(self):
        print("Some birds can fly")

class Parrot(Bird):
    def fly(self):
        print("Parrots can fly")

class Penguin(Bird):
    def fly(self):
        print("Penguins do not fly")
```

```python
# کاربرد چندریختی (Polymorphism) در عمل
def test_flying(bird):
    bird.fly()
```

```python
test_flying(Parrot())  # خروجی: Parrots can fly
```

```python
test_flying(Penguin())  # خروجی: Penguins do not fly
```

      <h3>۴. انتزاع (Abstraction)</h3>
      <p><strong>مفهوم:</strong> پنهان کردن پیچیدگی‌های پیاده‌سازی و نمایش دادن تنها بخش‌های ضروری. در پایتون این کار معمولاً با استفاده از کلاس‌های پایه انتزاعی (Abstract Base Classes) توسط ماژول <code>abc</code> انجام می‌شود.</p>
      <p>انتزاع یک نوع «قرارداد» ایجاد می‌کند: کلاس‌های فرزند <em>مجبورند</em> متدهای خاصی (مثلاً محاسبه <code>area</code> و <code>perimeter</code> در کلاس <code>Shape</code>) را پیاده‌سازی کنند.</p>
    </div>
  </div>
</div>

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass
```

```python
class Square(Shape):
    def __init__(self, side):
        self.__side = side

    def area(self):
        return self.__side * self.__side

    def perimeter(self):
        return 4 * self.__side
```

```python
square = Square(5)
square.area()
```
