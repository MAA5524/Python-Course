```python
# تسک ۱: تعریف کلاس Classroom
class Classroom:
    def __init__(self, teacher, students, subject, room_number):
        self.teacher = teacher
        self.students = students
        self.subject = subject
        self.room_number = room_number

    def display_info(self):
        print(f"Teacher: {self.teacher}")
        print(f"Students: {self.students}")
        print(f"Subject: {self.subject}")
        print(f"Room Number: {self.room_number}")
```

```python
# تسک ۲: نمونه‌سازی از کلاس
my_class = Classroom(
    teacher="Mr. Smith",
    students=["Alice", "Bob", "Charlie", "Diana"],
    subject="Mathematics",
    room_number=101
)
```

```python
# تسک ۳: چاپ اطلاعات شی‌ء
my_class.display_info()
```

```python
# بخش پاداش: دریافت لیست پویای دانش‌آموزان
class ClassroomWithDynamicStudents:
    def __init__(self, teacher, *students, subject, room_number):
        self.teacher = teacher
        self.students = list(students)  # تبدیل تاپل ورودی به لیست
        self.subject = subject
        self.room_number = room_number

    def display_info(self):
        print(f"Teacher: {self.teacher}")
        print(f"Students: {self.students}")
        print(f"Subject: {self.subject}")
        print(f"Room Number: {self.room_number}")
```

```python
# تست بخش پاداش
bonus_class = ClassroomWithDynamicStudents(
    "Ms. Johnson",
    "Eva", "Frank", "Grace", "Henry",
    subject="Science",
    room_number=102
)
```

```python
bonus_class.display_info()
```
