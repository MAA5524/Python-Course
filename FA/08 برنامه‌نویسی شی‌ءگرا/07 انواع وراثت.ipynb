```python
class Student(Person):
    def __init__(self, name, age, ID, studentID):
        super().__init__(name, age, ID)
        self.studentID = studentID
        self.courses = []

    def enroll_course(self, course_name):
        self.courses.append(course_name)
        print(f"{self.name} has enrolled in {course_name}")

    def display_courses(self):
        print(f"{self.name}'s Courses: {', '.join(self.courses)}")
```

  <h2>۲. ارث‌بری چندسطحی: مزایا و چالش‌ها</h2>
  <div class="container">
    <div class="card orange">
      <h3>مفهوم</h3>
      <p>ایجاد یک زنجیره از ارث‌بری: پدربزرگ -> پدر -> فرزند. هر سطح، سطح قبلی خود را توسعه می‌دهد.</p>
      <img src="../../images/multi-level-inheritance.png" alt="ارث‌بری چندسطحی">
    </div>
    <div class="card orange">
      <h3>مثال: سلسله‌مراتب کارمندان</h3>
      <p>یک ساختار ۳ سطحی می‌سازیم: <code>Person</code> (شخص) -> <code>Employee</code> (کارمند) -> <code>Manager</code> (مدیر).</p>
    </div>
  </div>
</div>

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")
```

```python
class Employee(Person):
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)
        self.employee_id = employee_id

    def display_employee_info(self):
        print(f"Employee ID: {self.employee_id}")
        self.display_info()
```

```python
class Manager(Employee):
    def __init__(self, name, age, employee_id, department):
        super().__init__(name, age, employee_id)
        self.department = department

    def display_manager_info(self):
        print(f"Department: {self.department}")
        self.display_employee_info()
```

      <p>کلاس پایه <code>Vehicle</code> که کلاس‌های <code>Car</code>، <code>Truck</code> و <code>Motorcycle</code> از آن ارث می‌برند.</p>

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.make} {self.model}")
```

```python
class Car(Vehicle):
    def __init__(self, make, model, year, doors):
        super().__init__(make, model, year)
        self.doors = doors

class Truck(Vehicle):
    def __init__(self, make, model, year, payload):
        super().__init__(make, model, year)
        self.payload = payload

class Motorcycle(Vehicle):
    def __init__(self, make, model, year, cc):
        super().__init__(make, model, year)
        self.cc = cc
```

  <div class="container">
    <div class="card red">
      <h3>مفهوم</h3>
      <p>یک کلاس فرزند، همزمان از <strong>چندین والد</strong> ارث‌بری می‌کند. این قابلیت در پایتون بسیار قدرتمند است، اما اگر با دقت استفاده نشود باعث پیچیدگی (مثل Diamond Problem) می‌شود.</p>
      <img src="../../images/multiple-inheritance.png" alt="ارث‌بری چندگانه">
    </div>
    <div class="card red">
      <h3>مثال: حیوان ترکیبی</h3>
      <p>کلاس تمساح (<code>Alligator</code>) که ویژگی‌های خود را هم از <code>Walker</code> (راه‌رونده) و هم از <code>Swimmer</code> (شناگر) به ارث می‌برد.</p>
    </div>
  </div>
</div>

```python
class Walker:
    def walk(self):
        print("این حیوان راه می‌رود")

class Swimmer:
    def swim(self):
        print("این حیوان شنا می‌کند")

class Alligator(Walker, Swimmer):
    pass
```

```python
# ساخت شی‌ء از تمساح
ally = Alligator()
```

```python
ally.walk()
```

```python
ally.swim()
```
