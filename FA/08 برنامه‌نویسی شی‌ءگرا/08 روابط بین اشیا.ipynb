```python
class Stethoscope:
    def __init__(self, brand):
        self.brand = brand

    def use(self):
        print(f"Using {self.brand} stethoscope.")

class Doctor:
    def __init__(self, name):
        self.name = name

    def check_patient(self, stethoscope):
        print(f"{self.name} is checking the patient.")
        stethoscope.use()
```

```python
# مثال استفاده
my_stethoscope = Stethoscope("Littmann")
dr_jones = Doctor("Dr. Jones")
```

```python
dr_jones.check_patient(my_stethoscope)
```

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def play(self):
        print(f"{self.name} is playing.")

class Owner:
    def __init__(self, name):
        self.name = name
        self.dogs = []  # می‌تواند چندین شیء از کلاس سگ را نگه دارد

    def add_dog(self, dog):
        self.dogs.append(dog)

    def play_with_dogs(self):
        print(f"{self.name} is playing with their dogs.")
        for dog in self.dogs:
            dog.play()
```

```python
# مثال استفاده
buddy = Dog("Buddy")
rover = Dog("Rover")
jane = Owner("Jane")
```

```python
jane.add_dog(buddy)
jane.add_dog(rover)
```

```python
jane.play_with_dogs()
```

```python
class Player:
    def __init__(self, name):
        self.name = name

class Team:
    def __init__(self, name):
        self.name = name
        self.players = []  # اینجا رابطه تجمیع برقرار می‌شود

    def add_player(self, player):
        self.players.append(player)

    def remove_player(self, player):
        self.players.remove(player)
```

```python
# چرخه حیات اشیاء بازیکن به شیء تیم وابسته نیست
player1 = Player("Alex")
team = Team("Sharks")
team.add_player(player1)
# در اینجا، بازیکن ۱ می‌تواند عضو تیم دیگری شود یا اصلاً تیمی نداشته باشد، که این نشان‌دهنده استقلال اوست.
```

```python
class Collar:
    def __init__(self, color):
        self.color = color

class Dog:
    def __init__(self, name):
        self.name = name
        self.collar = None  # در ابتدا سگ ممکن است قلاده نداشته باشد

    def add_collar(self, collar):
        self.collar = collar

    def remove_collar(self):
        self.collar = None
```

```python
# ساخت شیء قلاده
blue_collar = Collar("blue")
```

```python
# ساخت شیء سگ و متصل کردن قلاده به آن
buddy = Dog("Buddy")
buddy.add_collar(blue_collar)
```

```python
# سگ حالا قلاده دارد (رابطه تجمیع برقرار است)
# می‌توانیم قلاده را باز کنیم، که استقلال آن را نشان می‌دهد
buddy.remove_collar()
```

```python
# قلاده هنوز در سیستم وجود دارد و می‌تواند برای سگ دیگری استفاده شود
rover = Dog("Rover")
rover.add_collar(blue_collar)
```

```python
class Processor:
    def __init__(self, name):
        self.name = name
```

```python
class Memory:
    def __init__(self, size):
        self.size = size
```

```python
class HardDrive:
    def __init__(self, capacity):
        self.capacity = capacity
```

```python
class Computer:
    def __init__(self):
        self.processor = Processor("Intel i7")
        self.memory = Memory(16)  # بر حسب گیگابایت
        self.hard_drive = HardDrive(1)  # بر حسب ترابایت

    def describe(self):
        print(f"Computer with {self.processor.name} processor, {self.memory.size}GB RAM, and {self.hard_drive.capacity}TB storage.")
```

```python
class Tail:
    def __init__(self, length):
        self.length = length  # طول بر حسب سانتی‌متر

    def wag(self):
        print("The tail is wagging.")
```

```python
class Dog:
    def __init__(self, name):
        self.name = name
        self.tail = Tail(20)  # فرض می‌کنیم هر سگ یک دُم ۲۰ سانتی‌متری دارد

    def wag_tail(self):
        print(f"{self.name} is happy!")

        self.tail.wag()
```

```python
# با ایجاد شیء سگ، دُم نیز به عنوان بخشی لاینفک از او ساخته می‌شود.
buddy = Dog("Buddy")
buddy.wag_tail()
# در اینجا، دُم بادی کاملاً به او وابسته است و مفهوم «ترکیب» (Composition) را نشان می‌دهد.
```

  <h2>۴. ارث‌بری (Inheritance)</h2>
  <div class="container">
    <div class="card purple">
      <h3>مفهوم</h3>
      <p>رابطه «نوعی از است» (Is-a). یک کلاس جدید از کلاسی که قبلاً وجود داشته مشتق می‌شود و ویژگی‌ها و رفتارهای آن را به ارث می‌برد.</p>
      <img src="../../images/inheritance.png" alt="Inheritance">
    </div>
    <div class="card purple">
      <h3>انواع ارث‌بری</h3>
      <ul>
        <li><strong>یگانه (Single):</strong> یک والد.</li>
        <li><strong>چندگانه (Multiple):</strong> چند والد مختلف.</li>
        <li><strong>چندسطحی (Multilevel):</strong> پدربزرگ -> پدر -> فرزند.</li>
        <li><strong>سلسله‌مراتبی (Hierarchical):</strong> یک والد، چندین فرزند.</li>
      </ul>
      <img src="../../images/inheritance-types.png" alt="Inheritance Types">
    </div>
    <div class="card purple">
      <h3>مثال: حیوان خانگی و سگ</h3>
      <p>یک سگ، <strong>نوعی</strong> حیوان خانگی (Pet) است.</p>
    </div>
  </div>
</div>

```python
class Pet:
    def __init__(self, name):
        self.name = name

    def show_affection(self):
        print(f"{self.name} shows affection.")

class Dog(Pet):  # کلاس سگ از حیوان خانگی ارث می‌برد
    def __init__(self, name, breed):
        super().__init__(name)  # فراخوانی سازنده کلاس والد
        self.breed = breed

    def bark(self):
        print(f"{self.name}, the {self.breed}, barks.")
```

```python
# استفاده
my_pet = Pet("Buddy")
my_pet.show_affection()
```

```python
my_dog = Dog("Max", "Golden Retriever")
```

```python
my_dog.show_affection()  # متد به ارث برده شده
```

```python
my_dog.bark()  # متد اختصاصی کلاس سگ
```

  <h2>۵. راهنمای کاربردی: کِی از کدام استفاده کنیم؟</h2>
  <div class="container">
    <div class="card blue">
      <h3>تقلب‌نامه (Cheat Sheet)</h3>
      <ul>
        <li><strong>انجمن:</strong> برای روابط «استفاده می‌کند از» (Uses-a). (مشتری از روش پرداخت استفاده می‌کند)</li>
        <li><strong>تجمیع:</strong> برای روابط «دارد» (Has-a) به‌صورت <strong>مستقل</strong>. (ایرلاین هواپیماهایی دارد)</li>
        <li><strong>ترکیب:</strong> برای روابط «مالک بودن» (Owns-a) به‌صورت <strong>وابسته</strong>. (کتاب صفحاتی دارد)</li>
        <li><strong>ارث‌بری:</strong> برای روابط «نوعی از است» (Is-a). (ماشین نوعی وسیله نقلیه است)</li>

        <li>وابستگی‌ها (Coupling) را پایین نگه دارید (اشیاء باید تا حد امکان مستقل باشند).</li>
        <li>از ایجاد زنجیره‌های ارث‌بری طولانی و عمیق خودداری کنید.</li>
