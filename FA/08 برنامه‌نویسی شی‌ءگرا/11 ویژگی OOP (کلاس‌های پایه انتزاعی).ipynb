      <p>کلاسی است که نمی‌توان از روی آن شیء (نمونه) ساخت. این کلاس قالبی از متدهای انتزاعی تعریف می‌کند که زیرکلاس‌ها (کلاس‌های فرزند) مجبورند آن‌ها را پیاده‌سازی کنند.</p>
    </div>
    <div class="card green">
      <h3>چرا از <code>abc</code> استفاده می‌کنیم؟</h3>
      <p>ماژول <code>abc</code> در پایتون، زیرساختی را برای تعریف کلاس‌های انتزاعی و اجبار کردن اجرای رابط‌ها (Interfaces) فراهم می‌کند.</p>
    </div>
  </div>
</div>

```python
from abc import ABC, abstractmethod
```

```python
class Shape(ABC):  # کلاس Shape یک کلاس انتزاعی است
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):  # کلاس Rectangle زیرکلاسی از Shape است
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)
```

```python
# حالا، به دلیل ماهیت انتزاعی Shape، نمی‌توانید نمونه‌ای از آن بسازید
# shape = Shape()  # این خط باعث بروز خطا می‌شود
```

```python
# اما می‌توانید یک نمونه از Rectangle بسازید، زیرا تمام متدهای انتزاعی Shape را پیاده‌سازی کرده است
rectangle = Rectangle(10, 20)
```

```python
rectangle.area()
```

```python
rectangle.perimeter()
```

```python
class MyAbstractClass(ABC):
    @abstractmethod
    def an_abstract_method(self):
        pass
```

```python
from abc import ABC, abstractmethod

class DocumentHandler(ABC):

    @abstractmethod
    def read(self):
        pass

    @abstractmethod
    def write(self, content):
        pass

    @abstractmethod
    def validate(self):
        pass
```

<div class="section-card" id="toc3_">
  <h2>۳. پیاده‌سازی رابط‌ها (Interfaces)</h2>
  <div class="container">
    <div class="card red">
      <h3>رابط (Interface) چیست؟</h3>
      <p>یک رابط تعیین می‌کند که یک کلاس <strong>چه</strong> کارهایی باید انجام دهد، نه اینکه <strong>چگونه</strong> آن‌ها را انجام دهد. در پایتون، ABCها دقیقاً نقش رابط را ایفا می‌کنند.</p>
    </div>
    <div class="card red">
      <h3>مثال: پردازشگر پرداخت</h3>
      <p>تعریف یک رابط عمومی <code>PaymentProcessor</code> که توسط کلاس‌های کارت اعتباری (Credit Card) و پی‌پل (PayPal) پیاده‌سازی می‌شود.</p>
    </div>
  </div>
</div>

```python
from abc import ABC, abstractmethod
```

```python
class PaymentProcessor(ABC):

    @abstractmethod
    def process_payment(self, amount):
        """
        پرداخت را برای مبلغ مشخص‌شده انجام می‌دهد.
        """
        pass

class CreditCardPaymentProcessor(PaymentProcessor):

    def process_payment(self, amount):
        print(f"در حال پردازش پرداخت کارت اعتباری به مبلغ {amount} دلار")

class PayPalPaymentProcessor(PaymentProcessor):

    def process_payment(self, amount):
        print(f"در حال پردازش پرداخت پی‌پل به مبلغ {amount} دلار")
```

```python
# نحوه استفاده
# با فرض اینکه نمونه‌سازی انجام شده است، می‌توانیم فراخوانی کنیم:
processor = CreditCardPaymentProcessor()
processor.process_payment(100)

# این کار تضمین می‌کند که هر پردازشگر پرداخت، متد مورد نیاز را پیاده‌سازی کرده است.
```
