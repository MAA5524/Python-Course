<div class="section-card" id="toc1_">
  <h2>۱. رشته‌ها: <code>__str__</code> و <code>__repr__</code></h2>
  <div class="container">
    <div class="card green">
      <h3><code>__str__</code>: برای کاربر</h3>
      <p>وقتی از دستور <code>str(obj)</code> یا <code>print(obj)</code> استفاده می‌کنید، این متد فراخوانی می‌شود. هدف آن برگرداندن یک رشته خوانا و کاربرپسند است.</p>
    </div>
  </div>
</div>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"کتاب '{self.title}' اثر {self.author}"
```

```python
# ساخت یک نمونه از کتاب
book = Book("آموزش پایتون", "جین دو")
```

```python
print(str(book))  # فراخوانی صریح متد __str__
```

```python
print(book)  # فراخوانی ضمنی متد __str__
```

      <h3><code>__repr__</code>: برای توسعه‌دهنده</h3>
      <p>هدف این متد رفع ابهام و کمک به دیباگ کردن است. در حالت ایده‌آل، این متد باید رشته‌ای برگرداند که با اجرای آن بتوان شیء را دوباره در حافظه ساخت.</p>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        # !r باعث می‌شود مقدار داخل کوتیشن قرار بگیرد
        return f"Book({self.title!r}, {self.author!r})"
```

```python
# ساخت نمونه جدید
book = Book("آموزش پایتون", "جین دو")
```

```python
repr(book)  # فراخوانی صریح متد __repr__
```

```python
book  # در محیط جوپیتر متد __repr__ به صورت خودکار فراخوانی می‌شود
```

```python
class CustomList:
    def __init__(self, items):
        self.items = items

    def __len__(self):
        return len(self.items)
```

```python
len(my_list)  # فراخوانی متد __len__
```

```python
class Inventory:
    def __init__(self):
        self.items = {}

    def add_item(self, item, quantity):
        if item in self.items:
            self.items[item] += quantity
        else:
            self.items[item] = quantity

    def remove_item(self, item, quantity):
        if item in self.items:
            if self.items[item] <= quantity:
                del self.items[item]
            else:
                self.items[item] -= quantity

    def __len__(self):
        return sum(self.items.values())
```

```python
inventory = Inventory()
inventory.add_item('سیب', 5)
inventory.add_item('موز', 3)
inventory.add_item('پرتقال', 2)
```

```python
len(inventory)  # خروجی: 10
```

```python
inventory.remove_item('موز', 2)
```

```python
len(inventory)  # خروجی: 8
```

  <h2>۳. محاسبات: <code>__add__</code> و <code>__mul__</code></h2>
  <div class="container">
    <div class="card purple">
      <h3>عملیات ریاضی در اشیاء سفارشی</h3>
      <p>پایتون اجازه می‌دهد اشیاء ساخته شده توسط شما، درست مثل اعداد صحیح یا اعشاری، در عملیات ریاضی شرکت کنند. این کار با متدهایی مثل <code>__add__ (+)</code>، <code>__sub__ (-)</code> و <code>__mul__ (*)</code> انجام می‌شود.</p>
    </div>
  </div>
</div>

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        else:
            raise ValueError("عملوند باید از نوع Vector باشد")

    def __mul__(self, scalar):
        if isinstance(scalar, (float, int)):
            return Vector(self.x * scalar, self.y * scalar)
        else:
            raise ValueError("ضریب باید عدد صحیح یا اعشاری باشد")

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
```

```python
# جمع برداری
v1 = Vector(1, 2)
v2 = Vector(3, 4)
```

```python
v3 = v1 + v2
```

```python
v3
```

```python
# ضرب بردار در عدد ثابت
v4 = v1 * 3
```

```python
v4
```

  <h2>۴. مقایسه‌ها: <code>__eq__</code> و <code>__lt__</code></h2>
  <div class="container">
    <div class="card red">
      <h3>برابری و اولویت‌بندی</h3>
      <p>به جای مقایسه آدرس‌های حافظه، می‌توانید دقیقاً تعریف کنید چه چیزی باعث می‌شود دو شیء با هم «برابر» یا یکی از دیگری «بزرگتر» باشد.</p>
      <ul>
        <li><code>__eq__</code> (==), <code>__ne__</code> (!=)</li>
        <li><code>__lt__</code> (&lt;), <code>__le__</code> (&lt;=)</li>
        <li><code>__gt__</code> (&gt;), <code>__ge__</code> (&gt;=)</li>
      </ul>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __eq__(self, other):
        if not isinstance(other, Book):
            return NotImplemented
        return (self.title == other.title) and (self.author == other.author)
```

```python
book1 = Book("آموزش پایتون", "جین دو")
book2 = Book("آموزش پایتون", "جین دو")

book1 == book2
```

```python
book1 = Book("آموزش پایتون", "جین دو")
book2 = Book("آموزش پایتون", "جان اسمیت")

book1 == book2
```

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __lt__(self, other):
        # مقایسه بر اساس اندازه بردار
        return (self.x ** 2 + self.y ** 2) < (other.x ** 2 + other.y ** 2)

    def __le__(self, other):
        return (self.x ** 2 + self.y ** 2) <= (other.x ** 2 + other.y ** 2)

    def __gt__(self, other):
        return (self.x ** 2 + self.y ** 2) > (other.x ** 2 + other.y ** 2)

    def __ge__(self, other):
        return (self.x ** 2 + self.y ** 2) >= (other.x ** 2 + other.y ** 2)
```

```python
v1 = Vector(1, 2)
v2 = Vector(3, 4)
```

```python
v1 < v2
```

```python
v1 > v2
```

```python
v1 <= v2
```

```python
v1 >= v2
```
