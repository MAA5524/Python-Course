  <div class="container">
    <div class="card green">
      <h3>یادآوری: ارث‌بری و چندریختی</h3>
      <ul>
        <li><strong>ارث‌بری (Inheritance):</strong> یک زیرکلاس، تمام ویژگی‌ها و متدها را از کلاس والد (سوپرکلاس) به ارث می‌برد.</li>
        <li><strong>چندریختی (Polymorphism):</strong> اشیاء کلاس‌های مختلف می‌توانند مانند کلاس والد مشترک رفتار کنند.</li>
      </ul>
    </div>
    <div class="card green">
      <h3>بازنویسی متد چیست؟</h3>
      <p>بازنویسی متد (Method Overriding) به یک زیرکلاس اجازه می‌دهد تا پیاده‌سازی اختصاصی خودش را برای متدی که قبلاً در کلاس والد تعریف شده، ارائه دهد. این کار برای دستیابی به چندریختی در زمان اجرا حیاتی است.</p>
    </div>
  </div>
</div>

```python
class Animal:
    def speak(self):
        print("این حیوان به یک روش عمومی صدا درمی‌آورد.")

class Dog(Animal):
    def speak(self):
        print("هاپ! هاپ!")
```

  <h2>۲. تابع <code>super()</code></h2>
  <div class="container">
    <div class="card red">
      <h3>توسعه بدون جایگزینی</h3>
      <p>تابع <code>super()</code> به زیرکلاس‌ها اجازه می‌دهد تا متدهای کلاس والد را فراخوانی کنند. این تابع زمانی بسیار مفید است که می‌خواهید عملکرد یک متد را <strong>توسعه</strong> دهید، نه اینکه آن را به طور کامل جایگزین (و نابود) کنید.</p>
    </div>
    <div class="card red">
      <h3>مثال ۱: توسعه عملکردها</h3>
      <p>ما متد <code>work()</code> را در <code>SoftwareEngineer</code> بازنویسی می‌کنیم، اما همچنان با دستور <code>super().work()</code> وظایف عمومی کارمند (کلاس والد) را نیز حفظ می‌کنیم.</p>
      <p><strong>کلاس پایه:</strong></p>
    </div>
  </div>

```python
class Employee:
    def work(self):
        print("انجام وظایف محول شده توسط مدیر.")
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
```

```python
class Student(Person):
    def __init__(self, name, age, student_id):
        super().__init__(name, age)  # مقداردهی به ویژگی‌های کلاس Person
        self.student_id = student_id  # ویژگی جدید و اختصاصی کلاس Student
```

```python
alice = Student("Alice", 20, "S12345")
f"{alice.name}, {alice.age}, {alice.student_id}"
```

<div class="section-card" id="toc3_">
  <h2>۳. بازنویسی متدهای خاص (Magic Methods)</h2>
  <div class="container">
    <div class="card orange">
      <h3>مجیک متدها چیستند؟</h3>
      <p>متدهایی هستند که نام آن‌ها با دو زیرخط (Double Underscore) شروع و تمام می‌شود (مثل <code>__str__</code> و <code>__eq__</code>). بازنویسی آن‌ها باعث می‌شود اشیاء ساخته شده توسط شما، به صورت یکپارچه با عملگرها و توابع داخلی پایتون (مثل <code>print</code> یا عملگر <code>==</code>) کار کنند.</p>
    </div>
    <div class="card orange">
      <h3>بازنویسی متد <code>__str__</code></h3>
      <p>یک رشته متنی معنادار و خوانا برای انسان تولید می‌کند. این همان متنی است که وقتی شما دستور <code>()print</code> را روی شیء خود اجرا می‌کنید، نمایش داده می‌شود.</p>
    </div>
  </div>
</div>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"
```

```python
# ساخت یک نمونه از کلاس کتاب
book = Book("The Catcher in the Rye", "J.D. Salinger")
```

```python
# چاپ کردن آبجکت به صورت مستقیم
print(book)
```

      <p>این متد بیشتر برای برنامه‌نویسان (هنگام دیباگ) استفاده می‌شود. خروجی آن باید یک رشته صریح باشد که ترجیحاً بتوان با اجرای آن کد، همان شیء را دوباره در حافظه ساخت.</p>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        return f"Book({self.title!r}, {self.author!r})"
```

```python
# ساخت نمونه‌ای از کتاب
book = Book("The Catcher in the Rye", "J.D. Salinger")
```

```python
# مشاهده نمایش فنی و دقیق کتاب
print(repr(book))  # خروجی: Book('ناطور دشت', 'جی. دی. سلینجر')
```

      <p>منطق سفارشی را برای عملگر برابری (<code>==</code>) تعریف می‌کند. این کار به شما اجازه می‌دهد اشیاء را بر اساس ارزش واقعی ویژگی‌هایشان مقایسه کنید، نه بر اساس آدرس آن‌ها در حافظه رم.</p>

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
```

```python
# ساخت نمونه‌هایی از نقاط در محور مختصات
point1 = Point(1, 2)
point2 = Point(1, 2)
point3 = Point(3, 4)
```

```python
# مقایسه کردن نقاط
point1 == point2
```

```python
point1 == point3
```

  <h2>۴. بهترین روش‌ها برای بازنویسی متدها</h2>
  <div class="container">
    <div class="card purple">
      <h3>چه زمانی از بازنویسی استفاده کنیم؟</h3>
      <ul>
        <li><strong>سفارشی‌سازی رفتار:</strong> وقتی کلاس فرزند به نسخه خاص و متفاوتی از یک وظیفه عمومی نیاز دارد.</li>
        <li><strong>پیاده‌سازی متدهای انتزاعی:</strong> برای ایجاد یک منطق قطعی و اجرایی در کلاس‌هایی که به عنوان الگو (Template) تعریف شده‌اند.</li>
        <li><strong>توسعه عملکرد:</strong> وقتی می‌خواهید مراحل جدیدی را قبل یا بعد از اجرای متدِ والد اضافه کنید.</li>

    <div class="card purple">
      <h3>بهترین شیوه‌ها (Best Practices)</h3>
      <ul>
        <li><strong>حفظ ساختار (Signature):</strong> نام متد، تعداد پارامترها و نوع خروجی را دقیقاً مانند متد والد نگه دارید.</li>
        <li><strong>استفاده از <code>super()</code>:</strong> هنگام توسعه دادن (Extend)، همیشه متد والد را صدا بزنید تا کدهای پایه‌ای از بین نروند.</li>
      </ul>
    </div>
    <div class="card purple">
      <h3>تله‌های رایج ⚠️</h3>
      <ul>
        <li><strong>فراموش کردن <code>super()</code>:</strong> باعث می‌شود راه‌اندازی کلاس یا رفتارهای اصلی، ناقص اجرا شوند.</li>
        <li><strong>وابستگی شدید (Tight Coupling):</strong> استفاده بیش از حد از بازنویسی، کد را غیرمنعطف می‌کند. اگر احساس کردید ارث‌بری به زور اتفاق افتاده، از الگوی <em>ترکیب (Composition)</em> استفاده کنید.</li>
