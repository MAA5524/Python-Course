<div class="section-card">
  <h1>فاصله‌های خالی (Whitespace) در پایتون ⌨️</h1>
</div>
<div class="section-card">
  <div class="container">
    <div class="card">
      <p>در پایتون، فاصله‌های خالی (شامل اسپیس، تب و شکستن خط) تاثیر زیادی بر <strong>خوانایی</strong> و همچنین <strong>عملکرد</strong> کد شما دارند. این فاصله‌ها فقط برای زیبایی نیستند؛ بلکه می‌توانند رفتار کد شما را تغییر دهند.</p>
      <p>Whitespace در پایتون دو نقش اصلی ایفا می‌کند:</p>
      <ul>
        <li><strong>جدا کردن توکن‌ها (Tokens)</strong>: جدا کردن عناصر پایه زبان مثل متغیرها، کلمات کلیدی و عملگرها.</li>
        <li><strong>تورفتگی (Indentation)</strong>: پایتون از تورفتگی برای تعریف بلوک‌های کد استفاده می‌کند که برای ساختارهای کنترلی مثل حلقه‌ها و شرط‌ها حیاتی است.</li>
      </ul>
      <p style="margin-bottom:0;">در این درس، نقش عملکردی Whitespace، بهترین روش‌های استفاده از آن و اشتباهات رایجی که باید از آن‌ها دوری کنید را بررسی خواهیم کرد.</p>
    </div>
  </div>
</div>

  <h2>۵. بهترین روش‌های استفاده از Whitespace</h2>
  <div class="container">
    <div class="card">
      <p>راهنمای استایل رسمی پایتون (PEP 8)، موارد زیر را توصیه می‌کند:</p>
      <ul>
        <li>اطراف عملگرها و بعد از کاما از فاصله استفاده کنید: <code>a = f(1, 2) + g(3, 4)</code>.</li>
        <li>اطراف علامت <code>=</code> وقتی برای تعیین مقدار پیش‌فرض پارامتر استفاده می‌شود، فاصله نگذارید: <code>def complex(real, imag=0.0)</code>.</li>
        <li>برای مقداردهی به متغیرها، یک فاصله قبل و بعد از علامت <code>=</code> بگذارید: <code>x = 1</code>.</li>
      </ul>
       <p style="margin-bottom:0;">رعایت این قواعد به شما کمک می‌کند کدی تمیز، خوانا و مطابق با استانداردهای پایتون (Pythonic) بنویسید.</p>
