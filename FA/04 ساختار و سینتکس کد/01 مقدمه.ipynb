  <h2>۱. مروری بر سینتکس پایتون</h2>
  <div class="container">
    <div class="card">
      <img src="https://usemynotes.com/wp-content/uploads/2021/05/python-syntax.jpg" alt="تصویر سینتکس پایتون">
    </div>
    <div class="card">
      <p>سینتکس پایتون تعیین می‌کند که یک برنامه چگونه نوشته شود. برخلاف بسیاری از زبان‌ها، پایتون از <strong>تورفتگی (Indentation)</strong> برای تعیین بلوک‌های کد استفاده می‌کند که باعث اجباری شدنِ خوانایی کد می‌شود. پایتون از نظر بصری شلوغ نیست، چون از علامت سیمیکولن (;) برای پایان خطوط یا پرانتزهای بیش از حد برای تعریف بلوک‌ها استفاده نمی‌کند.</p>
    </div>
    <div class="card" id="toc1_1_">
      <h3>دستورات ساده (Simple Statements)</h3>
      <p>یک دستور ساده در پایتون معمولاً یک خط منطقی واحد است، مثل مقداردهی به یک متغیر یا فراخوانی یک تابع.</p>
      <pre><code>x = 10
print(x)</code></pre>
    </div>
    <div class="card" id="toc1_2_">
      <h3>دستورات ترکیبی (Compound Statements)</h3>
      <p>دستورات ترکیبی (مثل <code>if</code>، <code>for</code>، <code>while</code>) شامل گروه‌هایی از دستورات دیگر هستند. آن‌ها شامل یک خطِ هدر هستند که به <strong>دو نقطه (:)</strong> ختم می‌شود و به دنبال آن، مجموعه‌ای از دستورات تو رفته قرار می‌گیرند.</p>
      <pre><code>if x > 5:
    print("x بزرگتر از ۵ است")</code></pre>
    </div>
  </div>
</div>

  <div class="container">
    <div class="card">

    <div class="card">
      <p style="margin-bottom:0;">یک ویژگی منحصر به فرد پایتون این است که بلوک‌های کد با میزان <strong>تورفتگی</strong> آن‌ها تعریف می‌شوند. تمام خطوطِ داخل یک بلوک باید به یک اندازه تورفتگی داشته باشند. بلوک کد از جایی شروع می‌شود که تورفتگی افزایش می‌یابد و جایی تمام می‌شود که تورفتگی کم می‌شود؛ برخلاف زبان‌های دیگر که از آکولاد {} یا کلمات کلیدی استفاده می‌کنند.</p>
