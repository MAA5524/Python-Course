  <h2>۲. متدها و توابع کاربردی دیکشنری</h2>
  <div class="container">
    <div class="card">
      <p>دیکشنری‌ها مجموعه‌ای از متدها و توابع داخلی برای انجام کارهای رایج دارند. در اینجا برخی از کاربردی‌ترین آن‌ها آورده شده است:</p>
      <ul>
        <li><strong><code>.keys()</code>, <code>.values()</code>, <code>.items()</code></strong>: به ترتیب نمایش‌هایی (view) از کلیدها، مقادیر، یا جفت‌های کلید-مقدار برمی‌گردانند.</li>
        <li><strong><code>.get(key, default)</code></strong>: مقدار یک کلید را به صورت ایمن بازیابی می‌کند. اگر کلید پیدا نشود، مقدار پیش‌فرض (یا <code>None</code>) را برمی‌گرداند.</li>
        <li><strong><code>.update(other_dict)</code></strong>: یک دیکشنری دیگر را در دیکشنری فعلی ادغام می‌کند.</li>
        <li><strong><code>.pop(key, default)</code></strong>: یک کلید را حذف کرده و مقدار آن را برمی‌گرداند.</li>
        <li><strong><code>.popitem()</code></strong>: آخرین جفت کلید-مقدار وارد شده را حذف کرده و برمی‌گرداند.</li>
        <li><strong><code>.setdefault(key, default)</code></strong>: مقدار یک کلید را برمی‌گرداند؛ اگر کلید وجود نداشته باشد، آن را با مقدار پیش‌فرض درج می‌کند.</li>
        <li><strong><code>len(my_dict)</code></strong>: تعداد جفت‌های کلید-مقدار را برمی‌گرداند.</li>
        <li><strong><code>sorted(my_dict)</code></strong>: یک لیست مرتب‌شده جدید از کلیدهای دیکشنری برمی‌گرداند.</li>
      </ul>
    </div>
  </div>
</div>
