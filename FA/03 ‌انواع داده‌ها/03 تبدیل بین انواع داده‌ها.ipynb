        <li><strong>امنیت داده:</strong> پایتون فقط زمانی این کار را می‌کند که مطمئن باشد اطلاعاتی از دست نمی‌رود (مثلاً تبدیل عدد صحیح به اعشاری).</li>

    </div>

```python
integer_number = 123
float_number = 1.23

new_number = integer_number + float_number
```

```python
# نمایش مقدار جدید و نوع داده آن
print("Value:", new_number)
print("Data Type:", type(new_number))
```

<div class="section-card" id="toc4_">
  <h2>۴. تبدیل صریح (Explicit Type Conversion)</h2>
  <div class="container">
    <div class="card red">
      <p>این روش که به آن <strong>Typecasting</strong> هم می‌گویند، یعنی شما به عنوان برنامه‌نویس دستور می‌دهید که یک داده به نوع دیگری تبدیل شود.</p>
    </div>
    <div class="card red">
      <h3>توابع پرکاربرد برای تبدیل:</h3>
      <ul>
        <li><code>int()</code>: تبدیل به عدد صحیح.</li>
        <li><code>float()</code>: تبدیل به عدد اعشاری.</li>
        <li><code>str()</code>: تبدیل به رشته (متن).</li>
      </ul>
    </div>
     <div class="card red">
      <h3>مثال: جمع رشته و عدد</h3>
      <p>فرض کنید یک عدد به صورت متن (داخل کوتیشن) دارید و می‌خواهید آن را با یک عدد واقعی جمع کنید. اینجا باید از تبدیل صریح استفاده کنید.</p>
    </div>
  </div>
</div>

```python
num_string = '12'
num_integer = 23

print("Data type of num_string before Type Casting:", type(num_string))
```

```python
# انجام تبدیل صریح
num_string = int(num_string)

print("Data type of num_string after Type Casting:", type(num_string))
```

```python
num_sum = num_integer + num_string

print("Sum:", num_sum)
print("Data type of num_sum:", type(num_sum))
```
