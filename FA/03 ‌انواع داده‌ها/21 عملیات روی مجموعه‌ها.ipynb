      <h3>زیرمجموعه، ابرمجموعه و مجزا بودن</h3>
      <p>شما همچنین می‌توانید روابط بین دو مجموعه را بررسی کنید:</p>
      <ul>
        <li><strong>زیرمجموعه (Subset)</strong>: بررسی اینکه آیا تمام عناصر یک مجموعه در مجموعه دیگر هستند با <code>.issubset()</code> یا عملگر <code><=</code>.</li>
        <li><strong>ابرمجموعه (Superset)</strong>: بررسی اینکه آیا یک مجموعه شامل تمام عناصر مجموعه دیگر است با <code>.issuperset()</code> یا عملگر <code>>=</code>.</li>
        <li><strong>مجزا (Disjoint)</strong>: بررسی اینکه آیا دو مجموعه هیچ عنصر مشترکی ندارند با <code>.isdisjoint()</code>.</li>
      </ul>
    </div>
  </div>
</div>

```python
{1, 2}.issubset(set_a) # {1, 2} <= set_a
```

```python
set_b.issuperset({5}) # set_b >= {5}
```

```python
set_a.isdisjoint({6, 7})
```

  <h2>۲. تفاوت بین متدها و عملگرهای مجموعه</h2>
  <div class="container">
    <div class="card">
      <p>پایتون دو راه برای انجام عملیات مجموعه‌ها ارائه می‌دهد: متدها (مثل <code>.union()</code>) و عملگرها (مثل <code>|</code>). تفاوت اصلی در انعطاف‌پذیری آن‌هاست:</p>
      <ul>
        <li><strong>عملگرها</strong> (<code>|</code>, <code>&</code>, <code>-</code>, <code>^</code>) نیاز دارند که هر دو طرفِ عملگر حتماً <strong>مجموعه (set)</strong> باشند.</li>
        <li><strong>متدها</strong> (<code>.union()</code>, <code>.intersection()</code> و غیره) منعطف‌تر هستند و می‌توانند هر <strong>شیء قابل پیمایشی</strong> (مثل لیست یا تاپل) را به عنوان ورودی بپذیرند. آن‌ها به صورت داخلی ورودی را به مجموعه تبدیل کرده و سپس عملیات را انجام می‌دهند.</li>
      </ul>
      <p>برای کدهای کوتاه و خوانا وقتی فقط با مجموعه‌ها کار می‌کنید از عملگرها استفاده کنید. اما وقتی ترکیبی از مجموعه‌ها و سایر انواع داده (مثل لیست) دارید، متدها گزینه بهتری هستند.</p>
    </div>
  </div>
</div>
