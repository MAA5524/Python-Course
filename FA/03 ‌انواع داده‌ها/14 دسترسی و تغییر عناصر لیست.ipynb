  <h2>۴. اضافه کردن عناصر جدید</h2>
  <div class="container">
    <div class="card">
      <p>شما می‌توانید با استفاده از متدهای زیر، عناصر جدیدی به لیست اضافه کنید:</p>
      <ul>
        <li><code>.append()</code>: یک عنصر واحد را به <strong>انتهای</strong> لیست اضافه می‌کند.</li>
        <li><code>.extend()</code>: تمام عناصر یک مجموعه دیگر (مثل یک لیست دیگر) را به انتهای لیست فعلی اضافه می‌کند.</li>
        <li><code>.insert()</code>: یک عنصر را در یک <strong>موقعیت مشخص</strong> درج می‌کند.</li>
      </ul>

استفاده از `.append()`

```python
# اضافه کردن 'elderberry' به انتهای لیست
fruits.append('elderberry')
fruits
```

استفاده از `.extend()`

```python
# گسترش لیست با یک لیست دیگر
more_fruits = ['fig', 'grape']
fruits.extend(more_fruits)
fruits
```

استفاده از `.insert()`

```python
# درج 'apricot' در ابتدای لیست (ایندکس صفر)
fruits.insert(0, 'apricot')
fruits
```

  <h2>۵. حذف عناصر از لیست</h2>
  <div class="container">
    <div class="card">
      <p>برای حذف عناصر از لیست، چندین روش وجود دارد:</p>
      <ul>
        <li><code>.pop()</code>: عنصر موجود در یک ایندکس خاص را حذف کرده و آن را برمی‌گرداند (اگر ایندکس ندهید، آخرین عنصر را حذف می‌کند).</li>
        <li><code>.remove()</code>: اولین عنصری که با مقدار داده شده برابر باشد را پیدا و حذف می‌کند (نیازی به دانستن ایندکس نیست).</li>
        <li><code>del</code>: حذف یک آیتم یا یک بازه (Slice) بر اساس ایندکس.</li>
        <li><code>.clear()</code>: تمام آیتم‌های لیست را پاک کرده و یک لیست خالی باقی می‌گذارد.</li>
      </ul>
    </div>
  </div>

استفاده از `.pop()`

```python
# حذف آخرین آیتم و برگرداندن آن
popped_fruit = fruits.pop()
popped_fruit
```

```python
# حذف اولین آیتم (ایندکس صفر)
popped_first_fruit = fruits.pop(0)
popped_first_fruit
```

```python
fruits
```

استفاده از `.remove()`

```python
# حذف عنصر 'cherry' از لیست
fruits.remove('cherry')
fruits
```

استفاده از `del`

```python
# حذف اولین آیتم
del fruits[0]
fruits
```

```python
# حذف یک بازه (Slice)
del fruits[1:3]
fruits
```

استفاده از `.clear()`

```python
# حذف تمام آیتم‌های لیست
fruits.clear()
```
