<h1>لیست‌ها در پایتون (Lists) 📋</h1>
  <div class="container">
    <div class="card green">
      <p>یک <strong>لیست</strong> در پایتون، یک ساختار داده همه‌کاره است؛ درست مثل لیست خرید. لیست مجموعه‌ای مرتب و تغییرپذیر (Mutable) است که می‌تواند انواع مختلفی از آیتم‌ها شامل اعداد، رشته‌ها و حتی لیست‌های دیگر را در خود نگه دارد. لیست‌ها با قرار گرفتن عناصر داخل کروشه <code>[]</code> تعریف می‌شوند.</p>
      <p><strong>ویژگی‌های کلیدی:</strong></p>
      <ul>
        <li><strong>مرتب (Ordered)</strong>: آیتم‌ها ترتیب مشخصی دارند.</li>
        <li><strong>تغییرپذیر (Mutable)</strong>: می‌توانید عناصر را اضافه، حذف یا ویرایش کنید.</li>
        <li><strong>ناهمگن (Heterogeneous)</strong>: می‌تواند شامل انواع مختلف داده باشد.</li>
      </ul>
    </div>
    <div class="card green">
       <img src="../../images/list.png" alt="نمودار لیست پایتون">
    </div>
    <div class="card green">
    <h3>چه زمانی از لیست استفاده کنیم؟</h3>
      <p style="margin-bottom:0;">زمانی که به یک دنباله مرتب نیاز دارید که قصد دارید محتوای آن را تغییر دهید، بر اساس موقعیت (Index) به آن دسترسی داشته باشید، یا داده‌های متنوعی را در یکجا جمع کنید.</p>
      <ul>
        <li><strong>داده‌های متغیر</strong>: وقتی نیاز دارید عناصری را به مجموعه اضافه یا کم کنید.</li>
        <li><strong>دسترسی با ایندکس</strong>: لیست‌ها دسترسی سریع به عناصر از طریق شماره موقعیت آن‌ها را فراهم می‌کنند.</li>
        <li><strong>ترکیب انواع داده</strong>: محدودیتی در نوع داده‌های داخل یک لیست وجود ندارد.</li>
        <li><strong>پیمایش ساده</strong>: ساختار مرتب لیست‌ها، آن‌ها را برای استفاده در حلقه‌ها (Iteration) عالی می‌کند.</li>
      </ul>
    </div>

      <p><strong>لیست‌های خالی:</strong> می‌توانید یک لیست خالی با <code>[]</code> یا <code>list()</code> بسازید که برای پر کردن داده‌ها در مراحل بعدی برنامه مفید است.</p>
      <p><strong>از سایر انواع داده:</strong> سازنده <code>list()</code> می‌تواند سایر اشیاء قابل پیمایش (مثل رشته‌ها یا تاپل‌ها) را به لیست تبدیل کند.</p>
      <p><strong>با عملگر <code>*</code>:</strong> می‌توانید لیستی از عناصر تکراری بسازید؛ مثلاً <code>[0] * 10</code> لیستی با ده عدد صفر می‌سازد.</p>
      <p><strong>با تابع <code>range()</code>:</strong> استفاده از <code>list(range(10))</code> برای ساخت لیستی از اعداد متوالی.</p>
