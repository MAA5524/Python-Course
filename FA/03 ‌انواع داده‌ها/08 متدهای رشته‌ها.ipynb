      <h3>تغییر بزرگی و کوچکی حروف (Changing Case)</h3>
      <ul>
        <li><code>.upper()</code>: تمام حروف رشته را به حروف بزرگ تبدیل می‌کند.</li>
        <li><code>.lower()</code>: تمام حروف رشته را به حروف کوچک تبدیل می‌کند.</li>
        <li><code>.capitalize()</code>: فقط اولین حرف رشته را بزرگ می‌کند.</li>
        <li><code>.title()</code>: حرف اول تمام کلمات موجود در رشته را بزرگ می‌کند.</li>
      </ul>
    </div>
  </div>
</div>

```python
text = "Mohammad"
```

```python
text.upper()
```

```python
text.lower()
```

```python
text.capitalize()
```

```python
text.title()
```

       <h3>جستجو و جایگزینی (Search and Replace)</h3>
       <ul>
         <li><code>.find(substring)</code>: موقعیت (Index) اولین جایی که کلمه مورد نظر پیدا شد را برمی‌گرداند. اگر پیدا نشود، <code>-1</code> می‌دهد.</li>
         <li><code>.replace(old, new)</code>: تمام تکرارهای کلمه <code>old</code> را با <code>new</code> جایگزین کرده و یک رشته جدید برمی‌گرداند.</li>
       </ul>
    </div>
  </div>
</div>

```python
text = "hello world"
text.find("world")
```

```python
text.find("python")
```

```python
text = "hello world"
text.replace("world", "Python")
```

```python
# اگر کلمه مورد نظر در متن نباشد، هیچ تغییری رخ نمی‌دهد
text.replace("x", "X")
```
