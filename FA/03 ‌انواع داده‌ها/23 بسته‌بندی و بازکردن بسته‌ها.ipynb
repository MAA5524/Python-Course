<div class="section-card">
  <h1>بسته‌بندی و باز کردن (Packing & Unpacking) 📦</h1>
</div>
<div class="section-card">
  <div class="container">
    <div class="card">
      <p>در پایتون، شما می‌توانید چندین مقدار را در یک موجودیت واحد (مثل تاپل) گروه‌بندی کنید؛ به این کار <strong>بسته‌بندی (Packing)</strong> می‌گویند. برای مثال، <code>point = x, y, z</code> مقادیر x، y و z را در یک تاپل به نام <code>point</code> بسته‌بندی می‌کند.</p>
      <p>در مقابل، اگر یک تاپل داشته باشید و بخواهید عناصر آن را استخراج کنید، از <strong>باز کردن (Unpacking)</strong> استفاده می‌کنید. عبارت <code>x, y, z = point</code> تاپل را باز کرده و مقادیر آن را بر اساس موقعیتشان به متغیرهای جداگانه اختصاص می‌دهد.</p>
    </div>
  </div>
</div>

```python
# (Packing) بسته‌بندی مقادیر در یک تاپل
point = (7, 14, 21)
```

```python
# (Unpacking) باز کردن تاپل در متغیرها
x, y, z = point
```

```python
# دسترسی به مقادیر
print("The x coordinate is:", x)
print("The y coordinate is:", y)
print("The z coordinate is:", z)
```

      <p>پایتون ۳ قابلیت باز کردن گسترده را با عملگر <code>*</code> معرفی کرد. این کار به شما اجازه می‌دهد چندین آیتم از یک توالی را در یک لیست واحد جمع‌آوری کنید. این روش زمانی مفید است که می‌خواهید چند عنصر را استخراج کرده و بقیه را یکجا جمع کنید، یا زمانی که طول توالی متغیر است.</p>
      <p style="margin-bottom:0;">عملگر <code>*</code> فقط یک بار می‌تواند در دستور انتساب استفاده شود. یک قرارداد رایج استفاده از <code>*_</code> برای جمع‌آوری و نادیده گرفتن مقادیر غیرضروری است.</p>
    </div>
  </div>
</div>

#### استفاده در تاپل‌ها

```python
# یک تاپل با چندین مقدار
numbers = (1, 2, 3, 4, 5, 6)

# استخراج دو مقدار اول و مقدار آخر، و جمع‌آوری بقیه در وسط
first, second, *middle, last = numbers

print("First:", first)
print("Second:", second)
print("Middle:", middle)
print("Last:", last)
```

#### استفاده در لیست‌ها

```python
# لیستی شامل چندین عنصر
values = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# استخراج عناصر خاص و نادیده گرفتن بقیه
first, *_, last = values

print("First:", first)
print("Last:", last)
```
