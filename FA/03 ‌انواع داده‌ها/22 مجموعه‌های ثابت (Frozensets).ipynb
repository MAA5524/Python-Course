<div class="section-card">
  <h1>مجموعه‌های تغییرناپذیر (Frozenset) در پایتون ❄️</h1>
  <div class="container">
    <div class="card">
      <p>یک <code>frozenset</code> مجموعه‌ای نامرتب از عناصر منحصربه‌فرد است، دقیقاً شبیه به یک <code>set</code> معمولی. با این حال، ویژگی اصلی آن <strong>تغییرناپذیری (Immutability)</strong> است؛ یعنی پس از ساخته شدن، عناصر آن قابل تغییر، اضافه یا حذف نیستند.</p>
      <p><strong>تفاوت‌های اصلی بین Set و Frozenset:</strong></p>
      <ul>
        <li><strong>تغییرپذیری</strong>: یک <code>set</code> معمولی تغییرپذیر است، اما <code>frozenset</code> تغییرناپذیر است.</li>
        <li><strong>قابلیت هش (Hashability)</strong>: چون تغییرناپذیر است، <code>frozenset</code> می‌تواند به عنوان کلید در دیکشنری یا به عنوان عضوی از یک مجموعه دیگر استفاده شود (کاری که با <code>set</code> معمولی غیرممکن است).</li>
        <li><strong>متدهای موجود</strong>: در <code>frozenset</code> متدهایی که باعث تغییر می‌شوند (مثل <code>.add()</code> یا <code>.remove()</code>) وجود ندارند.</li>
      </ul>
    </div>
    <div class="card">
       <img src="../../images/frozenset.png" alt="نمودار Frozenset">
    </div>
    <div class="card">
      <p><strong>موارد استفاده از Frozenset:</strong></p>
      <p>زمانی که به مجموعه‌ای از عناصر یکتا نیاز دارید که نباید در طول برنامه تغییر کنند، مانند:</p>
      <ul>
        <li><strong>کلیدهای دیکشنری</strong>: وقتی می‌خواهید یک مجموعه را کلید قرار دهید.</li>
        <li><strong>مجموعه‌ای از مجموعه‌ها</strong>: اگر می‌خواهید مجموعه‌ای داشته باشید که شامل مجموعه‌های دیگر باشد، مجموعه‌های داخلی باید <code>frozenset</code> باشند.</li>
        <li><strong>ثابت‌ها</strong>: برای تعریف مجموعه‌های ثابتی که منطق برنامه به آن‌ها وابسته است.</li>
      </ul>
       <p style="margin-bottom:0;">به طور خلاصه، <code>frozenset</code> مفهوم مجموعه را با تضمین تغییرناپذیری گسترش می‌دهد که برای حفظ یکپارچگی داده‌ها بسیار ارزشمند است.</p>
    </div>
