  <h2>۱. اتصال (Concatenation) و تکرار لیست‌ها</h2>
  <div class="container">
    <div class="card orange">
      <p style="margin-bottom:0;">شما می‌توانید لیست‌ها را با استفاده از عملگر <code>+</code> به هم متصل کنید که دو لیست را در یک لیست جدید ادغام می‌کند. برای تکرار محتویات یک لیست، از عملگر <code>*</code> استفاده می‌شود که لیست را به تعداد مشخص تکرار می‌کند.</p>
    </div>

```python
list_one = [1, 2, 3]
list_two = [4, 5, 6]
```

اتصال (Concatenation)

```python
combined_list = list_one + list_two
combined_list
```

تکرار (Replication)

```python
repeated_list = list_one * 3
repeated_list
```

  <h2>۴. معکوس کردن و مرتب‌سازی لیست</h2>
  <div class="container">
    <div class="card">
      <p>برای معکوس کردن یک لیست، می‌توانید از متد <code>.reverse()</code> استفاده کنید که لیست را <strong>در جا (In-place)</strong> تغییر می‌دهد. همچنین، تابع <code>reversed()</code> یک تکرارکننده (Iterator) معکوس برمی‌گرداند بدون اینکه لیست اصلی را تغییر دهد.</p>
    </div>
    <div class="card">
      <p>به همین ترتیب، متد <code>.sort()</code> یک لیست را <strong>در جا</strong> مرتب می‌کند (با آرگومان اختیاری <code>reverse=True</code> برای مرتب‌سازی نزولی). تابع <code>sorted()</code> یک لیست مرتب‌شده‌ی <strong>جدید</strong> برمی‌گرداند و لیست اصلی را دست‌نخورده باقی می‌گذارد.</p>
    </div>
  </div>
</div>

معکوس کردن (Reverse)

```python
numbers = [1, 5, 2, 4, 3]
numbers.reverse()
```

```python
numbers
```

```python
list(reversed(numbers))
```

مرتب‌سازی (Sort)

```python
# مرتب‌سازی صعودی لیست اصلی
numbers.sort()
numbers
```

```python
# مرتب‌سازی نزولی لیست اصلی
numbers.sort(reverse=True)
numbers
```

```python
sorted(numbers)
```
