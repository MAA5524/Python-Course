      <h3>مدیریت کلیدهای ناموجود با متد .get()</h3>
      <p>برای جلوگیری از خطای <code>KeyError</code>، دیکشنری‌ها متدی به نام <code>.get()</code> دارند. این متد به شما اجازه می‌دهد به مقدار یک کلید دسترسی داشته باشید و یک «مقدار پیش‌فرض» تعیین کنید تا اگر کلید پیدا نشد، آن مقدار برگردانده شود.</p>
      <p style="margin-bottom:0;">در مثال زیر، کلید <code>'occupation'</code> وجود ندارد. به جای خطا، متد <code>.get()</code> مقدار پیش‌فرض <code>'تعیین نشده'</code> را برمی‌گرداند.</p>
    </div>
  </div>
</div>

```python
occupation = person_info.get('occupation', 'Not specified')
occupation
```

      <ul>
        <li><code>.keys()</code>: تمام کلیدهای دیکشنری را برمی‌گرداند.</li>
        <li><code>.values()</code>: تمام مقادیر دیکشنری را برمی‌گرداند.</li>
        <li><code>.items()</code>: تمام جفت‌ها را به صورت لیستی از تاپل‌های (key, value) برمی‌گرداند.</li>
      </ul>
      <p style="margin-bottom:0;">این اشیاء پویا هستند؛ یعنی اگر دیکشنری تغییر کند، این لیست‌ها هم خودکار بروز می‌شوند.</p>

```python
# دسترسی به تمام کلیدها
keys = person_info.keys()
keys
```

```python
# دسترسی به تمام مقادیر
values = person_info.values()
values
```

```python
# دسترسی به تمام آیتم‌ها
items = person_info.items()
items
```

        <li><strong>حذف آیتم‌ها:</strong> یک کتاب را با متد <code>.pop()</code> حذف کرده و ژانر حذف شده را نمایش دهید. سپس با دستور <code>del</code> یک کتاب دیگر را حذف کنید.</li>
        <li><strong>پیمایش (Iteration):</strong> لیستی از تمام نام کتاب‌ها (کلیدها) را ساخته و چاپ کنید. سپس لیستی از تمام ژانرها (بدون تکرار) بسازید و چاپ کنید.</li>
        <li><strong>پاکسازی:</strong> تمام محتویات دیکشنری <code>book_genre</code> را پاک کرده و آن را چاپ کنید تا مطمئن شوید خالی است.</li>
