  <div class="container">
    <div class="card" id="toc1_1_">
      <h3>ارجاع (Reference) چیست؟</h3>
      <p>یک «ارجاع» در پایتون، در واقع یک نام است که به یک شیء اشاره می‌کند. درست مثل اینکه اسم «علی» به یک شخص واقعی اشاره می‌کند، در پایتون هم یک متغیر مثل <code>friend</code> به یک شیء واقعی در حافظه (مثلاً رشته "Kian") اشاره می‌کند.</p>
    </div>
    <div class="card" id="toc1_2_">
       <h3>چگونه متغیرها به اشیاء اشاره می‌کنند؟</h3>
       <p style="margin-bottom:0;">وقتی می‌نویسید <code>friend = "Kian"</code>، شما در واقع یک برچسب به نام <code>friend</code> می‌سازید و آن را به شیء "Kian" می‌چسبانید. اگر بعداً بنویسید <code>friend = "Hamidreza"</code>، فقط دارید این برچسب را از روی شیء قبلی برمی‌دارید و روی شیء جدید می‌چسبانید. شیء قبلی ("Kian") هنوز وجود دارد، اما دیگر برچسبی روی آن نیست.</p>
    </div>
  </div>
</div>

<div class="section-card" id="toc2_">
  <h2>۲. بررسی هویت شیء (Object Identity)</h2>
  <div class="container">
    <div class="card" id="toc2_1_">
      <h3>تابع <code>id()</code> چیست؟</h3>
      <p>هر شیء در پایتون یک شناسه (ID) منحصر به فرد دارد؛ چیزی شبیه به کد ملی. تابع داخلی <code>id()</code> این شناسه (که در واقع آدرس حافظه آن شیء است) را به ما نشان می‌دهد.</p>
    </div>
    <div class="card" id="toc2_2_">
       <h3>کلمه کلیدی <code>is</code></h3>
       <p>دستور <code>is</code> بررسی می‌کند که آیا دو متغیر دقیقاً به <strong>یک شیء واحد</strong> اشاره می‌کنند یا نه. این کار را با مقایسه ID آن‌ها انجام می‌دهد.</p>
    </div>
    <div class="card" id="toc2_3_">
       <h3>تفاوت مهم بین <code>is</code> و <code>==</code></h3>
       <p>تفاوت اصلی در <strong>هویت (Identity) در برابر مقدار (Value)</strong> است:</p>
       <ul>
         <li>عملگر <code>is</code> چک می‌کند آیا دو متغیر <strong>یکسان</strong> هستند (آدرس حافظه یکی است).</li>
         <li>عملگر <code>==</code> چک می‌کند آیا <strong>مقدار</strong> دو متغیر برابر است (محتوایشان یکی است).</li>
       </ul>
    </div>
  </div>
</div>

```python
name1 = "Amin"
name2 = "Amin"
```

```python
# این ممکنه True باشه چون پایتون رشته‌های کوتاه رو کش می‌کنه
name1 is name2
```

```python
# این قطعاً True هست چون مقدارشون برابره
name1 == name2
```

  <h2>۳. اشتباهات رایج و بهترین روش‌ها</h2>
  <div class="container">

      <h3>اشتباه در استفاده از <code>is</code> به جای <code>==</code></h3>
      <p><strong>بهترین روش:</strong> فقط زمانی از <code>is</code> استفاده کنید که می‌خواهید چک کنید آیا دو متغیر به <strong>دقیقاً یک شیء</strong> در حافظه اشاره دارند یا خیر (مثلاً چک کردن با <code>None</code>). برای مقایسه مقدار (مثل عدد یا متن)، همیشه از <code>==</code> استفاده کنید.</p>

       <h3>کپی کردن ارجاع به جای کپی کردن شیء</h3>

       <p><strong>بهترین روش:</strong> اگر نیاز به یک کپی جداگانه دارید، باید صراحتاً یک کپی بسازید (مثلاً با متد <code>.copy()</code> که بعداً یاد می‌گیریم).</p>

       <p>خلاصه‌ای از آنچه درباره ارجاع و هویت اشیاء یاد گرفتیم:</p>
       <ol>
          <li><strong>ارجاع به شیء (Object References):</strong> متغیرها در پایتون، ظرف‌های داده نیستند؛ بلکه برچسب‌هایی (اشاره‌گرهایی) هستند که به داده‌های اصلی در حافظه اشاره می‌کنند.</li>
          <li><strong>هویت شیء (Object Identity):</strong> هر شیء یک ID منحصر به فرد دارد که با <code>id()</code> دیده می‌شود. دستور <code>is</code> این IDها را مقایسه می‌کند تا ببیند آیا دو متغیر دقیقاً یکی هستند یا خیر.</li>
       </ol>
       <p style="margin-bottom:0;">درک این تفاوت‌ها از باگ‌های رایج جلوگیری می‌کند و به شما کمک می‌کند کدهای بهینه‌تری بنویسید. حتماً این مفاهیم را تمرین کنید.</p>
