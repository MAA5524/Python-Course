  <div class="container">
    <div class="card">
      <p>متد <code>.format()</code> یک روش همه‌کاره و قدرتمند برای قرار دادن مقادیر در رشته‌هاست. این روش قبل از معرفی f-stringها در پایتون ۳.۶، روش استاندارد پایتون بود.</p>
    </div>
    <div class="card">
       <h3>نحوه استفاده</h3>
       <p>شما می‌توانید از آکولادهای خالی <code>{}</code> به عنوان جایگزین استفاده کنید و مقادیر را به متد پاس دهید، یا از نام‌ها برای وضوح بیشتر (مخصوصاً وقتی پارامترها زیاد هستند) استفاده کنید.</p>
    </div>
  </div>

<p><strong>استفاده پایه</strong>:</p>

```python
greeting = "Hello, {}"
name = "Alice"
```

```python
greeting.format(name)
```

<p><strong>استفاده از نام برای جایگزین‌ها</strong>:</p>

```python
template = 'Hey {name}, are you {age} years old?'
```

```python
template.format(name="ali", age=30)
```

  <div class="container">

      <ul>
        <li><strong>ایجاز و خوانایی:</strong> f-stringها به طور کلی خلاصه‌تر و خواناتر هستند، چون عبارات مستقیماً درون رشته قرار می‌گیرند.</li>
