      <p><strong>مجموعه‌ها (Sets)</strong> نوعی داده در پایتون هستند که برای ذخیره مجموعه‌ای نامرتب از آیتم‌های <strong>یکتا</strong> استفاده می‌شوند. مجموعه‌ها زمانی مفید هستند که وجود یک آیتم در مجموعه، مهم‌تر از ترتیب یا تعداد تکرار آن باشد.</p>
      <p><strong>ویژگی‌های کلیدی:</strong></p>

        <li><strong>عناصر یکتا</strong>: مجموعه اجازه تکرار نمی‌دهد؛ مقادیر تکراری به صورت خودکار حذف می‌شوند.</li>
        <li><strong>تغییرپذیر (Mutable)</strong>: شما می‌توانید آیتم‌ها را به مجموعه اضافه یا از آن حذف کنید.</li>
        <li><strong>بدون ترتیب و ایندکس</strong>: آیتم‌ها ترتیب مشخصی ندارند، بنابراین مجموعه‌ها از ایندکس‌گذاری یا برش زدن (Slicing) پشتیبانی نمی‌کنند.</li>
      </ul>

      <ul>

        <li><strong>دیکشنری (Dictionary)</strong>: شامل جفت‌های کلید-مقدار است و بر اساس کلیدها (که باید یکتا باشند) ایندکس می‌شود.</li>
        <li><strong>مجموعه (Set)</strong>: زمانی استفاده می‌شود که ترتیب اهمیتی ندارد و می‌خواهید مطمئن شوید هیچ مقدار تکراری در مجموعه وجود ندارد.</li>
