  <div class="container">
    <div class="card">
      <p>عملگرهای حسابی محاسبات پایه ریاضی را انجام می‌دهند. در اینجا مروری سریع داریم:</p>
      <ol>
        <li><strong>جمع (<code>+</code>):</strong> دو مقدار را با هم جمع می‌کند.</li>
        <li><strong>تفاضل (<code>-</code>):</strong> مقدار سمت راست را از چپ کم می‌کند.</li>
        <li><strong>ضرب (<code>*</code>):</strong> دو مقدار را در هم ضرب می‌کند.</li>
        <li><strong>تقسیم (<code>/</code>):</strong> تقسیم معمولی که نتیجه آن همیشه اعشاری (float) است.</li>
        <li><strong>تقسیم صحیح (<code>//</code>):</strong> تقسیم که بخش اعشاری را حذف کرده و عدد صحیح برمی‌گرداند.</li>
        <li><strong>باقیمانده (<code>%</code>):</strong> باقیمانده تقسیم را برمی‌گرداند (Modulus).</li>
        <li><strong>توان (<code>**</code>):</strong> عدد سمت چپ را به توان عدد سمت راست می‌رساند.</li>
        <li><strong>منفی کردن (<code>-</code>):</strong> علامت عدد را قرینه می‌کند.</li>
      </ol>

  <h2>۲. عملگرهای مقایسه‌ای</h2>
  <div class="container">

  <h2>۴. عملگرهای انتساب (Assignment Operators)</h2>
  <div class="container">
    <div class="card">
       <p>عملگرهای انتساب برای ذخیره مقادیر در متغیرها استفاده می‌شوند. عملگر پایه <code>=</code> است. اما عملگرهای ترکیبی (مثل <code>+=</code>) عملیات و انتساب را همزمان و خلاصه‌تر انجام می‌دهند.</p>
       <ol>
        <li><strong>مساوی (<code>=</code>)</strong>: مقدار راست را به متغیر چپ می‌دهد.</li>
        <li><strong>جمع و انتساب (<code>+=</code>)</strong>: <code>x += 3</code> معادل <code>x = x + 3</code> است.</li>
        <li><strong>تفاضل و انتساب (<code>-=</code>)</strong>: <code>x -= 2</code> معادل <code>x = x - 2</code> است.</li>
        <li><strong>ضرب و انتساب (<code>*=</code>)</strong>: <code>x *= 4</code> معادل <code>x = x * 4</code> است.</li>
        <li><strong>تقسیم و انتساب (<code>/=</code>)</strong>: <code>x /= 2</code> معادل <code>x = x / 2</code> است.</li>
        <li><strong>تقسیم صحیح و انتساب (<code>//=</code>)</strong>: <code>x //= 3</code> معادل <code>x = x // 3</code> است.</li>
        <li><strong>باقیمانده و انتساب (<code>%=</code>)</strong>: <code>x %= 4</code> معادل <code>x = x % 4</code> است.</li>
        <li><strong>توان و انتساب (<code>**=</code>)</strong>: <code>x **= 2</code> معادل <code>x = x ** 2</code> است.</li>
      </ol>
       <p style="margin-bottom:0;">استفاده از این روش کد شما را کوتاه‌تر و خواناتر می‌کند.</p>
    </div>
  </div>
</div>

<div class="section-card" id="toc5_">
  <h2>۵. عملگرهای عضویت (Membership Operators)</h2>
  <div class="container">
    <div class="card">
      <p>عملگرهای عضویت (<code>in</code> و <code>not in</code>) بررسی می‌کنند که آیا یک مقدار خاص در یک دنباله (مثل لیست یا رشته) وجود دارد یا خیر.</p>
      <ul>
        <li><strong>in</strong>: اگر مقدار در دنباله باشد، <code>True</code> برمی‌گرداند. (مثال: <code>'h' in 'hello'</code> ← درست)</li>
        <li><strong>not in</strong>: اگر مقدار در دنباله <strong>نباشد</strong>، <code>True</code> برمی‌گرداند.</li>
      </ul>
