<div class="section-card" id="toc3_">
  <h2>۳. نام‌گذاری متغیرها</h2>
  <div class="container">
    <div class="card green">
      <p>نام متغیرها می‌تواند کوتاه (مثل <code>x</code>) یا طولانی باشد. نام‌های طولانی‌تر معمولاً بهترند چون هدف متغیر را واضح‌تر بیان می‌کنند.</p>
      <strong>قوانین نام‌گذاری در پایتون:</strong>
      <ul>
      <li>می‌توانید از هر طولی استفاده کنید.</li>
      <li>حروف کوچک و بزرگ انگلیسی (a-z, A-Z) مجاز هستند.</li>
      <li>اعداد (0-9) مجاز هستند، اما <strong>نمی‌توانند در ابتدای نام</strong> باشند.</li>
      <li>استفاده از کاراکتر زیرخط (<code>_</code>) مجاز است.</li>
    </ul>
    <strong>روش‌های رایج نوشتن نام‌های چندکلمه‌ای:</strong>
    <ul>
      <li><strong>Camel Case:</strong> حرف اول کلمه اول کوچک، بقیه کلمات بزرگ (مثال: <code>numberOfStudents</code>).</li>
      <li><strong>Pascal Case:</strong> حرف اول تمام کلمات بزرگ (مثال: <code>NumberOfStudents</code>).</li>
      <li><strong>Snake Case:</strong> کلمات با زیرخط جدا می‌شوند (مثال: <code>number_of_students</code>).</li>
    </ul>
    <strong>قراردادهای مهم:</strong>
    <ul>
    <li>برای متغیرها و توابع از <strong>snake_case</strong> استفاده کنید (مثال: <code>my_variable</code>).</li>
    <li>برای کلاس‌ها از <strong>PascalCase</strong> استفاده کنید (مثال: <code>MyClass</code>).</li>
    </ul>
    </div>
</div>
