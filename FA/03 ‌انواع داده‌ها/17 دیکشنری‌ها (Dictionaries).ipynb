  <h2>۱. ویژگی‌های دیکشنری در پایتون</h2>
  <div class="container">
    <div class="card">
      <ul>
        <li><strong>بدون ترتیب (Unordered)</strong>: برخلاف لیست‌ها که آیتم‌ها ترتیب مشخصی دارند، در دیکشنری ترتیب اهمیتی ندارد. ترتیب وارد کردن آیتم‌ها، لزوماً همان ترتیب نمایش هنگام بازیابی آن‌ها نیست.</li>
        <li><strong>تغییرپذیر (Mutable)</strong>: دیکشنری‌ها قابل تغییر هستند؛ یعنی می‌توانید بعد از ساخته شدن، آیتم‌ها را اضافه، ویرایش یا حذف کنید.</li>
        <li><strong>پویا (Dynamic)</strong>: آن‌ها می‌توانند بر اساس نیاز بزرگ یا کوچک شوند و هیچ محدودیت اندازه از پیش تعریف شده‌ای ندارند.</li>
        <li><strong>ایندکس‌شده با کلید</strong>: دیکشنری‌ها با استفاده از «کلیدها» ایندکس می‌شوند. کلید می‌تواند هر نوع داده تغییرناپذیری باشد (معمولاً رشته یا عدد).</li>
        <li><strong>یکتا بودن کلیدها</strong>: هیچ دو آیتمی در یک دیکشنری نمی‌توانند کلید یکسان داشته باشند. اگر سعی کنید مقداری با کلید تکراری وارد کنید، مقدار قدیمی با مقدار جدید جایگزین می‌شود.</li>
        <li><strong>دسترسی سریع</strong>: مقادیر در دیکشنری با استفاده از کلید مربوطه فراخوانی می‌شوند، که بازیابی داده‌ها را بسیار سریع و بهینه می‌کند.</li>
      </ul>
    </div>
  </div>
</div>
