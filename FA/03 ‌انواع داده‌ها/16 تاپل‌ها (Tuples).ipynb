  <h2>۲. تغییرناپذیری تاپل‌ها (Immutability)</h2>
  <div class="container">
    <div class="card">
      <p>تغییرناپذیری یعنی محتویات یک تاپل پس از ساخت، نمی‌توانند تغییر کنند، اضافه شوند یا حذف شوند. این ویژگی مزایای زیر را دارد:</p>
      <ul>
        <li><strong>پایداری و امنیت:</strong> جلوگیری از تغییر تصادفی داده‌ها.</li>
        <li><strong>کارایی (Efficiency):</strong> تاپل‌ها می‌توانند سریع‌تر و بهینه‌تر از لیست‌ها باشند.</li>
        <li><strong>استفاده به عنوان کلید دیکشنری:</strong> چون تغییر نمی‌کنند، می‌توانند کلیدِ یک دیکشنری باشند.</li>
      </ul>
      <p style="margin-bottom:0;">اگر واقعاً نیاز دارید یک تاپل را «بروزرسانی» کنید، باید یک تاپل کاملاً جدید بسازید (مثلاً با ترکیب بخش‌هایی از تاپل قدیمی با مقادیر جدید).</p>

فرض کنید تاپلی به نام `my_info` داریم؛ وقتی مقداردهی شود، محتویات آن ثابت است:


```python
my_info = ("Jane Doe", "123 Elm Street", 28)
```

اگر سعی کنید یکی از مقادیر آن را تغییر دهید، با خطا مواجه می‌شوید:


```python
my_info[2] = 29  # خطای TypeError
```

برای «بروزرسانی» یک مقدار، باید یک تاپل کاملاً جدید ایجاد کنید:

```python
my_info = my_info[:2] + (29,)
```

  <h2>۳. عملیات روی تاپل‌ها</h2>
  <div class="container">
    <div class="card">
      <p>با وجود تغییرناپذیری، تاپل‌ها از چندین عملیات پشتیبانی می‌کنند که منجر به ساخت تاپل‌های جدید می‌شوند:</p>
      <ul>
        <li><strong>اتصال (<code>+</code>) و تکرار (<code>*</code>):</strong> چسباندن یا تکرار تاپل‌ها برای ساخت یک تاپل جدید.</li>
        <li><strong>بررسی عضویت (<code>in</code>):</strong> بررسی وجود یک عنصر در تاپل.</li>
        <li><strong>متدها:</strong> تاپل‌ها فقط دو متد دارند: <code>.count(value)</code> برای شمارش تکرار و <code>.index(value)</code> برای پیدا کردن اولین ایندکس یک مقدار.</li>
        <li><strong>مرتب‌سازی و معکوس کردن:</strong> تاپل‌ها متد <code>.sort()</code> یا <code>.reverse()</code> ندارند. در عوض باید از توابع داخلی <code>sorted(my_tuple)</code> و <code>reversed(my_tuple)</code> استفاده کنید که به ترتیب یک لیست جدید یا یک تکرارکننده برمی‌گردانند.</li>
      </ul>
    </div>
  </div>
</div>

### اتصال (Concatenation)

```python
tuple1 = (1, 2, 3)
tuple2 = (4, 5, 6)
combined_tuple = tuple1 + tuple2

combined_tuple
```

### تکرار (Repetition)

```python
repeated_tuple = tuple1 * 2
repeated_tuple
```

### بررسی عضویت (Membership Testing)

```python
5 in tuple2  # True برمی‌گرداند
```

```python
7 in tuple2  # False برمی‌گرداند
```

### متدهای Count و Index

```python
tuple3 = (1, 2, 3, 2, 4, 2)
```

```python
# شمارش تکرار
count_of_twos = tuple3.count(2)
count_of_twos
```

```python
# پیدا کردن ایندکس اولین تکرار
index_of_first_two = tuple3.index(2)
index_of_first_two
```

### مرتب‌سازی و معکوس کردن

```python
tuple4 = (1, 5, 3, 2, 4)
sorted(tuple4) # یک لیست مرتب‌شده جدید برمی‌گرداند
```

```python
tuple(reversed(tuple4)) # یک تاپل معکوس جدید برمی‌گرداند
```

        <li><strong>ساخت و دسترسی:</strong> از پرانتز <code>()</code> برای ساخت و از ایندکس‌گذاری/برش زدن استاندارد برای دسترسی استفاده می‌شود.</li>
        <li><strong>عملیات تاپل:</strong> عملیاتی مثل اتصال (<code>+</code>) و متدهایی مثل <code>.count()</code> اطلاعات را بدون تغییر در تاپل اصلی فراهم می‌کنند.</li>
        <li><strong>موارد استفاده:</strong> تغییرناپذیری آن‌ها باعث می‌شود قابل هش (Hashable) باشند و به عنوان کلید دیکشنری استفاده شوند. همچنین برای داده‌های ثابت مثل مختصات جغرافیایی عالی هستند.</li>
      </ol>
      <p style="margin-bottom:0;">تاپل‌ها مکمل لیست‌ها هستند و راهی قابل پیش‌بینی و مطمئن برای مدیریت مجموعه‌های ثابت داده فراهم می‌کنند.</p>
