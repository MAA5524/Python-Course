  <h2>۶. برش رشته‌ها (String Slicing)</h2>
  <div class="container">
    <div class="card">
      <p>برش یا Slicing به شما اجازه می‌دهد بخشی از رشته را با استفاده از الگوی <code>string[start:stop]</code> استخراج کنید. این دستور از ایندکس <code>start</code> شروع می‌کند و تا قبل از <code>stop</code> ادامه می‌دهد.</p>
