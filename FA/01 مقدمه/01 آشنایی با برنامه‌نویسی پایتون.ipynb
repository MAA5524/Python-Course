  <h2>۴. پایتون چگونه کار می‌کند؟ ⚙️</h2>
  <div class="container">
    <!-- Card 1: Interpreter vs Compiler -->
    <div class="card green">
      <h3>مفسر (Interpreter) در برابر کامپایلر</h3>
      <p>پایتون یک زبان <strong>مفسری (Interpreted)</strong> است. برخلاف زبان‌های کامپایلری (مثل جاوا یا ++C) که کل کد را یکجا ترجمه می‌کنند، پایتون کدها را <strong>خط به خط</strong> خوانده و اجرا می‌کند.</p>
      <ul>
          <li><strong>کامپایلر:</strong> مثل این است که یک کتاب را کامل ترجمه کنید و بعد برای خواندن چاپ کنید. (اجرای سریع‌تر).</li>
          <li><strong>مفسر:</strong> مثل یک مترجم همزمان است که هر جمله‌ای که می‌گویید را همان لحظه ترجمه می‌کند. (توسعه و کدنویسی سریع‌تر).</li>
      </ul>
      <img src="../../images/interpreter-vs-compiler-example.png" alt="تفاوت کامپایلر و مفسر">
    </div>
    <!-- Card 2: Pros & Cons -->
    <div class="card green">
      <h3>مزایا و معایب زبان مفسری</h3>
      <div style="display: grid; grid-template-columns: 1fr 1fr; gap: 15px;">
          <div>
              <p><strong>✅ مزایا:</strong></p>
              <ul>
                  <li><strong>دیباگ آسان:</strong> دقیقاً در همان خطی که مشکل دارد متوقف می‌شود.</li>
                  <li><strong>تست سریع:</strong> نیازی به صبر کردن برای کامپایل شدن کل برنامه نیست.</li>
              </ul>
          </div>
          <div>
              <p><strong>❌ معایب:</strong></p>
              <ul>
                  <li><strong>سرعت:</strong> برای پردازش‌های سنگین ریاضی کُندتر از ++C است.</li>
                  <li><strong>مصرف منابع:</strong> معمولاً حافظه رم (RAM) بیشتری اشغال می‌کند.</li>
              </ul>
          </div>
      </div>
    </div>
  </div>
