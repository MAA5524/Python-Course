  <h2>۱. کامپریهنشن‌ لیست (List Comprehensions)</h2>
  <div class="container">
    <div class="card orange">
      <p><strong>سینتکس پایه: <code>[expression for item in iterable]</code></strong></p>
      <p>این روش راهی کوتاه برای ساخت لیست با استفاده از یک عبارت و یک حلقه <code>for</code> در داخل کروشه فراهم می‌کند.</p>
      <p><strong>استفاده از شرط: <code>[expression for item in iterable if condition]</code></strong></p>
      <p>می‌توانید با اضافه کردن یک شرط <code>if</code>، آیتم‌ها را فیلتر کنید.</p>
      <p><strong>حلقه‌های تودرتو:</strong></p>
      <p>کامپریهنشن‌ لیست می‌تواند شامل حلقه‌های تودرتو نیز باشد.</p>

**سینتکس پایه**

```python
squares = [x**2 for x in range(10)]
squares
```

**استفاده از شرط (Filtering)**

```python
# فقط توان دوم اعداد زوج
even_squares = [x**2 for x in range(10) if x % 2 == 0]
even_squares
```

**حلقه‌های تودرتو (Nested Loops)**

```python
# تبدیل یک ماتریس به یک لیست ساده (Flattening)
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
flattened = [num for row in matrix for num in row]
flattened
```

- ساخت لیستی از مکعب ۱۰ عدد اول.
- استخراج تمام حروف صدا‌دار (Vowels) از یک رشته متنی.
- تبدیل لیستی از دماها از سانتی‌گراد به فارنهایت.

**سینتکس پایه دیکشنری**

```python
squared_dict = {x: x**2 for x in range(10)}
squared_dict
```

**استفاده از شرط در دیکشنری**

```python
even_squares_dict = {x: x**2 for x in range(10) if x % 2 == 0}
even_squares_dict
```

**معکوس کردن یک دیکشنری (Inverting)**

```python
my_dict = {'a': 1, 'b': 2, 'c': 3}
inverted_dict = {value: key for key, value in my_dict.items()}
inverted_dict
```

- ساخت دیکشنری نگاشت اعداد به توان دوم آن‌ها.

- ساخت دیکشنری از کلمات و طول آن‌ها از یک لیست متنی.
