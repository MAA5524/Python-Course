<div class="section-card" id="toc1_">
  <h2>۱. تکرار معین (Definite Iteration) در برنامه‌نویسی</h2>
  <div class="container">
    <div class="card">
      <p>تکرار معین که معمولاً با حلقه <code>for</code> شناخته می‌شود، یک بلوک کد را به تعداد مشخصی اجرا می‌کند. در طول تاریخ، زبان‌های مختلف انواع متفاوتی از این حلقه را ارائه داده‌اند، مانند حلقه‌های محدوده عددی یا حلقه‌های سه-عبارتی (رایج در زبان C).</p>
      <p style="margin-bottom:0;">پایتون از یک حلقه <strong>مبتنی بر مجموعه (Collection-based)</strong> یا <strong>مبتنی بر تکرارکننده (Iterator-based)</strong> استفاده می‌کند. این حلقه روی اعضای یک مجموعه (مثل لیست یا رشته) پیمایش می‌کند که باعث می‌شود بسیار منعطف باشد.</p>
    </div>
    <div class="card">
      <h3>حلقه محدوده عددی (Numeric Range Loop)</h3>
        <p>یک حلقه ساده که روی بازه‌ای از اعداد حرکت می‌کند؛ مثلاً از ۱ تا ۱۰. این مدل در زبان‌هایی مثل BASIC یا Pascal رایج است.</p>
        <pre><code>for i := 1 to 10 do
    loop body</code></pre>
    </div>
    <div class="card">
      <h3>حلقه سه-عبارتی (Three-Expression Loop)</h3>
        <p>این مدل که در زبان C محبوب شد، دارای سه بخش است: مقداردهی اولیه (<code>i = 0</code>)، شرط تکرار (<code>i < 10</code>) و بروزرسانی (<code>i++</code>). این ساختار در ++C، جاوا و PHP هم دیده می‌شود.</p>
        <pre><code>for (i = 0; i < 10; i++){
    loop body
}</code></pre>
    </div>
    <div class="card">
      <h3>حلقه مبتنی بر تکرارکننده (پایتون)</h3>
        <p>این نوع حلقه به جای اعداد، روی اعضای یک مجموعه حرکت می‌کند. در هر بار تکرار، متغیر <code>item</code> مقدار بعدی توالی را می‌گیرد. این روش کد را بسیار تمیز و خوانا می‌کند.</p>
        <pre><code>for item in collection:
    loop body</code></pre>
    </div>
  </div>
</div>

<div class="section-card" id="toc2_">
  <h2>۲. حلقه <code>for</code> در پایتون</h2>
  <div class="container">
    <div class="card">
      <h3>پیمایش روی توالی‌ها (Iterating)</h3>
      <p>حلقه <code>for</code> ابزاری همه‌کاره برای پیمایش روی عناصر هر شیء قابل تکرار (Iterable) است:</p>
      <ul>
        <li><strong>لیست‌ها، تاپل‌ها و مجموعه‌ها:</strong> روی تک‌تک آیتم‌های موجود حرکت می‌کند.</li>
        <li><strong>رشته‌ها:</strong> روی تک‌تک کاراکترهای متن حرکت می‌کند.</li>
        <li><strong>دیکشنری‌ها:</strong> می‌توانید روی کلیدها (با <code>.keys()</code>)، مقادیر (با <code>.values()</code>) یا جفت‌های کلید-مقدار (با <code>.items()</code>) پیمایش کنید.</li>
      </ul>
    </div>
  </div>
</div>

**استفاده در لیست‌ها (Lists)**

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

**استفاده در تاپل‌ها (Tuples)**

```python
coordinates = (1, 2, 3)
for coordinate in coordinates:
    print(coordinate)
```

**استفاده در مجموعه‌ها (Sets)**

```python
numbers_set = {1, 2, 3, 4, 5}
for number in numbers_set:
    print(number)
```

**استفاده در دیکشنری‌ها (Dictionaries)**

```python
person = {"name": "Alice", "age": 25, "gender": "Female"}
# پیمایش روی کلیدها
for key in person:
    print(key)

# پیمایش روی مقادیر
for value in person.values():
    print(value)

# پیمایش روی جفت‌های کلید-مقدار
for key, value in person.items():
    print(f"{key}: {value}")
```

**استفاده در رشته‌ها (Strings)**

```python
greeting = "Hello"
for letter in greeting:
    print(letter)
```

<div class="section-card" id="toc3_">
  <h2>۳. استفاده از <code>range()</code></h2>
  <div class="container">
    <div class="card">
      <p style="margin-bottom:0;">تابع <code>range()</code> یک دنباله از اعداد تولید می‌کند که معمولاً برای تکرار یک حلقه به تعداد دفعات مشخص استفاده می‌شود.</p>

```python
# استفاده از range برای تکرار روی دنباله‌ای از اعداد
for i in range(5):  # اعداد ۰، ۱، ۲، ۳، ۴ را تولید می‌کند
    print(i)
```

```python
# استفاده از range با تعیین نقطه شروع، پایان و گام حرکت
for i in range(2, 10, 2):  # اعداد ۲، ۴، ۶، ۸ را تولید می‌کند
    print(i)
```

<div class="section-card" id="toc4_">
  <h2>۴. توابع کمکی در حلقه‌های For</h2>
  <div class="container">
    <div class="card">
      <p>پایتون توابع داخلی متعددی دارد که قابلیت‌های حلقه‌های <code>for</code> را گسترش می‌دهند.</p>
    </div>
    <div class="card" id="toc4_1_">
       <h3>تابع <code>enumerate()</code></h3>
       <p>این تابع زمانی استفاده می‌شود که در کنار مقدارِ هر آیتم، به <strong>شمارنده (ایندکس)</strong> آن هم نیاز داشته باشید. این تابع ایندکس و مقدار را به صورت همزمان برمی‌گرداند.</p>
    </div>
  </div>

```python
# مثال ماتریس
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
```

```python
# پیمایش روی هر سطر در ماتریس
for row in matrix:
    # پیمایش روی هر عنصر در آن سطر
    for element in row:
        print(element)
```
