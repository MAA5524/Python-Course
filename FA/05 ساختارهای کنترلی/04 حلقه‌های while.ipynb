<div class="section-card" id="toc1_">
  <h2>۱. نوشتن حلقه‌های <code>while</code> با شرط کنترلی</h2>
  <div class="container">
    <div class="card orange">
      <p>یک حلقه <code>while</code> بلوک کد خود را تا زمانی که شرط کنترلی <strong>غلط (False)</strong> شود، به طور مداوم اجرا می‌کند. شرطِ حلقه، قبل از هر بار اجرای بدنه بررسی می‌شود.</p>
      <p style="margin-bottom:0;">در این مثال، حلقه تا زمانی که <code>counter</code> کمتر از ۵ باشد اجرا می‌شود. دستور <code>counter += 1</code> بسیار حیاتی است؛ زیرا تضمین می‌کند که شرط در نهایت غلط شده و حلقه به پایان برسد.</p>
    </div>
  </div>
</div>

```python
counter = 0
while counter < 5:
    print(f"The counter is at {counter}")
    counter += 1  # افزایش شمارنده
```

  <h2>۳. حلقه‌های <code>while</code> تودرتو (Nested)</h2>
  <div class="container">
    <div class="card">
      <p style="margin-bottom:0;">حلقه تودرتو به معنای وجود یک حلقه در دل حلقه‌ای دیگر است. در حلقه‌های <code>while</code>، استفاده از حالت تودرتو برای انجام کارهای پیچیده‌ای که نیاز به تکرار چندباره یک عملیات تکراری دارند (مثل پیمایش ماتریس‌ها) بسیار مفید است.</p>
    </div>

```python
outer_counter = 0
while outer_counter < 3:
    inner_counter = 0
    while inner_counter < 2:
        print(f"Outer counter: {outer_counter}, Inner counter: {inner_counter}")
        inner_counter += 1
    outer_counter += 1
```
