      <p>ساختارهای انتخابی (یا دستورات شرطی) به برنامه اجازه می‌دهند بر اساس اینکه آیا شرایط خاصی برقرار هست یا خیر، بلوک‌های متفاوتی از کد را اجرا کند. پایتون دستورات <code>if</code>، <code>elif</code> و <code>else</code> را برای این منظور ارائه می‌دهد.</p>

```python
count = 0
while count < 5:
    print(f"شمارش: {count}")
    count += 1
```

```python
# حلقه‌ای که تعداد تکرار آن مشخص نیست و به ورودی کاربر بستگی دارد
user_input = int(input("یک عدد وارد کنید (برای خروج 0 را بزنید): "))
while user_input != 0:
    print(f"عدد شما: {user_input}")
    user_input = int(input("یک عدد وارد کنید (برای خروج 0 را بزنید): "))
```
