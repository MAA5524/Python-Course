```python
# استفاده از break برای خروج از حلقه
for i in range(1, 10):
    if i == 5:
        break
    print(i)
print("Loop ended.")
```

  <h2>۳. تمرین: کنترل اجرای حلقه‌ها</h2>
  <div class="container">
    <div class="card">
      <p>با داشتن لیست <code>numbers = [7, 8, 0, 4, 3, 0, 5, 6, 0, 1]</code> کارهای زیر را انجام دهید:</p>
      <ol>
        <li><strong>یافتن اولین صفر</strong>: از یک حلقه <code>for</code> استفاده کنید تا اولین عدد صفر را پیدا کرده و ایندکس آن را چاپ کنید، سپس از حلقه خارج شوید.</li>
        <li><strong>چاپ اعداد غیرصفر</strong>: از یک حلقه <code>for</code> استفاده کنید تا تمام اعداد لیست به جز صفرها را چاپ کنید.</li>
        <li><strong>استفاده از <code>pass</code></strong>: روی لیست پیمایش کنید. برای اعداد غیرصفر، از <code>pass</code> به عنوان نگهدارنده کد استفاده کنید و برای صفرها، عبارت "Zero found" را چاپ کنید.</li>
      </ol>
    </div>
  </div>

```python
# راه‌حل تمرین
numbers = [7, 8, 0, 4, 3, 0, 5, 6, 0, 1]

print("--- تمرین ۱: یافتن اولین صفر ---")
for index, number in enumerate(numbers):
    if number == 0:
        print(f"اولین صفر در ایندکس {index} پیدا شد")
        break

print("\n--- تمرین ۲: چاپ اعداد غیرصفر ---")
for number in numbers:
    if number == 0:
        continue
    print(number)

print("\n--- تمرین ۳: استفاده از pass ---")
for number in numbers:
    if number == 0:
        print("Zero found")
    else:
        pass
```
