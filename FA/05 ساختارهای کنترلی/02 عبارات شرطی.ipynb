<div class="section-card" id="toc1_">
  <h2>۱. دستور <code>if</code></h2>
  <div class="container">
    <div class="card">
      <img src="../../images/if-condition.png" alt="فلوچارت دستور If">
    </div>
    <div class="card">
      <p>دستور <code>if</code> بنیادی‌ترین ساختار کنترل شرطی است. این دستور یک شرط را ارزیابی می‌کند و تنها در صورتی که آن شرط <strong>درست (True)</strong> باشد، بلوک کد مربوط به خود را اجرا می‌کند. اگر شرط غلط باشد، برنامه از آن بلوک عبور می‌کند.</p>
    </div>
    <div class="note">
      <strong>یادآوری:</strong> یک <strong>عبارت (Expression)</strong> واحدی از کد است که به یک مقدار ختم می‌شود (مثلاً <code>x > 5</code>). یک <strong>دستور (Statement)</strong> یک خط کامل کد است که عملیاتی انجام می‌دهد (مثلاً <code>print("Hello")</code>).
    </div>
  </div>
</div>

```python
x = 10
if x > 5:
    print("x is greater than 5.")
```

  <h2>۲. دستور <code>else</code></h2>
  <div class="container">
    <div class="card">
      <p style="margin-bottom:0;">دستور <code>else</code> مکمل دستور <code>if</code> است. بلوک <code>else</code> زمانی اجرا می‌شود که شرطِ دستور <code>if</code> <strong>غلط (False)</strong> ارزیابی شده باشد.</p>

```python
x = 3
if x > 5:
    print("x is greater than 5.")
else:
    print("x is less than or equal to 5.")
```

  <h2>۳. دستور <code>elif</code></h2>
  <div class="container">
    <div class="card">
      <p style="margin-bottom:0;">کلمه کلیدی <code>elif</code> مخفف "else if" است و اجازه می‌دهد بیش از دو حالت ممکن را بررسی کنیم. این دستور بعد از <code>if</code> و قبل از <code>else</code> قرار می‌گیرد تا شرط‌های اضافی را یکی پس از دیگری چک کند.</p>

<div class="section-card" id="toc4_">
  <h2>۴. درک مقادیر Truthy و Falsy</h2>
  <div class="container">
    <div class="card">
      <p>در ساختارهای شرطی، پایتون «ارزشِ منطقی» یک عبارت را بررسی می‌کند. مقادیر در پایتون یا به عنوان <strong>Truthy</strong> (درست‌گونه) شناخته می‌شوند یا <strong>Falsy</strong> (غلط‌گونه).</p>
      <p><strong>مواردی که در پایتون <code>False</code> در نظر گرفته می‌شوند (Falsy):</strong></p>
      <ul>
        <li>مقدار بولی خودِ <code>False</code>.</li>
        <li>هر عدد صفر (<code>0</code>, <code>0.0</code>, <code>0j</code>).</li>
        <li>هر توالی یا مجموعه خالی (<code>[]</code>, <code>()</code>, <code>{}</code>, <code>set()</code>, <code>''</code>).</li>
        <li>آبجکت خاص <code>None</code>.</li>
      </ul>
      <p style="margin-bottom:0;"><strong>تمام مقادیر دیگر</strong> در پایتون به عنوان <code>True</code> (Truthy) در نظر گرفته می‌شوند.</p>
    </div>
  </div>
</div>

```python
bool(False)
```

```python
bool(0), bool(0.0), bool(0j)
```

```python
bool([]), bool(()), bool({})
```

```python
bool(set()), bool('')
```

```python
bool(None)
```

<div class="section-card"  id="toc5_">
  <h2>۵. ساختارهای شرطی تودرتو (Nested)</h2>
  <div class="container">
    <div class="card">
      <p>دستورات شرطی می‌توانند درون یکدیگر قرار بگیرند. این کار اجازه می‌دهد فرآیندهای تصمیم‌گیری پیچیده‌ای بسازیم که در آن نتیجه به چندین شرط وابسته است.</p>
    </div>
  </div>
</div>

```python
# مثال از if تودرتو
x = 25
if x > 10:
    if x > 20:
        print("x is greater than 20.")
    else:
        print("x is greater than 10 but not greater than 20.")
else:
    print("x is less than or equal to 10.")
```

<div class="section-card"  id="toc6_">
  <h2>۶. عبارات شرطی (Ternary Operator)</h2>
  <div class="container">
    <div class="card">
      <p>پایتون روشی فشرده برای نوشتن دستورات ساده‌ی <code>if-else</code> در یک خط فراهم کرده است که به آن «عبارات شرطی» یا «عملگر سه‌تایی» می‌گویند: <br><code>value_if_true if condition else value_if_false</code></p>
    </div>
  </div>
</div>

```python
# مثال از عبارت شرطی تک خطی
x = 10
result = "Greater than 5" if x > 5 else "Less than or equal to 5"
print(result)
```
