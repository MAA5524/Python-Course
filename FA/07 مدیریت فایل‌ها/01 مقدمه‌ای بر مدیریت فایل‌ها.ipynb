      <p>مدیریت فایل به برنامه‌ها اجازه می‌دهد تا داده‌ها را حتی پس از بسته شدن برنامه حفظ کنند (Persistence) و با سیستم‌های ذخیره‌سازی ارتباط برقرار کنند.</p>
      <strong>کاربردهای اصلی:</strong>
      <ul>
        <li><strong>ذخیره و بازیابی:</strong> ذخیره تنظیمات کاربر، لاگ‌ها (Logs) یا مراحل بازی (Save Game).</li>

        <li><strong>اشتراک داده:</strong> انتقال اطلاعات بین برنامه‌های مختلف.</li>

        <li><code>'w'</code>: نوشتن (Write) — اگر فایل باشد، محتوایش را پاک می‌کند! اگر نباشد، می‌سازد.</li>

        <li><code>'r+'</code>: خواندن و نوشتن — فایل باید وجود داشته باشد.</li>
        <li><code>'w+'</code>: نوشتن و خواندن — فایل قبلی را پاک می‌کند.</li>
      </ul>
      <p>نکته: با اضافه کردن <code>'b'</code> به این حالت‌ها (مثلاً <code>'rb'</code> یا <code>'wb'</code>)، فایل در حالت <strong>باینری</strong> باز می‌شود (برای عکس، صدا و...).</p>

**باز کردن فایل برای خواندن (`'r'`)**

```python
#  باز کردن فایل
file_object = open('files/example.txt', 'r')

content = file_object.read()

print(content)

file_object.close()
```

**باز کردن فایل برای نوشتن (`'w'`)**
- *هشدار: محتوای قبلی فایل پاک می‌شود*

```python
# نوشتن در فایل متنی
file_object = open('files/write-example.txt', 'w')

file_object.write('Hello, World!')

file_object.close()
```

**باز کردن فایل برای الحاق (`'a'`)** - *به انتهای فایل اضافه می‌کند*

```python
# اضافه کردن متن جدید به انتهای فایل
file_object = open('files/write-example.txt', 'a')

file_object.write('\nAppend this line.')

file_object.close()
```

**خواندن و نوشتن همزمان (`'r+'`)**

```python
# خواندن و سپس نوشتن در فایل
file_object = open('files/write-example.txt', 'r+')



print('Current content:', content)

file_object.write('\nAdd this new line.')

file_object.close()
```

**کار با فایل‌های باینری (`'rb'`, `'wb'`)**

```python
# خواندن یک عکس (باینری)
binary_file = open('files/image.png', 'rb')

binary_content = binary_file.read()

file_object.close()
```

```python
# ذخیره کردن همان عکس با نام جدید
binary_file = open('files/image.jpg', 'wb')

binary_file.write(binary_content)

file_object.close()
```
