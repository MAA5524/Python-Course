      <p>از دستور <code>open(filename, 'r')</code> برای خواندن فایل‌های متنی استفاده کنید. اگر حالت (mode) را ننویسید، <code>'r'</code> پیش‌فرض است. همیشه فایل‌ها را با <code>close()</code> ببندید یا از دستور <code>with</code> برای بستن خودکار استفاده کنید.</p>
    </div>

```python
# باز کردن فایل از مسیر جاری
file_path = 'files/example.txt'
file = open(file_path, 'r')

# خواندن محتوا
content = file.read()

# بستن فایل (بسیار مهم)
file.close()
```

```python
# استفاده از 'with' که فایل را به صورت خودکار می‌بندد
with open('files/example.txt', 'r') as file:
    content = file.read()

# در اینجا فایل بسته شده است
```
