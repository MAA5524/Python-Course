  <h2>۱. باز کردن پروژه‌ها و فایل‌ها</h2>
  <div class="container">
    <div class="card purple">
      <h3>قانون طلایی VS Code</h3>
      <p>هرگز فایل‌های پایتون را به صورت تکی (تک فایل) در VS Code باز نکنید. همیشه کل <strong>پوشه پروژه (Workspace)</strong> را باز کنید. این کار باعث می‌شود تنظیمات محیط مجازی، خطایابی و مسیر فایل‌ها به درستی کار کنند.</p>
    </div>
    <div class="card purple">
      <h3>روش اول: از طریق ترمینال (روش پیشنهادی حرفه‌ای‌ها)</h3>
      <p>اگر هنگام نصب، VS Code را به PATH اضافه کرده باشید، این سریع‌ترین راه برای شروع کدنویسی است:</p>
      <ol>
        <li>ترمینال (یا Command Prompt) خود را باز کنید.</li>
        <li>با استفاده از دستور <code>cd</code> به پوشه پروژه خود بروید (مثلاً <code>cd Documents/MyProject</code>).</li>
        <li>دستور <code>code .</code> (کلمه code، یک فاصله، و بعد یک نقطه) را تایپ کرده و Enter بزنید.</li>
      </ol>
      <div class="note">
        <strong>دستور <code>code .</code> به چه معناست؟</strong><br>
        کلمه <code>code</code> ویرایشگر را باز می‌کند و علامت <code>.</code> (نقطه) به ویرایشگر می‌گوید که "همین پوشه‌ای که الان داخلش هستم" را به عنوان محیط کار (Workspace) باز کن.
      </div>
    </div>
    <div class="card purple">
      <h3>روش دوم: از طریق محیط گرافیکی (موس)</h3>
      <ul>
        <li><strong>در ویندوز:</strong> روی پوشه پروژه خود کلیک راست کرده و گزینه <strong>"Open with Code"</strong> را انتخاب کنید.</li>
        <li><strong>در همه سیستم‌عامل‌ها:</strong> برنامه VS Code را باز کنید، از منوی بالا روی <code>File</code> کلیک کرده، <code>Open Folder...</code> را انتخاب کنید و پوشه پروژه خود را به آن بدهید.</li>
      </ul>
    </div>
  </div>

        <li><strong>نوار وضعیت (Status Bar):</strong> نوار پایین صفحه که نسخه پایتون، وضعیت اتصال و اطلاعات پروژه را نشان می‌دهد.</li>
