<div class="section-card" id="toc_venv">
  <h2>۱. venv (سریع و سبک)</h2>
  <div class="container">
    <div class="card green">
      <h3>venv چیست؟</h3>
      <p>ابزار <b>venv</b> به صورت پیش‌فرض روی پایتون نصب است و نیازی به دانلود چیزی ندارید. این ابزار برای ساخت پروژه‌های سبک و استاندارد عالی است.</p>
    </div>
    <div class="card green">
      <h3>مراحل کار (Workflow)</h3>
      <p>کافیست ترمینال را باز کنید و دستورات زیر را به ترتیب اجرا کنید:</p>
      <ul>
        <li><strong>ساخت محیط:</strong> <code>python -m venv .venv</code></li>
        <li><strong>فعال‌سازی (ویندوز):</strong> <code>.venv\Scripts\activate</code></li>
        <li><strong>فعال‌سازی (مک/لینوکس):</strong> <code>source .venv/bin/activate</code></li>
      </ul>
      <div class="note">
        <strong>💡 نکته حرفه‌ای:</strong> همیشه نام پوشه محیط مجازی را <code>.venv</code> بگذارید. VS Code به طور هوشمند این نام را تشخیص می‌دهد و آن را مخفی نگه می‌دارد.
      </div>
    </div>

  </div>

<!-- SECTION 2: CONDA -->
<div class="section-card" id="toc_conda">
  <h2>۲. Conda (قدرت مطلق علم داده)</h2>
  <div class="container">
    <!-- Conda vs Anaconda -->
    <div class="card orange">
      <h3>تفاوت Conda، Miniconda و Anaconda چیست؟</h3>
      <img src="../../images/conda-anaconda-workbench.png" alt="تفاوت نسخه‌های کاندِا">
      <ul>
        <li><b>Conda:</b> موتور اصلی مدیریت پکیج‌هاست.</li>
        <li><b>Miniconda:</b> نسخه‌ی سبک که فقط شامل پایتون و Conda است. <strong>(پیشنهاد ما برای نصب)</strong></li>
        <li><b>Anaconda:</b> نسخه‌ی سنگین که هزاران کتابخانه را از قبل نصب کرده (حجم بالا).</li>
      </ul>
    </div>
    <div class="card green">
      <h3>نصب Miniconda</h3>
      <p>۱. از <a href="https://docs.anaconda.com/miniconda/" target="_blank" style="color: #29b6f6;">سایت رسمی</a> فایل نصبی را دانلود کنید.</p>
      <p>۲. <b>در ویندوز:</b> فایل <code>.exe</code> را اجرا کنید و تیک "Add to PATH" را بزنید (اختیاری اما مفید).</p>
      <p>۳. <b>در مک/لینوکس:</b> دستور <code>bash Miniconda3-latest-xxx.sh</code> را در ترمینال اجرا کنید.</p>
    </div>
    <div class="card pink">
      <h3>چرخه مدیریت محیط‌ها (Workflow)</h3>
      <p>مدیریت محیط‌ها به شما اجازه می‌دهد برای هر پروژه، نسخه پایتون و کتابخانه‌های جداگانه داشته باشید.</p>
      <!-- Step 0: Conceptual Diagram -->
      <img src="../../images/how-conda-works.png" style="max-width: 65%; opacity: 0.9;" alt="نحوه کار کاندِا">
      <div style="margin-top: 1rem;">
        <ul style="list-style-type: none; padding-left: 0;">
          <!-- Step 1 -->
          <li style="margin-bottom: 1.2rem; border-bottom: 1px solid rgba(255,255,255,0.05); padding-bottom: 0.8rem;">
            <strong style="color: #f06292; font-size: 1rem;">✨ ۱. ساختن (Creation):</strong> 
            ساخت یک فضای ایزوله با نسخه پایتون دلخواه.
            <pre><code>conda create --name my_env python=3.10</code></pre>
          </li>
          <!-- Step 2 -->
          <li style="margin-bottom: 1.2rem; border-bottom: 1px solid rgba(255,255,255,0.05); padding-bottom: 0.8rem;">
            <strong style="color: #f06292; font-size: 1rem;">🚀 ۲. فعال‌سازی (Activation):</strong> 
            وارد شدن به محیط برای استفاده از ابزارهای آن.
            <pre><code>conda activate my_env</code></pre>
          </li>
          <!-- Step 3 -->
          <li style="margin-bottom: 1.2rem; border-bottom: 1px solid rgba(255,255,255,0.05); padding-bottom: 0.8rem;">
            <strong style="color: #f06292; font-size: 1rem;">📋 ۳. مشاهده لیست (Inventory):</strong> 
            دیدن تمام محیط‌های ساخته شده روی سیستم.
            <pre><code>conda env list</code></pre>
          </li>
          <!-- Step 4 -->
          <li>
            <strong style="color: #f06292; font-size: 1rem;">🧹 ۴. حذف کردن (Cleanup):</strong> 
            پاک کردن محیطی که دیگر به آن نیاز ندارید.
            <pre><code>conda env remove -n my_env</code></pre>
          </li>
        </ul>
      </div>
    </div>
    <div class="card purple">
      <h3>مدیریت پکیج‌ها و خروجی گرفتن</h3>
      <p>Conda در مدیریت وابستگی‌های پیچیده (Dependency) بسیار بهتر از pip عمل می‌کند.</p>
      <ul>
        <li><b>نصب پکیج:</b> <code>conda install numpy</code></li>
        <li><b>بروزرسانی همه:</b> <code>conda update --all</code></li>
        <li><b>خروجی گرفتن (Export):</b> <code>conda env export > env.yml</code> (برای اشتراک‌گذاری پروژه)</li>
        <li><b>ساخت از روی فایل:</b> <code>conda env create -f env.yml</code></li>
      </ul>
    </div>
 </div>
</div>
