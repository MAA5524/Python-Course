  <div class="container">
    <div class="card purple">
      <h3>چرا این کار لازم است؟</h3>
      <p>VS Code باید بداند که از کدام «موتور پایتون» و کدام پکیج‌های نصب شده برای اجرای کد شما استفاده کند. این کار باعث می‌شود تکمیل خودکار (Auto-completion) و خطایابی (Linting) به درستی کار کنند.</p>
      <h3 style="margin-top: 15px;">مراحل کار (مشترک برای venv و Conda):</h3>
      <ul>
        <li>پوشه پروژه خود را در VS Code باز کنید.</li>
        <li>کلیدهای <code>Ctrl + Shift + P</code> (ویندوز/لینوکس) یا <code>Cmd + Shift + P</code> (مک) را بزنید تا Command Palette باز شود.</li>
        <li>عبارت <strong>Python: Select Interpreter</strong> را تایپ کرده و اینتر بزنید.</li>
        <li><strong>برای venv:</strong> مسیری را انتخاب کنید که به <code>.venv/Scripts/python.exe</code></li>
        <li><strong>برای Conda:</strong> محیطی را انتخاب کنید که نام آن (مثلاً <code>my_env</code>) نوشته شده و برچسب "conda" دارد.</li>
      </ul>

  <div class="container">

      <h3 style="margin-top: 15px;">فعال‌سازی دستی (اگر خودکار انجام نشد):</h3>
      <p>اگر نام محیط خود (مثلاً <code>(my_env)</code> یا <code>(.venv)</code>) را در ابتدای خط فرمان نمی‌بینید، می‌توانید دستی فعالش کنید:</p>
      <ul>
        <li><strong>برای venv (ویندوز):</strong> <code>.venv\Scripts\activate</code></li>
        <li><strong>برای venv (مک/لینوکس):</strong> <code>source .venv/bin/activate</code></li>
        <li><strong>برای Conda:</strong> <code>conda activate my_env</code></li>
      </ul>
