<div class="section-card" id="toc2_">
  <h2>۲. استفاده از Git (پیشنهادی)</h2>
  <div class="container">
    <!-- Install Git -->
    <div class="card">
      <h3>مرحله ۱: نصب Git</h3>
      <ul>
        <li><strong>ویندوز:</strong> فایل نصب را از سایت <a href="https://git-scm.com/" target="_blank" style="color:#ffa726">git-scm.com</a> دانلود و نصب کنید.</li>
        <li><strong>مک (macOS):</strong> ترمینال را باز کرده و تایپ کنید <code>git --version</code>؛ اگر نصب نباشد، سیستم به شما پیشنهاد نصب می‌دهد.</li>
        <li><strong>لینوکس:</strong> دستور <code>sudo apt install git</code> را اجرا کنید.</li>
      </ul>
    </div>
    <!-- Clone Repo -->
    <div class="card">
      <h3>مرحله ۲: کلون کردن (Clone) مخزن</h3>
      <p>۱. آدرس URL را از دکمه سبز رنگ <strong>Code</strong> در گیت‌هاب کپی کنید.</p>
      <p>۲. ترمینال را باز کرده و دستورات زیر را اجرا کنید:</p>
      <pre><code>git clone <REPO_URL>
cd <REPO_NAME>
code .</code></pre>
      <p>این کار تمام کدها را دانلود کرده و مستقیماً در VS Code باز می‌کند.</p>
    </div>
    <!-- Update Repo -->
    <div class="card">
      <h3>مرحله ۳: چطور کدهایتان را بروزرسانی کنید؟</h3>
      <p>اگر درس جدیدی اضافه کنم یا باگی را رفع کنم، نیازی نیست دوباره همه چیز را دانلود کنید. فقط کافیست در پوشه پروژه دستور زیر را بزنید:</p>
      <pre><code>git pull origin main</code></pre>
      <p>با این دستور، فایل‌های شما فوراً با آخرین نسخه آنلاین هماهنگ می‌شوند.</p>
    </div>
  </div>
</div>
