  <h2>۱. ابزار مدیریت پکیج چیست؟</h2>
  <div class="container">
    <div class="card image-card orange">
      <h3>نگاه کلی</h3>
      <p>مدیریت‌کننده پکیج (Package Manager) ابزاری است که فرایند نصب، بروزرسانی، تنظیم و حذف کتابخانه‌ها را خودکار می‌کند. در پایتون، این ابزار به شما کمک می‌کند کدهایی که توسط دیگران نوشته شده را به راحتی به پروژه خود اضافه کنید.</p>
      <img src="../../images/packege.jpg" alt="ابزار مدیریت پکیج">
    </div>
    <div class="card orange">
      <h3>چرا به آن نیاز داریم؟</h3>
      <ul>
        <li><strong>حل وابستگی‌ها (Dependency Resolution):</strong> اگر کتابخانه‌ی شما برای کار کردن به ۳ کتابخانه دیگر نیاز داشته باشد، مدیریت پکیج آن ۳ مورد را هم اتوماتیک نصب می‌کند.</li>
        <li><strong>کنترل نسخه:</strong> تضمین می‌کند که دقیقاً از ورژن درست یک کتابخانه استفاده کنید تا کدها خراب نشوند.</li>
        <li><strong>سهولت استفاده:</strong> ساعت‌ها جستجو و دانلود دستی در اینترنت، جای خود را به یک خط دستور ساده می‌دهد:</li>
      </ul>
      <pre><code>pip install pandas</code></pre>
    </div>
    <div class="card orange">
      <h3>ابزارهای محبوب</h3>
      <ul>
        <li><strong>pip:</strong> ابزار پیش‌فرض خود پایتون. عالی برای نصب کتابخانه‌های پایتونی.</li>
        <li><strong>Conda:</strong> ابزاری قدرتمند که علاوه بر پایتون، نرم‌افزارهای مرتبط را هم نصب می‌کند و استاندارد طلاییِ علم داده است.</li>
      </ul>
    </div>
  </div>
</div>

  <h2>۲. مدیریت‌کننده‌های پکیج چگونه کار می‌کنند؟</h2>
  <div class="container">
    <div class="card image-card purple">
      <h3>فرایند کار</h3>
      <p>این ابزارها مثل یک دستیار شخصی برای کدهای شما عمل می‌کنند. آن‌ها در اینترنت جستجو کرده، ابزارهای لازم را پیدا می‌کنند و آن‌ها را به صورت منظم در پروژه شما می‌چینند.</p>
      <img src="../../images/how-package-manager-work.png" alt="روند کار مدیریت پکیج">
    </div>
    <div class="card purple">
      <h3>مراحل چهارگانه</h3>
      <ul>
        <li><strong>جستجو (Search):</strong> مخازن آنلاین (مثل PyPI) را می‌گردند تا پکیج مورد نظر را پیدا کنند.</li>
        <li><strong>دانلود (Download):</strong> فایل‌های پکیج و تمام پیش‌نیازهایش را دانلود می‌کنند.</li>
        <li><strong>نصب (Install):</strong> فایل‌ها را متناسب با سیستم‌عامل شما استخراج و نصب می‌کنند.</li>
        <li><strong>مدیریت وابستگی‌ها:</strong> تداخل بین نسخه‌های مختلف برنامه‌ها را حل می‌کنند.</li>
      </ul>
    </div>
  </div>

        <li><strong>جلوگیری از تداخل:</strong> ابزارهای هر پروژه را از پروژه‌های دیگر جدا نگه می‌دارد.</li>
        <li><strong>تضمین پایداری:</strong> نسخه‌های نصب شده را یادداشت می‌کند تا وقتی پروژه را به همکارانتان می‌دهید، روی سیستم آن‌ها هم به درستی کار کند.</li>
        <li><strong>بروزرسانی آسان:</strong> ارتقاء به آخرین نسخه فقط با تایپ <code>pip install --upgrade</code>.</li>

  <div class="container">
    <div class="card image-card orange">

    </div>
    <div class="card orange">

      <ul>
        <li><strong>PyPI:</strong> مخزن رسمی پایتون با بیش از ۷۰۰ هزار کتابخانه. (منبع اصلی ابزار <code>pip</code>).</li>
        <li><strong>Conda Forge:</strong> یک مخزن جامع‌تر شامل پکیج‌های پایتون و نرم‌افزارهای خارجی که معمولاً در هوش مصنوعی استفاده می‌شوند.</li>
      </ul>
      <div class="note">
        <strong>💡 نکته:</strong> مخازنی مثل PyPI یک بازار دیجیتال هستند که برای هر نیازی (از ساخت سایت گرفته تا بازی‌سازی) یک ابزار آماده در آن‌ها وجود دارد.
      </div>

  <h2>۶. محیط‌های مجازی (venv)</h2>
  <div class="container">
    <div class="card green">
      <h3>مفهوم ایزوله کردن (Isolation)</h3>
      <p>فرض کنید یک پروژه شما به نسخه ۳ برنامه جنگو (Django 3) نیاز دارد و پروژه دیگر به جنگو ۴. اگر پکیج‌ها را به صورت عمومی روی سیستم نصب کنید، کدهایتان خراب می‌شود. <strong>محیط مجازی (Virtual Environment)</strong> برای هر پروژه یک «اتاق ایزوله و جداگانه» می‌سازد تا هیچ تداخلی ایجاد نشود.</p>
    </div>
    <div class="card green">
      <h3>چرا در این دوره ترکیب Pip + venv را ترجیح می‌دهیم؟</h3>
      <p>با وجود اینکه Conda برای علم داده عالی است، اما به دلیل حجم بالا و مشکلات تحریم و کندی اینترنت در ایران، ممکن است کلافه‌کننده باشد. با استفاده از ترکیب <b>Pip</b> و <b>venv</b>، شما کنترل کاملی روی سیستم دارید و می‌توانید با استفاده از میرورهای ایرانی، با بالاترین سرعت دانلود کنید.</p>
