  <div class="container">

      <p>ویندوز به صورت پیش‌فرض پایتون ندارد. باید آن را به صورت دستی از سایت رسمی نصب کنید.</p>
      <ul>
        <li>به سایت رسمی <a href="https://www.python.org/downloads/" target="_blank" style="color:#66bb6a; direction: ltr; display: inline-block;">python.org/downloads</a> بروید و آخرین نسخه (Installer) را دانلود کنید.</li>
        <li>فایل <code>.exe</code> دانلود شده را باز کنید تا صفحه نصب باز شود.</li>
      </ul>
      <!-- <div class="note-danger">
        <strong>⚠️ قدم بسیار حیاتی:</strong> در همان صفحه اول نصب، در پایین پنجره، <strong>حتماً</strong> تیک گزینه <strong>"Add Python to PATH"</strong> را فعال کنید و سپس روی Install Now کلیک کنید. اگر این تیک را نزنید، ترمینال ویندوز پایتون را نمی‌شناسد!
      </div> -->
    </div>
    <div class="card green" id="toc1_2_">
      <h3>نصب در مک (macOS)</h3>
      <ul>
        <li>مک‌های جدید نسخه مناسبی از پایتون را ندارند.</li>
        <li>به آدرس <a href="https://www.python.org/downloads/macos/" target="_blank" style="color:#66bb6a; direction: ltr; display: inline-block;">python.org/downloads/macos</a> بروید.</li>
        <li>نسخه <strong>macOS 64-bit universal2 installer</strong> را دانلود کنید.</li>
        <li>فایل <code>.pkg</code> را باز کرده و مراحل نصب استاندارد را طی کنید (Next > Agree > Install).</li>
      </ul>
    </div>
    <div class="card green" id="toc1_3_">
      <h3>نصب در لینوکس</h3>
      <p>اکثر توزیع‌های لینوکس پایتون را از قبل دارند. اما برای نصب آخرین نسخه یا ابزارهای توسعه، ترمینال را باز کرده و دستورات زیر را اجرا کنید:</p>
      <ul>
        <li><strong>در اوبونتو / دبیان:</strong> <code>sudo apt update && sudo apt install python3 python3-pip python3-venv</code></li>
        <li><strong>در فدورا:</strong> <code>sudo dnf install python3</code></li>
        <li><strong>در آرچ لینوکس:</strong> <code>sudo pacman -S python</code></li>
      </ul>
      <p>برای اطمینان از نصب موفق، در ترمینال دستور <code>python3 --version</code> یا <code>python --version</code> را تایپ کنید.</p>
    </div>

<div class="section-card" id="toc2_">
  <h2>۲. نصب VS Code</h2>
  <div class="container">
    <div class="card orange">
      <h3>ویندوز</h3>
      <ul>
        <li>فایل نصبی را از سایت <a href="https://code.visualstudio.com/Download" target="_blank" style="color:#ffa726; direction: ltr; display: inline-block;">code.visualstudio.com</a> دانلود کنید.</li>
        <li>مراحل نصب را طی کنید. وقتی به بخش "Select Additional Tasks" رسیدید، حتماً تیک گزینه‌های <strong>"Add to PATH"</strong> و <strong>"Add 'Open with Code' action to Windows Explorer"</strong> را بزنید (بسیار کاربردی است!).</li>
      </ul>
    </div>
    <div class="card orange">
      <h3>مک (macOS)</h3>
      <ul>
        <li>فایل <code>.zip</code> مخصوص مک را دانلود کنید.</li>
        <li>فایل را استخراج (Extract) کنید تا برنامه <code>Visual Studio Code.app</code> ظاهر شود.</li>
        <li><strong>مهم:</strong> این برنامه را به پوشه <strong>Applications</strong> خود Drag & Drop (بکشید و رها کنید) تا در Launchpad نمایش داده شود.</li>
      </ul>
    </div>
    <div class="card orange">
      <h3>لینوکس</h3>
      <ul>
        <li>در دبیان/اوبونتو، فایل <code>.deb</code> را دانلود کرده و با این دستور نصب کنید: <code>sudo apt install ./&lt;filename&gt;.deb</code></li>
        <li>همچنین می‌توانید از Snap Store لینوکس خود آن را مستقیم نصب کنید.</li>
      </ul>
    </div>
