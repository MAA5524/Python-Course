      <p>توابع لامبدا که به آن‌ها «توابع بی‌نام» (Anonymous Functions) هم گفته می‌شود، ویژگی قدرتمندی در پایتون برای ساخت توابع کوچک و یک‌بارمصرف هستند. این توابع معمولاً برای مدت کوتاهی مورد نیازند و استفاده از آن‌ها در کنار توابع مرتبه بالا (Higher-Order Functions) مثل <code>map()</code> و <code>filter()</code> بسیار رایج است.</p>

       <h3>ساختار و سینتکس (Syntax)</h3>
       <p>ساختار کلی بسیار ساده است: <code style="color: #fff;">lambda arguments: expression</code></p>
       <ul>
        <li><code>lambda</code>: کلمه کلیدی برای تعریف تابع.</li>
        <li><code>arguments</code>: پارامترهای ورودی (یک یا چند تا، جدا شده با کاما).</li>
        <li><code>:</code>: جداکننده آرگومان‌ها از بدنه تابع.</li>
        <li><code>expression</code>: تنها دستوری که اجرا شده و نتیجه‌اش <strong>اتوماتیک</strong> برگردانده (Return) می‌شود.</li>
      </ul>

```python
# تسک ۱: ساخت لامبدا ساده
square_function = lambda x: x ** 2
print(f"Square of 5 is {square_function(5)}")
```

```python
# تسک ۲: لامبدا با چند ورودی
product_function = lambda a, b: a * b
print(f"Product of 4 and 6 is {product_function(4, 6)}")
```

```python
# تسک ۳: استفاده با filter
ages = [14, 18, 21, 16, 30, 45, 17, 22, 61]
adult_ages = list(filter(lambda age: age >= 18, ages))
print(f"Adult ages: {adult_ages}")
```

```python
# تسک ۴: استفاده با map
prices = [19.99, 35.50, 89.99, 43.75]
discounted_prices = list(map(lambda price: round(price * 0.9, 3), prices))
print(f"Discounted prices: {discounted_prices}")
```

```python
# تسک ۵: مرتب‌سازی با کلید سفارشی
products = [('apple', 1.0), ('banana', 0.5), ('cherry', 1.5), ('date', 0.75)]
sorted_products = sorted(products, key=lambda item: item[1])
print(f"Sorted products by price: {sorted_products}")
```

```python
# تسک ۶: مرتب‌سازی دیکشنری‌ها (پاداش)
books = [
    {'title': 'Book A', 'year': 2001},
    {'title': 'Book B', 'year': 1999},
    {'title': 'Book C', 'year': 2010},
    {'title': 'Book D', 'year': 1985}
]
sorted_books = sorted(books, key=lambda book: book['year'])
print(f"Sorted books by year: {sorted_books}")
```
