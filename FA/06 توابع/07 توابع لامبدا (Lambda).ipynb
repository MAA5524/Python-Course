       <ul>

        <li><code>:</code>: جداکننده آرگومان‌ها از بدنه تابع.</li>
        <li><code>expression</code>: تنها دستوری که اجرا شده و نتیجه‌اش <strong>اتوماتیک</strong> برگردانده (Return) می‌شود.</li>
      </ul>

**آرگومان‌های متغیر:**

می‌توانید از سینتکس <code>*args</code> و <code>**kwargs</code> هم استفاده کنید.

```python
# جمع تمام اعداد ورودی (تعداد نامشخص)
sum_all = lambda *args: sum(args)
sum_all(1, 2, 3, 4)
```

```python
# چسباندن کلمات ورودی (کلید-مقدار)
merge_strings = lambda **kwargs: " ".join(kwargs.values())
merge_strings(a='Python', b='is', c='awesome')
```
