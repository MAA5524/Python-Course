      <p style="margin-bottom:0;">در این درس یاد می‌گیریم که چگونه به طور بهینه از توابع خارج شویم، مقادیر تکی یا چندگانه را بازگردانیم و مفهوم مقدار بازگشتی پیش‌فرض را درک کنیم. تسلط بر این مفاهیم به شما اجازه می‌دهد توابعی بسازید که نه تنها وظایف را انجام می‌دهند، بلکه نتایج را به شکلی موثر منتقل می‌کنند.</p>

```python
def find_book(title):
    # شبیه‌سازی لیستی از کتاب‌هایی که داریم
    my_books = ['Python Basics', 'Advanced Python', 'Data Science with Python']
    if title in my_books:
        return True
    else:
        return False
```

```python
# استفاده از توابع
book_title = 'Python Basics'
if find_book(book_title):
    print(f"کتاب '{book_title}' در قفسه موجود است.")
else:
    print(f"کتاب '{book_title}' در مجموعه ما نیست.")
```

**مثال ۱: بازگرداندن نام و نام خانوادگی به صورت جداگانه**

**مثالی از چاپ کردن (Printing)**

```python
def print_hello():
    print("Hello, World!")

print_hello()
# خود تابع مقداری را برنمی‌گرداند
```

```python
output = print_hello()
print(output) # خروجی None است
```

**مثالی از بازگرداندن (Returning)**

```python
def return_hello():
    return "Hello, World!"

greeting = return_hello()
print(greeting)
# متغیر greeting مقدار بازگشتی را در خود نگه می‌دارد
```
