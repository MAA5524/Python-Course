  <h2>۱. بسته‌بندی آرگومان‌ها در تاپل (*args)</h2>
  <div class="container">
    <div class="card">
      <p>بسته‌بندی آرگومان‌ها به صورت تاپل به یک تابع اجازه می‌دهد تا هر تعداد آرگومان موقعیتی را دریافت کرده و در داخل تابع به صورت یک تاپل به آن‌ها دسترسی داشته باشد. این کار با قرار دادن یک ستاره (<code>*</code>) قبل از نام پارامتر انجام می‌شود.</p>
      <p style="margin-bottom:0;">اگرچه هر نامی را می‌توان انتخاب کرد، اما طبق یک قرارداد جهانی از نام <code>*args</code> استفاده می‌شود.</p>
    </div>
  </div>
</div>

یک سناریوی واقعی را در نظر بگیرید که در آن نیاز دارید میانگین تعداد متغیری از نمرات آزمون را محاسبه کنید:

```python
def calculate_average(*scores):
    if scores:
        return sum(scores) / len(scores)
    else:
        return 0
```

```python
# محاسبه میانگین ۳ نمره
calculate_average(85, 90, 95)
```

```python
# محاسبه میانگین ۵ نمره
calculate_average(70, 75, 80, 85, 90)
```

یک کاربرد عملی دیگر، متصل کردن تعداد دلخواهی از رشته‌ها با یک جداکننده مشخص است:

```python
def concatenate_strings(separator, *args):
    return separator.join(args)
```

```python
# اتصال با جداکننده فاصله
concatenate_strings(" ", "Python", "is", "awesome")
```

```python
# اتصال با جداکننده خط تیره
concatenate_strings("-", "2023", "04", "21")
```

  <h2>۲. باز کردن تاپل آرگومان‌ها (Tuple Unpacking)</h2>
  <div class="container">
    <div class="card">
      <p>باز کردن تاپل آرگومان‌ها به شما اجازه می‌دهد یک مجموعه (مثل لیست یا تاپل) را بگیرید و عناصر آن را هنگام فراخوانی تابع، به صورت آرگومان‌های تکی باز کنید. این تکنیک زمانی مفید است که داده‌های شما از قبل در یک قالب ساختاریافته قرار دارند.</p>
      <p>عملگر بازکننده (<code>*</code>) محدود به تاپل‌ها نیست؛ بلکه با هر شیء قابل پیمایشی کار می‌کند. با این حال، چون مجموعه‌ها (Sets) بدون ترتیب هستند، باز کردن یک مجموعه در تابعی که انتظار داده‌های مرتب را دارد، ممکن است نتایج درستی ندهد.</p>
    </div>
  </div>

**مثال ۱**

```python
def print_book_info(title, author, year):
    print(f"Title: {title}, Author: {author}, Year: {year}")
```

```python
book_details = ('Black Horse', 'Veronica Blake', 2009)
```

```python
# باز کردن مستقیم تاپل در هنگام فراخوانی تابع
print_book_info(*book_details)
```

```python
def print_authors(*authors):
    for author in authors:
        print(author)
```

```python
author_list = ['J.K. Rowling', 'George R.R. Martin', 'J.R.R. Tolkien']
```

```python
# لیست ابتدا باز شده و به صورت آرگومان‌های تکی ارسال می‌شود، سپس در تابع دوباره به صورت یک تاپل بسته‌بندی می‌گردد
print_authors(*author_list)
```

  <h2>۳. بسته‌بندی آرگومان‌ها به صورت دیکشنری (**kwargs)</h2>
  <div class="container">
    <div class="card">
      <p>بسته‌بندی آرگومان‌ها در قالب دیکشنری از عملگر دو ستاره (<code>**</code>) استفاده می‌کند تا به تابع اجازه دهد تعداد دلخواهی از آرگومان‌های کلیدواژه‌ای را بپذیرد. این جفت‌های کلید-مقدار در قالب یک دیکشنری بسته‌بندی می‌شوند.</p>
      <p>این روش زمانی مفید است که می‌خواهید یک تابع منعطف داشته باشید که ورودی‌های نام‌گذاری شده‌ی متنوعی را بپذیرد. این قابلیت همچنین می‌تواند با آرگومان‌های موقعیتی استاندارد ترکیب شود.</p>

```python
def introduce_yourself(**kwargs):
    for key, value in kwargs.items():
        print(f"{key}: {value}")
```

```python
# معرفی یک شخص با ویژگی‌های مختلف
introduce_yourself(name="Alice", age=30, profession="Engineer")
```

**باز کردن تاپل (Tuple Unpacking)**

```python
def set_coordinates(x, y):
    print(f"X: {x}, Y: {y}")
```

```python
# روش صحیح (Best Practice)
coords = (12, 5)
set_coordinates(*coords)
```
