      <p>توابعی مثل <code>abs()</code> (قدر مطلق)، <code>sum()</code> (جمع)، <code>min()</code> (کمترین)، <code>max()</code> (بیشترین)، <code>pow()</code> (توان) و <code>round()</code> (گرد کردن) عملیات ریاضی را بسیار ساده می‌کنند.</p>

```python
numbers = [2, -5, 3.5, 4]
```

```python
# دریافت قدر مطلق تمام اعداد لیست
[abs(num) for num in numbers]
```

```python
# جمع تمام اعداد
sum(numbers)
```

```python
# پیدا کردن کوچک‌ترین و بزرگ‌ترین عدد
min(numbers), max(numbers)
```

```python
# گرد کردن اعداد به نزدیک‌ترین عدد صحیح
[round(num) for num in numbers]
```

```python
# فیلتر کردن اعداد بزرگتر از ۵
list(filter(lambda x: x > 5, numbers))
```

```python
# به توان ۲ رساندن اعداد با استفاده از map
list(map(lambda x: x**2, numbers))
```

```python
text = "The quick brown fox jumps over the lazy dog"
numbers_list = [8, 23, 45, 12, 78]
mixed_list = ["hello", 10, "world", 42]
```

```python
# تسک ۱: آنالیز رشته
print(f"تعداد کاراکترها: {len(text)}")
print(f"متن با حروف بزرگ: {text.upper()}")
```

```python
# تسک ۲: محاسبات عددی
print(f"مجموع اعداد: {sum(numbers_list)}")
print(f"بزرگترین عدد: {max(numbers_list)}")
print(f"کوچک‌ترین عدد: {min(numbers_list)}")
```

```python
# تسک ۴: تکرار با توابع داخلی
print("کلمات فیلتر شده (بدون 'the'):", end=" ")
for word in filter(lambda w: w.lower() != "the", text.split()):
    print(word, end=", ")
print()
squared_numbers = list(map(lambda x: x**2, numbers_list))
print(f"اعداد به توان ۲: {squared_numbers}")
```
