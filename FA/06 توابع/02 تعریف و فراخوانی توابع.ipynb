  <h2>۱. تابع چیست؟</h2>
  <div class="container">
    <div class="card">
      <p>در پایتون، تابع دنباله‌ای نام‌گذاری شده از دستورات است که یک محاسبه یا عملیات خاص را انجام می‌دهد. شما ابتدا یک تابع را تعریف می‌کنید و سپس آن را برای اجرای دستوراتش «فراخوانی» می‌کنید. توابع روش اصلی برای دسته‌بندی کدهای شما به گروه‌های منطقی هستند که باعث ترویج موارد زیر می‌شوند:</p>

        <li>تست‌پذیری (Testability)</li>
      </ul>

<div class="section-card" id="toc2_">
  <h2>۲. تعریف یک تابع</h2>
  <div class="container">
    <div class="card" id="toc2_1_">
       <h3>سینتکس تعریف تابع</h3>
       <img src="../../images/function-parts.png" alt="اجزای یک تابع در پایتون">
       <p>قواعد نوشتاری برای تعریف یک تابع به صورت زیر است:</p>
       <pre><code>def function_name(parameters):
    """Docstring (اختیاری اما توصیه می‌شود)"""
    # بدنه تابع
    return result  # اختیاری</code></pre>
       <ul>
         <li><code>def</code>: کلمه کلیدی که تعریف تابع را آغاز می‌کند.</li>
         <li><code>function_name</code>: نام تابع.</li>
         <li><code>parameters</code>: متغیرهای اختیاری که تابع می‌تواند دریافت کند.</li>
         <li><code>"""Docstring"""</code>: یک رشته اختیاری برای توصیف هدف تابع.</li>
         <li><strong>بدنه تابع</strong>: بلوک کد تورفته که عملیات را انجام می‌دهد.</li>
         <li><code>return</code>: کلمه کلیدی اختیاری برای خروج از تابع و بازگرداندن یک مقدار.</li>
       </ul>
    </div>
    <div class="card" id="toc2_2_">
      <h3>قراردادهای نام‌گذاری</h3>
      <p>هنگام نام‌گذاری توابع، مهم است که از نام‌های توصیفی استفاده کنید که هدف تابع را مشخص کنند.</p>
      <ul>
        <li>از حروف کوچک استفاده کنید.</li>
        <li>کلمات باید با زیرخط (Underscore) از هم جدا شوند (snake_case).</li>
        <li>از نام‌های خیلی کلی یا خیلی طولانی پرهیز کنید.</li>
        <li>اگر تابع کاری را انجام می‌دهد از «فعل» و اگر مقداری را برمی‌گرداند از «اسم» استفاده کنید.</li>
      </ul>
    </div>
    <div class="card" id="toc2_3_">
      <h3>مثال ساخت یک تابع ساده</h3>
      <p>بیایید این مفاهیم را با ساخت یک تابع ساده که دو عدد را با هم جمع می‌کند، تمرین کنیم:</p>
    </div>
  </div>
</div>

```python
def add_numbers(a, b):
    """Add two numbers and return the result."""
    result = a + b
    return result
```

```python
# حالا می‌توانیم از تابع add_numbers استفاده کنیم
add_numbers(5, 3)
```

<div class="section-card" id="toc3_">
  <h2>۳. فراخوانی تابع (Calling)</h2>
  <div class="container">
  <div class="card">
  <p>برای استفاده از یک تابع تعریف شده، آن را با نامش و سپس پرانتز فراخوانی کنید:</p>
  <pre><code>function_name(arguments)</code></pre>
<p><strong>نام تابع</strong>: نامی که برای تابع انتخاب کرده‌اید (باید دقیقاً مطابق تعریف باشد).</p>
<p><strong>آرگومان‌ها</strong>: مقادیری که به پارامترهای تابع پاس داده می‌شوند.</p>
  </div>
    <div class="card" id="toc3_1_">
      <h3>جریان اجرا (Flow of Execution)</h3>
      <p>هنگامی که یک تابع فراخوانی می‌شود، پایتون دستورات داخل بدنه آن را اجرا کرده و سپس به نقطه‌ای که تابع فراخوانی شده بود بازمی‌گردد. دقت کنید که صرفاً «تعریف» یک تابع باعث اجرای آن نمی‌شود؛ شما حتماً باید آن را «فراخوانی» کنید.</p>
    </div>
  </div>
</div>

```python
def add_numbers(a, b):
    """Add two numbers and return the result."""
    result = a + b
    return result
```

```python
# فراخوانی تابع با آرگومان‌های ۵ و ۳
add_numbers(5, 3)
```

```python
# فراخوانی تابع با آرگومان‌های متفاوت
add_numbers(10, 15)
```

  <h2>۴. پارامترها و آرگومان‌ها</h2>
  <div class="container">
    <div class="card">
       <img src="../../images/parameters-arguments.png" alt="تفاوت پارامتر و آرگومان">
    </div>
    <div class="card">
      <p>یک تفاوت ظریف اما مهم بین پارامتر و آرگومان وجود دارد:</p>
      <ul>
        <li><strong>پارامترها (Parameters)</strong>: متغیرهایی هستند که در هنگام تعریف تابع نوشته می‌شوند و نقش نگهدارنده جایگاه را دارند.</li>
        <li><strong>آرگومان‌ها (Arguments)</strong>: مقادیر واقعی هستند که در هنگام فراخوانی به تابع پاس داده می‌شوند تا جای پارامترها را پر کنند.</li>
      </ul>
    </div>
  </div>
</div>

```python
# تعریف تابع با پارامترهای x و y
def multiply_numbers(x, y):
    """Multiply two numbers and return the result."""
    result = x * y
    return result
```

```python
# فراخوانی تابع با آرگومان‌های ۶ و ۷
multiply_numbers(6, 7)
```

```python
# فراخوانی تابع با استفاده از متغیرها به عنوان آرگومان
number_one = 8
number_two = 10
multiply_numbers(number_one, number_two)
```

**تابع با دستور return**

```python
def calculate_area(width, height):
    """Calculate the area of a rectangle."""
    area = width * height
    return area
```

```python
# استفاده از تابع و ذخیره خروجی آن در یک متغیر
rect_area = calculate_area(10, 5)
print(rect_area)  # خروجی: 50
```

```python
def print_welcome_message(username):
    """Print a welcome message to the user."""
    print(f"Welcome, {username}!")
```

```python
# استفاده از تابع
print_welcome_message("Alice")  # خروجی: !Welcome, Alice
```

```python
result = print_welcome_message("Alice")
print(result)  # خروجی: None
```

        <li><strong>امتیازی (خروجی چندگانه)</strong>: تابعی به نام <code>min_max</code> بنویسید که لیستی از اعداد را بگیرد و همزمان کوچکترین و بزرگترین عدد آن لیست را برگرداند.</li>

```python
# راه‌حل تمرین‌ها
def greet():
    print("Hello, World!")
greet()

def personalize_greeting(name):
    print(f"Hello, {name}!")
personalize_greeting("Alice")

def circle_area(radius):
    pi = 3.14159
    return pi * radius ** 2
area = circle_area(5)
print(f"The area of the circle with radius 5 is {area}.")

def add_numbers(num1, num2):
    return num1 + num2
sum_of_numbers = add_numbers(3, 7)
print(f"The sum of 3 and 7 is {sum_of_numbers}.")

def print_menu():
    menu_items = ["Pizza", "Salad", "Soup"]
    print("Menu:", ", ".join(menu_items))
print_menu()

def min_max(numbers):
    return min(numbers), max(numbers)
min_number, max_number = min_max([2, 3, 5, 7, 11, 2, 10])
print(f"Minimum: {min_number}, Maximum: {max_number}")
```
