<div class="section-card">
  <h2>What is Abstraction?</h2>
  <div class="container">
    <div class="card blue">
      <h3>Concept</h3>
      <p>Abstraction hides complex reality while exposing only the necessary parts. It focuses on <strong>what</strong> an object does instead of <strong>how</strong> it does it.</p>
    </div>
    <div class="card purple">
      <h3>Abstract Base Class (ABC)</h3>
      <p>A class that cannot be instantiated. It defines a template with abstract methods that must be implemented by subclasses.</p>
    </div>
    <div class="card green">
      <h3>Why use `abc`?</h3>
      <p>Python's `abc` module provides infrastructure for defining abstract classes and enforcing interfaces.</p>
    </div>
  </div>
</div>

```python
from abc import ABC, abstractmethod
```

```python
class Shape(ABC):  # Shape is an abstract class
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass

class Rectangle(Shape):  # Rectangle is a subclass of Shape
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

    def perimeter(self):
        return 2 * (self.width + self.height)
```

```python
# Now, you can't create an instance of Shape due to its abstract nature
# shape = Shape()  # This would raise an error
```

```python
# But you can create an instance of Rectangle, which implements all the abstract methods from Shape
rectangle = Rectangle(10, 20)
```

```python
rectangle.area()
```

```python
rectangle.perimeter()
```

<div class="section-card" id="toc1_">
  <h2>1. How to Define Abstract Base Classes</h2>
  <div class="container">
    <div class="card blue">
      <h3>Steps</h3>
      <ol>
        <li><strong>Import:</strong> `from abc import ABC, abstractmethod`</li>
        <li><strong>Inherit:</strong> Class must inherit from `ABC`.</li>
        <li><strong>Decorate:</strong> Use `@abstractmethod` for methods that must be overridden.</li>
      </ol>
    </div>
  </div>
</div>

```python
from abc import ABC, abstractmethod
```

```python
class MyAbstractClass(ABC):
    @abstractmethod
    def an_abstract_method(self):
        pass
```

<div class="section-card" id="toc2_">
  <h2>2. The ABC Module Importance</h2>
  <div class="container">
    <div class="card orange">
      <h3>Key Benefits</h3>
      <ul>
        <li><strong>Contracts:</strong> Ensures consistency across subclasses.</li>
        <li><strong>Safety:</strong> Prevents instantiation of base classes meant only as templates.</li>
        <li><strong>Validation:</strong> Runtime checks ensure all abstract methods are implemented.</li>
      </ul>
    </div>
    <div class="card green">
      <h3>Example: Document Handler</h3>
      <p>Defining a common interface for reading/writing different file formats (PDF, Word, etc).</p>
    </div>
  </div>
</div>

```python
from abc import ABC, abstractmethod

class DocumentHandler(ABC):

    @abstractmethod
    def read(self):
        pass

    @abstractmethod
    def write(self, content):
        pass

    @abstractmethod
    def validate(self):
        pass
```

<div class="section-card" id="toc3_">
  <h2>3. Implementing Interfaces</h2>
  <div class="container">
    <div class="card purple">
      <h3>What is an Interface?</h3>
      <p>An interface defines <strong>what</strong> actions a class should perform, not <strong>how</strong>. In Python, ABCs act as interfaces.</p>
    </div>
    <div class="card red">
      <h3>Example: Payment Processor</h3>
      <p>Defining a generic `PaymentProcessor` interface implemented by Credit Card and PayPal classes.</p>
    </div>
  </div>
</div>

```python
from abc import ABC, abstractmethod
```

```python
class PaymentProcessor(ABC):

    @abstractmethod
    def process_payment(self, amount):
        """
        Processes payment for a specified amount.
        """
        pass

class CreditCardPaymentProcessor(PaymentProcessor):

    def process_payment(self, amount):
        print(f"Processing credit card payment for ${amount}")

class PayPalPaymentProcessor(PaymentProcessor):

    def process_payment(self, amount):
        print(f"Processing PayPal payment for ${amount}")
```

```python
# Usage
# Assuming instantiation and setup done accordingly, we can call:
processor = CreditCardPaymentProcessor()
processor.process_payment(100)

# This ensures that each payment processor implements the required method.
```
