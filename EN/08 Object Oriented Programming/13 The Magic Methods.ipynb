<!-- Introduction -->
<div class="section-card">
  <h2>What are Magic Methods?</h2>
  <div class="container">
    <div class="card blue">
      <h3>The Hooks of Python</h3>
      <p>Magic methods (or special methods) are predefined methods in Python recognized by double underscores (e.g., <code>__init__</code>, <code>__str__</code>). There is no actual "magic" involved; they are simply the hooks that let your custom classes interact with Python's core features.</p>
    </div>
    <div class="card blue">
      <h3>What do they do?</h3>
      <ul>
        <li><strong>Initialization & Destruction:</strong> <code>__init__</code>, <code>__del__</code></li>
        <li><strong>String Representation:</strong> <code>__str__</code>, <code>__repr__</code></li>
        <li><strong>Arithmetic Operations:</strong> <code>__add__</code>, <code>__mul__</code></li>
        <li><strong>Comparisons:</strong> <code>__eq__</code>, <code>__lt__</code></li>
      </ul>
      <p>They are rarely called directly. For example, using <code>len(obj)</code> automatically triggers the <code>__len__</code> method under the hood.</p>
    </div>
  </div>
</div>

  <h2>1. Strings: <code>__str__</code> and <code>__repr__</code></h2>
  <div class="container">
    <div class="card green">
      <h3><code>__str__</code>: For the User</h3>
      <p>Invoked by <code>str(obj)</code> or <code>print(obj)</code>. It should return a readable, user-friendly string.</p>
    </div>
  </div>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"
```

```python
# Creating a Book instance
book = Book("Python Programming", "Jane Doe")
```

```python
print(str(book))  # Explicitly calling __str__ method
```

```python
print(book)  # Implicitly calls __str__
```

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        return f"Book({self.title!r}, {self.author!r})"
```

  <h2>2. Length: <code>__len__</code></h2>
  <div class="container">
    <div class="card orange">
      <h3>Concept</h3>
      <p>Defines how the built-in <code>len()</code> function behaves when applied to your custom objects.</p>
    </div>
  </div>
</div>

```python
class CustomList:
    def __init__(self, items):
        self.items = items

    def __len__(self):
        return len(self.items)
```

```python
my_list = CustomList([1, 2, 3, 4, 5])
```

```python
len(my_list)  # Calls __len__
```

```python
class Inventory:
    def __init__(self):
        self.items = {}

    def add_item(self, item, quantity):
        if item in self.items:
            self.items[item] += quantity
        else:
            self.items[item] = quantity

    def remove_item(self, item, quantity):
        if item in self.items:
            if self.items[item] <= quantity:
                del self.items[item]
            else:
                self.items[item] -= quantity

    def __len__(self):
        return sum(self.items.values())
```

```python
inventory = Inventory()
inventory.add_item('apple', 5)
inventory.add_item('banana', 3)
inventory.add_item('orange', 2)
```

```python
len(inventory)  # Output: 10
```

```python
inventory.remove_item('banana', 2)
```

```python
len(inventory)  # Output: 8
```

<div class="section-card" id="toc3_">
  <h2>3. Arithmetic: <code>__add__</code>, <code>__mul__</code></h2>
  <div class="container">
    <div class="card purple">
      <h3>Arithmetic in Custom Objects</h3>
      <p>Python allows instances of your class to partake in arithmetic operations just like built-in ints or floats. This is done via methods like <code>__add__ (+)</code>, <code>__sub__ (-)</code>, and <code>__mul__ (*)</code>.</p>
    </div>
  </div>
</div>

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        else:
            raise ValueError("Operand must be an instance of Vector")

    def __mul__(self, scalar):
        if isinstance(scalar, (float, int)):
            return Vector(self.x * scalar, self.y * scalar)
        else:
            raise ValueError("Scalar must be a float or an integer")

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"
```

```python
# Vector addition
v1 = Vector(1, 2)
v2 = Vector(3, 4)
```

```python
v3 = v1 + v2
```

```python
v3
```

```python
# Vector scalar multiplication
v4 = v1 * 3
```

```python
v4
```

<div class="section-card" id="toc4_">
  <h2>4. Comparisons: <code>__eq__</code>, <code>__lt__</code></h2>
  <div class="container">
    <div class="card red">
      <h3>Equality & Ordering</h3>
      <p>Instead of comparing memory addresses, you can define exactly what makes two objects "equal" or "greater than" each other.</p>
      <ul>
        <li><code>__eq__</code> (==), <code>__ne__</code> (!=)</li>
        <li><code>__lt__</code> (&lt;), <code>__le__</code> (&lt;=)</li>
        <li><code>__gt__</code> (&gt;), <code>__ge__</code> (&gt;=)</li>
      </ul>
    </div>
  </div>
</div>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __eq__(self, other):
        if not isinstance(other, Book):
            return NotImplemented
        return (self.title == other.title) and (self.author == other.author)
```

```python
book1 = Book("Python Programming", "Jane Doe")
book2 = Book("Python Programming", "Jane Doe")

book1 == book2
```

```python
book1 = Book("Python Programming", "Jane Doe")
book2 = Book("Python Programming", "John Smith")

book1 == book2
```

      <h3>Example: Ordering Vectors</h3>
    </div>

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __lt__(self, other):
        return (self.x ** 2 + self.y ** 2) < (other.x ** 2 + other.y ** 2)

    def __le__(self, other):
        return (self.x ** 2 + self.y ** 2) <= (other.x ** 2 + other.y ** 2)

    def __gt__(self, other):
        return (self.x ** 2 + self.y ** 2) > (other.x ** 2 + other.y ** 2)

    def __ge__(self, other):
        return (self.x ** 2 + self.y ** 2) >= (other.x ** 2 + other.y ** 2)
```

```python
v1 = Vector(1, 2)
v2 = Vector(3, 4)
```

```python
v1 < v2
```

```python
v1 > v2
```

```python
v1 <= v2
```

```python
v1 >= v2
```

<div class="section-card" id="toc6_">
  <h2>6. Containers: <code>__getitem__</code> and <code>__setitem__</code></h2>
  <div class="container">
    <div class="card green">
      <h3><code>__getitem__</code> & <code>__setitem__</code></h3>
      <p>These methods allow your objects to respond to square bracket notation (<code>obj[key]</code>), making them act like lists or dictionaries.</p>
    </div>
  </div>
</div>

```python
class CustomDict:
    def __init__(self):
        self.data = {}

    def __getitem__(self, key):
        return self.data[key]
```

```python
my_dict = CustomDict()
my_dict.data = {'a': 1, 'b': 2}
```

```python
my_dict['a']  # Calls __getitem__
```

```python
class CustomDict:
    def __init__(self):
        self.data = {}

    def __setitem__(self, key, value):
        self.data[key] = value
```

```python
my_dict = CustomDict()
my_dict['a'] = 1  # Calls __setitem__
my_dict['b'] = 2
```

```python
my_dict.data
```

<div class="section-card" id="toc7_">
  <h2>7. Callable: <code>__call__</code></h2>
  <div class="container">
    <div class="card pink">
      <h3>Concept</h3>
      <p>Allows an instance of a class to be called as if it were a function (<code>my_instance()</code>).</p>
    </div>
  </div>
</div>

```python
class Adder:
    def __init__(self, initial_value):
        self.total = initial_value

    def __call__(self, value):
        self.total += value
        return self.total
```

```python
adder = Adder(10)
```

```python
adder(5)  # Calls __call__
```

```python
adder(3)
```

      <h3>Example 2: A Counter Class</h3>
      <p>Useful when an object encapsulates a single action.</p>

```python
class Counter:
    def __init__(self, initial_count=0):
        self.count = initial_count

    def __call__(self, increment=1):
        self.count += increment
        return self.count
```

```python
counter = Counter()
```

```python
counter()
```

```python
counter()
```

```python
counter(5)
```

```python
counter()
```
