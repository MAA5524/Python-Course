<!-- Introduction Section -->
<div class="section-card">
  <h2>Introduction</h2>
  <div class="container">
    <div class="card blue">
      <h3>Inheritance & Polymorphism Refresher</h3>
      <ul>
        <li><strong>Inheritance:</strong> A subclass inherits properties and methods from a superclass.</li>
        <li><strong>Polymorphism:</strong> Objects of different classes can be treated as objects of a common superclass.</li>
      </ul>
    </div>
    <div class="card purple">
      <h3>What is Method Overriding?</h3>
      <p>Method overriding allows a subclass to provide a specific implementation of a method that is already defined in its superclass. It is crucial for achieving runtime polymorphism.</p>
    </div>
  </div>
</div>

```python
class Animal:
    def speak(self):
        print("This animal speaks in a generic way.")

class Dog(Animal):
    def speak(self):
        print("Bark! Bark!")
```

      <p>When you call a method, Python looks it up in the object's immediate class. If not found, it traverses up the superclasses (Method Resolution Order or MRO). Since <code>WatercolorPainter</code> has its own <code>paint</code>, that is executed instead of the parent's.</p>

<div class="section-card" id="toc2_">
  <h2>2. The `super()` Function</h2>
  <div class="container">
    <div class="card red">
      <h3>Extending Without Replacing</h3>
      <p>The <code>super()</code> function allows subclasses to call methods from their superclass. This is extremely useful when you want to <strong>extend</strong> an existing method rather than completely replacing it.</p>
    </div>
    <div class="card red">
      <h3>Example 1: Extending Functionality</h3>
      <p>We override <code>work()</code> in <code>SoftwareEngineer</code>, but we still call <code>super().work()</code> to keep the generic employee workflow.</p>
    </div>
  </div>
</div>

```python
class Employee:
    def work(self):
        print("Completing tasks assigned by the manager.")
```

```python
class SoftwareEngineer(Employee):
    def work(self):
        super().work()  # Call the work method from Employee
        print("Writing code and fixing bugs.")  # Additional behavior for SoftwareEngineer
```

      <h3>Example 2: Using `super()` in `__init__`</h3>

<div class="section-card" id="toc3_">
  <h2>3. Overriding Special Methods</h2>
  <div class="container">
    <div class="card blue">
      <h3>What are Special (Magic) Methods?</h3>
      <p>Methods bounded by double underscores (e.g., <code>__str__</code>, <code>__eq__</code>). Overriding them lets your custom objects integrate seamlessly with Python's built-in operations (+, ==, print, etc.).</p>
    </div>
    <div class="card blue">
      <h3>Overriding <code>__str__</code></h3>
      <p>Provides a meaningful, human-readable string representation (used when you <code>print()</code> the object).</p>
    </div>
  </div>
</div>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f"'{self.title}' by {self.author}"
```

```python
# Creating an instance of Book
book = Book("The Catcher in the Rye", "J.D. Salinger")
```

```python
# Printing the book
print(book)
```

    <div class="card green">
      <h3>Overriding <code>__repr__</code></h3>
      <p>Intended for developers. It produces an unambiguous string that ideally could be used to recreate the object. Great for debugging.</p>
    </div>
  </div>

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __repr__(self):
        return f"Book({self.title!r}, {self.author!r})"
```

```python
# Creating an instance of Book
book = Book("The Catcher in the Rye", "J.D. Salinger")
```

```python
# Using the representation of a book
print(repr(book))  # Output: Book('The Catcher in the Rye', 'J.D. Salinger')
```

  <div class="container">
    <div class="card orange">
      <h3>Overriding <code>__eq__</code></h3>
      <p>Defines custom logic for the equality operator (<code>==</code>). This lets you compare objects based on their actual attribute values rather than memory addresses.</p>
    </div>
  </div>

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y
```

```python
# Creating instances of Point
point1 = Point(1, 2)
point2 = Point(1, 2)
point3 = Point(3, 4)
```

```python
# Comparing points
point1 == point2
```

```python
point1 == point3
```

<div class="section-card" id="toc4_">
  <h2>4. Method Overriding Best Practices</h2>
  <div class="container">
    <div class="card purple">
      <h3>When to Use Overriding?</h3>
      <ul>
        <li><strong>Customizing Behavior:</strong> The subclass needs a specific version of a generic task.</li>
        <li><strong>Implementing Abstract Methods:</strong> Providing concrete logic for templates.</li>
        <li><strong>Enhancing Functionality:</strong> Adding steps before or after the parent method runs.</li>
      </ul>
    </div>
    <div class="card purple">
      <h3>Best Practices</h3>
      <ul>
        <li><strong>Maintain Signature:</strong> Keep the same name, parameters, and return type as the parent method.</li>
        <li><strong>Use <code>super()</code>:</strong> When extending, always call the parent method to preserve base functionality.</li>
      </ul>
    </div>
    <div class="card red">
      <h3>Common Pitfalls ⚠️</h3>
      <ul>
        <li><strong>Forgetting <code>super()</code>:</strong> Leads to broken or incomplete initializations/behaviors.</li>
        <li><strong>Tight Coupling:</strong> Overusing overriding can make the codebase rigid. Consider <em>Composition</em> if inheritance feels forced.</li>
      </ul>
    </div>
  </div>
</div>
