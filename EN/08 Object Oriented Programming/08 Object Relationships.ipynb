<div class="section-card" id="toc1_">
  <h2>1. Association</h2>
  <div class="container">
    <div class="card green">
      <h3>Concept</h3>
      <p>Objects interact but remain independent. No ownership.</p>
      <img src="../../images/association.png" alt="Association">
    </div>
    <div class="card green">
      <h3>Example 1: Doctor & Stethoscope</h3>
      <img src="../../images/doctor.png" alt="Doctor">
      <p>A doctor uses a stethoscope, but both can exist independently.</p>
    </div>
  </div>
</div>

```python
class Stethoscope:
    def __init__(self, brand):
        self.brand = brand

    def use(self):
        print(f"Using {self.brand} stethoscope.")

class Doctor:
    def __init__(self, name):
        self.name = name

    def check_patient(self, stethoscope):
        print(f"{self.name} is checking the patient.")
        stethoscope.use()
```

```python
# Example usage
my_stethoscope = Stethoscope("Littmann")
dr_jones = Doctor("Dr. Jones")
```

```python
dr_jones.check_patient(my_stethoscope)
```

<div class="section-card">
  <div class="container">
    <div class="card green">
      <h3>Example 2: Owner & Dog</h3>
      <p>One owner can have multiple dogs. The dogs are not part of the owner, they just relate to them.</p>
    </div>
  </div>
</div>

```python
class Dog:
    def __init__(self, name):
        self.name = name

    def play(self):
        print(f"{self.name} is playing.")

class Owner:
    def __init__(self, name):
        self.name = name
        self.dogs = []  # This can hold multiple Dog objects

    def add_dog(self, dog):
        self.dogs.append(dog)

    def play_with_dogs(self):
        print(f"{self.name} is playing with their dogs.")
        for dog in self.dogs:
            dog.play()
```

```python
# Example usage
buddy = Dog("Buddy")
rover = Dog("Rover")
jane = Owner("Jane")
```

```python
jane.add_dog(buddy)
jane.add_dog(rover)
```

```python
jane.play_with_dogs()
```

<div class="section-card" id="toc2_">
  <h2>2. Aggregation</h2>
  <div class="container">
    <div class="card orange">
      <h3>Concept</h3>
      <p>A "whole-part" relationship. Weak ownership. The part <strong>can exist independently</strong> of the whole.</p>
      <img src="../../images/aggregation.png" alt="Aggregation">
    </div>
    <div class="card orange">
      <h3>Example: Team & Players</h3>
      <p>A team consists of players. If the team is dissolved, the players still exist.</p>
    </div>
  </div>
</div>

```python
class Player:
    def __init__(self, name):
        self.name = name

class Team:
    def __init__(self, name):
        self.name = name
        self.players = []  # This is where the aggregation relationship is established

    def add_player(self, player):
        self.players.append(player)

    def remove_player(self, player):
        self.players.remove(player)
```

```python
# The existence of Player objects is not tied to the lifecycle of a Team object
player1 = Player("Alex")
team = Team("Sharks")
team.add_player(player1)
# Here, 'player1' can equally be part of another team or none, signifying the independence characteristic of aggregation.
```

<div class="section-card">
  <div class="container">
    <div class="card orange">
      <h3>Example 2: Dog & Collar</h3>
      <p>A dog can wear a collar, but the collar can be removed and used by another dog.</p>
    </div>
  </div>
</div>

```python
class Collar:
    def __init__(self, color):
        self.color = color

class Dog:
    def __init__(self, name):
        self.name = name
        self.collar = None  # Initially, the dog might not have a collar

    def add_collar(self, collar):
        self.collar = collar

    def remove_collar(self):
        self.collar = None
```

```python
# Create a Collar object
blue_collar = Collar("blue")
```

```python
# Create a Dog object and associate it with the Collar
buddy = Dog("Buddy")
buddy.add_collar(blue_collar)
```

```python
# The Dog has a Collar, establishing an aggregation relationship
# The Collar can be removed, emphasizing its independence
buddy.remove_collar()
```

```python
# The Collar still exists and can potentially be used by another Dog
rover = Dog("Rover")
rover.add_collar(blue_collar)
```

<div class="section-card" id="toc3_">
  <h2>3. Composition</h2>
  <div class="container">
    <div class="card red">
      <h3>Concept</h3>
      <p>A strong "whole-part" relationship. Strong ownership. If the whole is destroyed, the parts are destroyed too.</p>
      <img src="../../images/composition.png" alt="Composition">
    </div>
    <div class="card red">
      <h3>Difference from Aggregation</h3>
      <p><strong>Aggregation:</strong> Loose. Shared ownership. (House & Residents)<br>
      <strong>Composition:</strong> Strong. Exclusive ownership. (House & Rooms)</p>
    </div>
    <div class="card red">
      <h3>Example 1: Computer & Parts</h3>
      <p>A processor, memory, and hard drive are internal parts of a specific computer. They are created with it.</p>
    </div>
  </div>
</div>

```python
class Processor:
    def __init__(self, name):
        self.name = name
```

```python
class Memory:
    def __init__(self, size):
        self.size = size
```

```python
class HardDrive:
    def __init__(self, capacity):
        self.capacity = capacity
```

```python
class Computer:
    def __init__(self):
        self.processor = Processor("Intel i7")
        self.memory = Memory(16)  # in GB
        self.hard_drive = HardDrive(1)  # in TB

    def describe(self):
        print(f"Computer with {self.processor.name} processor, {self.memory.size}GB RAM, and {self.hard_drive.capacity}TB storage.")
```

```python
# When a Computer instance is created, the Processor, Memory, and HardDrive instances are also created, and their lifecycle is tied to the Computer.
my_computer = Computer()
my_computer.describe()
# Deleting 'my_computer' will also 'delete' its Processor, Memory, and HardDrive, demonstrating composition's strong whole-part relationship.
```

  <div class="container">
    <div class="card red">
      <h3>Example 2: Dog & Tail</h3>
      <p>A tail is an integral part of a dog. It doesn't exist meaningfully on its own.</p>
    </div>
  </div>
</div>

```python
class Tail:
    def __init__(self, length):
        self.length = length  # Length in cm

    def wag(self):
        print("The tail is wagging.")
```

```python
class Dog:
    def __init__(self, name):
        self.name = name
        self.tail = Tail(20)  # Assuming every dog has a tail of length 20 cm

    def wag_tail(self):
        print(f"{self.name} is happy!")
        self.tail.wag()
```

```python
# When a Dog instance is created, a Tail instance is also created as an integral part of the Dog.
buddy = Dog("Buddy")
buddy.wag_tail()
# Here, 'buddy's tail is intrinsically linked to him, exemplifying composition's inseparable whole-part dynamic.
```

<div class="section-card" id="toc4_">
  <h2>4. Inheritance</h2>
  <div class="container">
    <div class="card purple">
      <h3>Concept</h3>
      <p>"Is-a" relationship. A new class derives from an existing one, inheriting properties and behaviors.</p>
      <img src="../../images/inheritance.png" alt="Inheritance">
    </div>
    <div class="card purple">
      <h3>Types</h3>
      <ul>
        <li><strong>Single:</strong> One parent.</li>
        <li><strong>Multiple:</strong> Multiple parents.</li>
        <li><strong>Multilevel:</strong> Grandparent -> Parent -> Child.</li>
        <li><strong>Hierarchical:</strong> One parent, many children.</li>
      </ul>
      <img src="../../images/inheritance-types.png" alt="Inheritance Types">
    </div>
    <div class="card purple">
      <h3>Example: Pet & Dog</h3>
      <p>A Dog <strong>is a</strong> Pet.</p>
    </div>
  </div>
</div>

```python
class Pet:
    def __init__(self, name):
        self.name = name

    def show_affection(self):
        print(f"{self.name} shows affection.")

class Dog(Pet):  # Dog inherits from Pet
    def __init__(self, name, breed):
        super().__init__(name)  # Call the constructor of the base class
        self.breed = breed

    def bark(self):
        print(f"{self.name}, the {self.breed}, barks.")
```

```python
# Usage
my_pet = Pet("Buddy")
my_pet.show_affection()
```

```python
my_dog = Dog("Max", "Golden Retriever")
```

```python
my_dog.show_affection()  # Inherited method
```

```python
my_dog.bark()  # Dog-specific method
```

  <h2>5. Practical Guide: When to use what?</h2>
  <div class="container">
    <div class="card blue">
      <h3>Cheat Sheet</h3>
      <ul>
        <li><strong>Inheritance:</strong> Use for "Is-a". (Car is a Vehicle)</li>
        <li><strong>Association:</strong> Use for "Uses-a". (Customer uses PaymentMethod)</li>
        <li><strong>Aggregation:</strong> Use for "Has-a" (independent). (Airline has Airplanes)</li>
        <li><strong>Composition:</strong> Use for "Has-a" (dependent). (Book has Pages)</li>
      </ul>
    </div>
    <div class="card green">
      <h3>Best Practices</h3>
      <ul>
        <li>Prefer <strong>Composition over Inheritance</strong> for flexibility.</li>
        <li>Keep coupling low (objects should be independent).</li>
        <li>Avoid deep inheritance hierarchies.</li>
      </ul>
    </div>

      <ol>
        <li><strong>Association:</strong> Zookeeper & Tools.</li>
        <li><strong>Aggregation:</strong> Habitat & Animals.</li>
        <li><strong>Composition:</strong> Zoo & Habitats.</li>
        <li><strong>Inheritance:</strong> Species & Animals.</li>
      </ol>

      <h3>Solution</h3>
      <p>Below is a complete Python solution demonstrating all these relationships.</p>

```python
# Task 1: Association - Zookeeper and Tools
class Tool:
    def __init__(self, name):
        self.name = name

    def use(self):
        print(f"Using the tool: {self.name}")

class Zookeeper:
    def __init__(self, name):
        self.name = name

    def use_tool(self, tool):
        print(f"{self.name} is now ")
        tool.use()

# Task 2: Aggregation - Habitat and Animals
class Animal:
    def __init__(self, name):
        self.name = name

class Habitat:
    def __init__(self):
        self.animals = []

    def add_animal(self, animal):
        self.animals.append(animal)

    def remove_animal(self, animal):
        self.animals.remove(animal)

# Task 3: Composition - Zoo and Habitats
class Zoo:
    def __init__(self):
        self.habitats = []

    def add_habitat(self, habitat):
        self.habitats.append(habitat)

    def __del__(self):
        print("Closing the zoo and removing all habitats.")
        del self.habitats

# Task 4: Inheritance - Species and Animals
class Lion(Animal):
    def roar(self):
        print(f"{self.name} roars loudly!")

class Penguin(Animal):
    def swim(self):
        print(f"{self.name} swims gracefully.")

# Task 5: Practical Application - Decision Making
# Association is chosen because a care record should not solely exist within the scope of an animal or a zookeeper
class VeterinaryCareRecord:
    def __init__(self, animal, description):
        self.animal = animal
        self.description = description

# Demonstration
broom = Tool("broom")
zookeeper_jane = Zookeeper("Jane")
zookeeper_jane.use_tool(broom)

lion_habitat = Habitat()
leo = Animal("Leo the Lion")
lion_habitat.add_animal(leo)

alex = Lion("Alex")
marty = Penguin("Marty")
alex.roar()
marty.swim()

leo_care_record = VeterinaryCareRecord(leo, "Annual vaccination.")
print(f"{leo_care_record.animal.name} has a care record for: {leo_care_record.description}")
```
