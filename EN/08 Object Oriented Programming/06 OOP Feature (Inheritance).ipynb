        <li><strong>Extensibility:</strong> Easily add features to subclasses.</li>
        <li><strong>Hierarchy:</strong> Organizes code logically (e.g., Vehicle -> Car, Bike).</li>

<div class="section-card" id="toc1_">
  <h2>1. The Basics of Inheriting Classes</h2>
  <div class="container">
    <div class="card green">
      <h3>How to Inherit?</h3>
      <p>Define a new class and put the parent class name in parentheses: <code>class Child(Parent):</code></p>
    </div>
  </div>
</div>

```python
class Vehicle:  # This is the base class
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.make} {self.model}")
```

```python
# Inheriting from Vehicle
class Car(Vehicle):  # Car is the derived class that inherits from Vehicle
    pass
```

```python
# Creating an instance of Car
my_car = Car("Toyota", "Corolla", 2021)
```

```python
my_car.display_info()  # Outputs: 2021 Toyota Corolla
```

<div class="section-card">
  <div class="container">
    <div class="card green">
      <p><code>Car</code> automatically has access to methods and attributes of <code>Vehicle</code>.</p>
    </div>
    <div class="card green">
      <h3>Adding & Overriding Methods</h3>
      <p>Subclasses can define their own methods or override inherited ones to change behavior.</p>
    </div>
  </div>
</div>

```python
class Car(Vehicle):
    def __init__(self, make, model, year, trunk_capacity):
        super().__init__(make, model, year)  # Calls the __init__ of Vehicle
        self.trunk_capacity = trunk_capacity  # New attribute specific to Car

    def display_car_info(self):  # New method specific to Car
        print(f"{self.year} {self.make} {self.model} with trunk capacity: {self.trunk_capacity} liters")

    # Overriding an inherited method
    def display_info(self):
        print(f"This is a {self.year} {self.make} {self.model}.")
```

```python
# Creating an instance of Car and using its methods
my_car = Car("Honda", "Civic", 2020, 450)
```

```python
my_car.display_car_info()  # Outputs: 2020 Honda Civic with trunk capacity: 450 liters
```

```python
my_car.display_info()  # Outputs: This is a 2020 Honda Civic.
```

<div class="section-card" id="toc2_">
  <h2>2. The <code>super()</code> Function & Method Overriding</h2>
  <div class="container">
    <div class="card purple">
      <h3>What is <code>super()</code>?</h3>
      <p>The <code>super()</code> function allows access to methods of a parent class. It is crucial for extending functionality rather than just replacing it.</p>
    </div>
    <div class="card purple">
      <h3>Method Overriding</h3>
      <p>When a subclass provides a specific implementation of a method that is already defined in its parent class.</p>
    </div>
  </div>
</div>

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

class Car(Vehicle):
    def __init__(self, make, model, trunk_capacity):
        super().__init__(make, model)  # Reuse parent logic
        self.trunk_capacity = trunk_capacity
```

<div class="section-card">
  <div class="container">
    <div class="card purple">
      <h3>Extending Behavior</h3>
      <p>We can also call the parent method inside the overridden method to add to it, instead of replacing it entirely.</p>
    </div>
  </div>
</div>

```python
class Vehicle:
    def display_info(self):
        print("This is a Vehicle")

class Car(Vehicle):
    def display_info(self):
        super().display_info()  # Calls display_info from Vehicle
        print("and more specifically, this is a Car")
```

  <h2>3. Exercise: Employee Hierarchy</h2>
  <div class="container">
    <div class="card green">
      <h3>Your Task</h3>
      <ol>
        <li><strong>Define <code>Employee</code> (Base):</strong> With <code>name</code>, <code>email</code> and <code>get_info()</code>.</li>
        <li><strong>Define <code>Manager</code> (Subclass):</strong> Inherit from <code>Employee</code>, add <code>department</code>, and override methods using <code>super()</code>.</li>
        <li><strong>Test:</strong> Create instances and verify output.</li>
      </ol>
    </div>
    <div class="card green">
      <h3>Solution</h3>
      <p>Here is the implementation using <code>super()</code> to handle initialization and method extension.</p>
    </div>
  </div>
</div>

```python
# Define the base class 'Employee'
class Employee:
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def get_info(self):
        print(f"Employee Name: {self.name}")
        print(f"Employee Email: {self.email}")
```

```python
# Define the subclass 'Manager' which inherits from 'Employee'
class Manager(Employee):
    def __init__(self, name, email, department):
        super().__init__(name, email) # Use super() to call the __init__ method of the parent class
        self.department = department

    def get_info(self):
        # Override the get_info() method to include department information
        super().get_info() # Call the get_info() method from the parent class
        print(f"Department: {self.department}")
```

```python
# Instantiate and Test Classes

# Create an instance of 'Employee'
alice = Employee("Alice Smith", "alice@example.com")
# Call get_info() method to display 'Employee' information
alice.get_info()
```

```python
# Create an instance of 'Manager'
bob = Manager("Bob Johnson", "bob@example.com", "Engineering")
# Call get_info() method to display 'Manager' information
bob.get_info()
```
