  <h2>Object Initialization</h2>
  <div class="container">
    <div class="card blue">
      <h3>What is it?</h3>
      <p>Object initialization is the process where an object obtains its initial state from its class blueprint. This stage sets the foundation for how the object will behave.</p>
    </div>
    <div class="card blue">
      <h3>The Role of <code>__init__</code></h3>
      <p>The <code>__init__</code> method is automatically called when a new object is created. It serves as the <strong>constructor</strong>, setting initial values for attributes.</p>
    </div>
  </div>
</div>

```python
class Car:
    def __init__(self, color, model):
        self.color = color
        self.model = model
```

  <div class="container">
    <div class="card green">
      <h3>Best Practice: Validation</h3>
      <p>Keep <code>__init__</code> minimal. If you need complex validation, delegate it to a method (like <code>set_price</code>).</p>
    </div>
  </div>
</div>

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.set_price(price)

    def set_price(self, price):
        if price < 0:
            raise ValueError("Price cannot be negative")
        self.__price = price
```

```python
# Task 1: Define the Classroom Class
class Classroom:
    def __init__(self, teacher, students, subject, room_number):
        self.teacher = teacher
        self.students = students
        self.subject = subject
        self.room_number = room_number

    def display_info(self):
        print(f"Teacher: {self.teacher}")
        print(f"Students: {self.students}")
        print(f"Subject: {self.subject}")
        print(f"Room Number: {self.room_number}")
```

```python
# Task 2: Instantiate the Classroom
my_class = Classroom(
    teacher="Mr. Smith",
    students=["Alice", "Bob", "Charlie", "Diana"],
    subject="Mathematics",
    room_number=101
)
```

```python
# Task 3: Print Object Attributes
my_class.display_info()
```

```python
# Bonus: Dynamic Student List
class ClassroomWithDynamicStudents:
    def __init__(self, teacher, *students, subject, room_number):
        self.teacher = teacher
        self.students = list(students)
        self.subject = subject
        self.room_number = room_number

    def display_info(self):
        print(f"Teacher: {self.teacher}")
        print(f"Students: {self.students}")
        print(f"Subject: {self.subject}")
        print(f"Room Number: {self.room_number}")
```

```python
# Bonus Task: Test Dynamic Student List
bonus_class = ClassroomWithDynamicStudents(
    "Ms. Johnson",
    "Eva", "Frank", "Grace", "Henry",
    subject="Science",
    room_number=102
)
```

```python
bonus_class.display_info()
```
