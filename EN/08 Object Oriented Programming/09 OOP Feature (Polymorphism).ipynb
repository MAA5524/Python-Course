<div class="section-card">
  <h2>What is Polymorphism?</h2>
  <div class="container">
    <div class="card blue">
      <h3>Concept</h3>
      <p>From Greek "poly" (many) and "morph" (form). It allows objects of different classes to be treated as objects of a common superclass.</p>
    </div>
    <div class="card purple">
      <h3>Two Key Aspects</h3>
      <ul>
        <li><strong>Method Overriding:</strong> A subclass provides a specific implementation of a method defined in its superclass.</li>
        <li><strong>Duck Typing:</strong> "If it walks like a duck and quacks like a duck, it's a duck." Python focuses on behavior, not type.</li>
      </ul>
    </div>
    <div class="card blue">
      <h3>Example: Vehicles</h3>
      <p>Imagine a base <code>Vehicle</code> class with a <code>display_info()</code> method. <code>Car</code> and <code>Motorcycle</code> subclasses can implement this method differently.</p>
    </div>
  </div>
</div>

```python
class Vehicle:
    def display_info(self):
        print("This is a vehicle")

class Car(Vehicle):
    def display_info(self):
        print("This is a car")

class Motorcycle(Vehicle):
    def display_info(self):
        print("This is a motorcycle")
```

```python
# Function that takes any vehicle and calls its display_info method
def vehicle_info(vehicle):
    vehicle.display_info()
```

```python
# Create instances
car = Car()
motorcycle = Motorcycle()
```

```python
# Demonstrating polymorphism
vehicle_info(car)
```

```python
vehicle_info(motorcycle)
```

<div class="section-card" id="toc1_">
  <h2>1. Implementing Polymorphism via Inheritance</h2>
  <div class="container">
    <div class="card green">
      <h3>Method Overriding</h3>
      <p>Subclasses can provide a specific implementation of a method already defined in their parent class.</p>
    </div>
    <div class="card green">
      <h3>Example: Animal Sounds</h3>
      <p>A base <code>Animal</code> class with a generic <code>speak()</code> method, overridden by <code>Dog</code> and <code>Cat</code>.</p>
    </div>
  </div>
</div>

```python
class Animal:
    def speak(self):
        return "This animal does not have a sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

class Cat(Animal):
    def speak(self):
        return "Meow"
```

```python
# Instantiate objects
dog = Dog()
cat = Cat()
```

```python
# Utilizing polymorphism
dog.speak()
```

```python
cat.speak()
```

<div class="section-card" id="toc2_">
  <h2>2. Dynamic Typing and Polymorphic Behavior</h2>
  <div class="container">
    <div class="card orange">
      <h3>Dynamic Typing</h3>
      <p>Python doesn't require explicit type declarations. A variable can hold an integer now and a string later. This enables flexible polymorphism.</p>
    </div>
    <div class="card orange">
      <h3>Example</h3>
      <p>A single function <code>fetch_item</code> works on lists, dictionaries, and strings because they all support index access.</p>
    </div>
  </div>
</div>

```python
def fetch_item(collection, position):
    return collection[position]
```

```python
# This function can operate on various types, thanks to dynamic typing
fetch_item([1, 2, 3], 1)
```

```python
fetch_item({'a': 1, 'b': 2}, 'b')
```

```python
fetch_item("Python", 3)
```

<div class="section-card" id="toc3_">
  <h2>3. Exercise: Document Handling</h2>
  <div class="container">
    <div class="card blue">
      <h3>Your Task</h3>
      <ol>
        <li><strong>Base Class:</strong> <code>Document</code> with <code>display()</code>.</li>
        <li><strong>Subclasses:</strong> <code>PDF</code>, <code>Word</code>, <code>Spreadsheet</code> overriding <code>display()</code>.</li>
        <li><strong>Polymorphism:</strong> A function <code>display_document()</code> that works on any document.</li>
        <li><strong>Dynamic Typing:</strong> A function to change a document's type at runtime.</li>
      </ol>
    </div>
    <div class="card green">
      <h3>Solution</h3>
      <p>Here is the implementation demonstrating polymorphism and dynamic typing.</p>
    </div>
  </div>
</div>

```python
# Superclass
class Document:
    def display(self):
        print("Displaying a document.")

# Subclasses
class PDF(Document):
    def display(self):
        print("Displaying PDF document.")

class Word(Document):
    def display(self):
        print("Displaying Word document.")

class Spreadsheet(Document):
    def display(self):
        print("Displaying Spreadsheet document.")

# Task 2: Demonstrate Polymorphism
def display_document(document):
    document.display()

# Create instances of each document type
pdf = PDF()
word = Word()
spreadsheet = Spreadsheet()

# Store the document instances in a list
documents = [pdf, word, spreadsheet]

# Loop through the list and display each document
for document in documents:
    display_document(document)

# Task 3: Explore Dynamic Typing
def change_document_type(document, new_type):
    # Create a new instance of the specified type
    return new_type()

# Initially, create a Word document
current_document = Word()
current_document.display()  # Should print "Displaying Word document."

# Change the document type to PDF
current_document = change_document_type(current_document, PDF)
current_document.display()  # Should print "Displaying PDF document."
```
