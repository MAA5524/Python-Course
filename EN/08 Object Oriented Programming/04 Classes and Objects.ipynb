  <div class="container">
    <div class="card blue">
      <h3>Classes and Objects</h3>
      <p>At the heart of Object-Oriented Programming (OOP) in Python lie two fundamental concepts: classes and objects. Grasping these concepts is essential for understanding how OOP facilitates the creation and management of complex software systems.</p>
    </div>
    <div class="card blue">
      <h3>Classes as Blueprints</h3>
      <p>Imagine you are an architect. Before a building is constructed, you draft a detailed blueprint. In OOP, a <strong>class</strong> serves as this blueprint or template. It outlines a specific set of attributes (data) and methods (functions) that objects based on this blueprint will possess.</p>
    </div>
    <div class="card blue">
      <h3>Objects as Instances</h3>
      <p>If a class is the blueprint, an <strong>object</strong> is the actual building. An object is a specific instance of a class; it embodies the structured data and behaviors defined by its class but with concrete values.</p>
      <img src="../../images/class-objects.png" alt="Class vs Objects">
    </div>
    <div class="card purple">
      <h3>Relationship between Classes and Objects</h3>
      <p>Classes create a framework under which individual objects operate. When an object is instantiated, it fills out the framework provided by its class with actual, specific data. This allows for flexibility and reusability.</p>
    </div>
  </div>
</div>

<div class="section-card" id="toc2_">
  <h2>2. Instance Methods and the Use of <code>self</code></h2>
  <div class="container">
    <div class="card purple">
      <h3>What is an Instance Method?</h3>
      <p>Instance methods define the behaviors of objects. They interact with the object's data (attributes). The <code>self</code> parameter is crucial here, acting as a reference to the <strong>current instance</strong> of the class.</p>
    </div>
    <div class="card purple">
      <h3>Example</h3>
      <p>Let's define a <code>Car</code> class with a <code>honk</code> instance method.</p>
    </div>
  </div>
</div>

```python
class Car:
    def honk(self):
        return "Honk! Honk!"
```

<div class="section-card">
  <div class="container">
    <div class="card purple">
      <h3>Practical Example with <code>self</code></h3>
      <p>Let's expand <code>Car</code> to include a <code>color</code> attribute. Notice how <code>self.color</code> allows the method to access the data specific to that instance.</p>
    </div>
  </div>
</div>

```python
class Car:
    def __init__(self, color):
        self.color = color

    def honk(self):
        return f"The {self.color} car says Honk! Honk!"
```

```python
blue_car = Car("blue")
blue_car.honk()
```

      <h3>Your Task</h3>
      <p>Define a <code>Book</code> class to apply what you've learned.</p>
      <ul>
        <li><strong>1. Define:</strong> Attributes for <code>title</code>, <code>author</code>, <code>pages</code>.</li>
        <li><strong>2. Method:</strong> <code>description()</code> to print details.</li>
        <li><strong>3. Init:</strong> Use <code>__init__</code> to set attributes.</li>
        <li><strong>4. Class Attr:</strong> Add <code>book_count</code> to track total books.</li>
        <li><strong>5. Logic:</strong> Add <code>is_long()</code> (True if > 500 pages).</li>
      </ul>
    </div>
    <div class="card green">
      <h3>Solution</h3>
      <p>Here is the complete solution implementing all the tasks above.</p>
    </div>
  </div>

```python
# Task 1: Define a Simple `Book` Class
class Book:
    # Task 5: Class attribute to keep track of the total number of book instances
    book_count = 0

    # Task 4: Incorporate the `__init__` Method
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        Book.book_count += 1  # Increment the book_count class attribute

    # Instance method to print a description of the book
    def description(self):
        print(f"Title: {self.title}, Author: {self.author}, Pages: {self.pages}")

    # Task 7: Add a method `is_long`
    def is_long(self):
        return self.pages > 500
```

```python
# Task 2: Instantiate an Object from the `Book` Class
python_book = Book("Python Programming", "John Doe", 350)

# Call the `description` method on this instance
python_book.description()
```

```python
# Task 3: Use of `self`
# The `self` keyword is used to refer to the instance upon which the method is
# being called. It allows access to the instance's attributes and methods.
# Without `self`, the method would not have access to the instance's `title`,
# `author`, and `pages` attributes.
```

```python
# Task 6: Instantiate Multiple Objects
book1 = Book("Learning Python", "Mark Lutz", 624)
book2 = Book("Automate the Boring Stuff", "Al Sweigart", 504)

# Print out the book_count to show the total number of books created
print(f"Total number of books created: {Book.book_count}")
```

```python
# Task 7: Use the `is_long` method to check if the book is considered long
print(f"This book is considered long: {book1.is_long()}")
```
