<div class="section-card">
  <h2>What is Encapsulation?</h2>
  <div class="container">
    <div class="card blue">
      <h3>The Concept</h3>
      <p>Encapsulation bundles data (attributes) and methods into a single unit (class) and controls access to the internal state. It's about hiding complexity and protecting data integrity.</p>
    </div>
    <div class="card purple">
      <h3>Python's Approach</h3>
      <p>Unlike Java/C++ (strict <code>private</code>, <code>public</code>), Python uses conventions: "We are all consenting adults here."</p>
      <ul>
        <li><strong>Public:</strong> Accessible everywhere.</li>
        <li><strong>Protected (`_`):</strong> Internal use convention.</li>
        <li><strong>Private (`__`):</strong> Name mangling to prevent accidental access.</li>
      </ul>
      <img src="../../images/access-modification-python.png" alt="Access Modifiers">
    </div>
  </div>
</div>

<div class="section-card" id="toc2_">
  <h2>2. Public Attributes and Methods</h2>
  <div class="container">
    <div class="card orange">
      <h3>Open Access</h3>
      <p>By default, everything in Python is public. No special syntax needed.</p>
    </div>
  </div>

```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make  # Public attribute
        self.model = model  # Public attribute

    def display_info(self):  # Public method
        print(f"This vehicle is a {self.make} {self.model}.")
```

```python
# Creating an instance of Vehicle
my_car = Vehicle("Toyota", "Camry")
```

```python
# Accessing public attributes directly
my_car.make
```

```python
my_car.model
```

```python
# Calling a public method
my_car.display_info()
```

<div class="section-card" id="toc3_">
  <h2>3. Protected Attributes and Methods</h2>
  <div class="container">
    <div class="card purple">
      <h3>The Single Underscore (`_`)</h3>
      <p>A convention indicating "Internal Use Only". Python doesn't enforce it, but developers respect it.</p>
    </div>
  </div>
</div>

```python
class Vehicle:
    def __init__(self, make, fuel_capacity):
        self.make = make  # Public
        self._fuel_capacity = fuel_capacity  # Protected
        self._fuel_level = fuel_capacity  # Protected

    def _consume_fuel(self, amount):  # Protected method
        self._fuel_level -= amount
        if self._fuel_level < 0:
            self._fuel_level = 0
        print(f"Fuel left in {self.make}: {self._fuel_level} liters")
```

```python
class Sedan(Vehicle):
    def drive(self, distance):
        print(f"Driving the sedan for {distance} miles.")
        fuel_consumption_rate = 0.04
        self._consume_fuel(distance * fuel_consumption_rate)
```

```python
class SUV(Vehicle):
    def drive(self, distance):
        print(f"Driving the SUV for {distance} miles.")
        fuel_consumption_rate = 0.07
        self._consume_fuel(distance * fuel_consumption_rate)
```

```python
# Creating instances
toyota_camry = Sedan("Toyota Camry", 60)
ford_explorer = SUV("Ford Explorer", 70)
```

```python
# Driving the cars
toyota_camry.drive(200)
```

```python
ford_explorer.drive(200)
```

<div class="section-card" id="toc4_">
  <h2>4. Private Attributes and Methods</h2>
  <div class="container">
    <div class="card red">
      <h3>The Double Underscore (`__`)</h3>
      <p>True encapsulation. Python performs <strong>Name Mangling</strong> (`_ClassName__attribute`) to make access difficult from outside.</p>
    </div>
  </div>
</div>

```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner             # Public attribute
        self.__balance = balance       # Private attribute

    def __update_balance(self, amount):  # Private method
        self.__balance += amount
        return self.__balance

    def deposit(self, amount):
        if amount > 0:
            new_balance = self.__update_balance(amount)
            print(f"Deposit successful. New balance: {new_balance}")
        else:
            print("Deposit amount must be positive.")
```

```python
# Creating an instance of Account
my_account = Account("John Doe", 1000)
```

```python
# Trying to access private members from outside the class
# my_account.__balance  # This will raise an AttributeError
```

```python
my_account.deposit(500)  # This works as intended
```

<div class="section-card" id="toc6_">
  <h2>5. Practical Example: Blog Post System</h2>
  <div class="container">
    <div class="card blue">
      <h3>Scenario</h3>
      <p>A <code>BlogPost</code> with public title, protected content, and private status (draft/published).</p>
    </div>
  </div>
</div>

```python
class BlogPost:
    def __init__(self, title, content):
        self.title = title  # Public attribute
        self._content = content  # Protected attribute
        self.__status = "draft"  # Private attribute

    def edit_content(self, new_content):  # Public method
        self._content = new_content
        print("The content has been updated.")

    def __publish(self):  # Private method
        self.__status = "published"
        print("The post has been published.")

    def make_public(self):  # Public method
        if self.__can_publish():
            self.__publish()
        else:
            print("Cannot publish the post.")

    def __can_publish(self):  # Private method
        return len(self._content) > 100  # Example condition for publishing
```

```python
# Creating an instance of BlogPost
my_post = BlogPost("My First Post", "This is the initial content of the post.")
```
