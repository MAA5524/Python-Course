<div class="section-card" id="toc_intro">
  <h2>1. What is OOP? (vs Procedural)</h2>
  <div class="container">
    <div class="card orange">
      <h3>The OOP Paradigm</h3>
      <p>Object-Oriented Programming (OOP) is a programming paradigm based on the concept of <strong>"objects"</strong>, which can contain data (attributes/properties) and code (methods/procedures).</p>
      <p>It aims to implement real-world entities like inheritance and polymorphism, prioritizing the objects we want to manipulate rather than the logic required to manipulate them.</p>
      <p><em>History:</em> Traced back to the 1960s (Simula) and 1970s (Smalltalk). Today, it dominates languages like Python, Java, and C++.</p>
    </div>
    <div class="card orange">
      <h3>OOP vs Procedural Programming</h3>
      <p><strong>Procedural Programming</strong> focuses on calling procedures/routines (functions) and separates data from procedures.</p>
      <ul>
        <li><strong>Approach:</strong> OOP encapsulates data and methods together. Procedural separates them into sequences of instructions.</li>
        <li><strong>Organization:</strong> OOP organizes around data/objects. Procedural organizes around functions/logic.</li>
        <li><strong>Reusability:</strong> OOP promotes reusability through inheritance. Procedural is harder to maintain as changes affect the entire code.</li>
      </ul>
    </div>
    <div class="card orange">
      <h3>Benefits of OOP Over Procedural</h3>
      <ul>
        <li><strong>Modularity:</strong> Everything is an object, making code easier to understand and maintain.</li>
        <li><strong>Reusability:</strong> Mechanisms like inheritance reduce redundancy.</li>
        <li><strong>Scalability:</strong> Perfect for large, complex applications.</li>
        <li><strong>Security:</strong> Encapsulation hides data, enforcing access restrictions.</li>
      </ul>
    </div>
  </div>
</div>

<!-- Core Concepts Section -->
<div class="section-card" id="toc1_">
  <h2>2. Core Concepts of OOP</h2>
  <div class="container">
    <div class="card purple" id="toc1_1_">
      <h3>Classes and Objects</h3>
      <ul>
        <li><strong>Classes:</strong> Templates or blueprints for creating objects. They encapsulate data for the object.</li>
        <li><strong>Objects:</strong> Instances of classes. They embody the attributes and methods defined by their class.</li>
      </ul>
      <p><strong>Example:</strong> A <code>Dog</code> class defines attributes (age, breed) and methods (bark). An object is a specific dog (Fido, 5 years old, Labrador).</p>
      <p><strong>Real-world Analogy:</strong> Think of a class as a cookie cutter, and objects as the cookies.</p>
      <div style="display: flex; gap: 10px; justify-content: center; flex-wrap: wrap;">
          <img src="../../images/class-objects.png" alt="Class and Objects" style="max-width: 45%; margin: 10px 0;">
      </div>
    </div>
    <div class="card purple" id="toc1_2_">
      <h3>Attributes and Methods</h3>
      <ul>
        <li><strong>Attributes (Properties):</strong> Characteristics of an object, storing data relevant to the object.</li>
        <li><strong>Methods (Functions):</strong> Behaviors or actions that an object can perform.</li>
      </ul>
      <p><strong>Example:</strong> In a <code>Car</code> class, attributes = color, model. Methods = drive(), honk().</p>
      <img src="../../images/object.png" alt="Attributes and Methods" style="max-width: 60%;">
    </div>
    <div class="card purple" id="toc1_3_">
      <h3>Encapsulation</h3>
      <p>Bundling data and methods into a single unit (class) and restricting access to some components.</p>
      <ul>
        <li><strong>Example:</strong> A <code>BankAccount</code> class hides the balance from direct access. You must use <code>deposit()</code> or <code>withdraw()</code> methods to alter it safely.</li>
        <li><strong>Importance:</strong> Ensures data hiding and integrity using access modifiers (private, public, protected).</li>
      </ul>
    </div>
    <div class="card purple" id="toc1_4_">
      <h3>Inheritance</h3>
      <p>Allows a derived class (child) to acquire properties and methods from a base class (parent), promoting code reusability.</p>
      <ul>
        <li><strong>Single:</strong> One child inherits from one parent.</li>
        <li><strong>Multiple:</strong> Inheriting from multiple parents.</li>
        <li><strong>Multilevel:</strong> A class inherits from a child class (Parent -> Child -> Grandchild).</li>
      </ul>
    </div>
    <div class="card purple" id="toc1_5_">
      <h3>Polymorphism</h3>
      <p>Means "many forms". Allows objects of different classes to be treated as objects of a common superclass.</p>
      <ul>
        <li><strong>Static (Compile-time):</strong> Method overloading (same name, different parameters).</li>
        <li><strong>Dynamic (Run-time):</strong> Method overriding (child overrides parent's method).</li>
        <li><strong>Example:</strong> A <code>draw()</code> function manifests differently depending on whether the object is a Circle, Square, or Triangle.</li>
      </ul>
    </div>
    <div class="card purple" id="toc1_6_">
      <h3>Abstraction</h3>
      <p>Hides complex implementation details and exposes only the necessary parts. Focuses on <em>what</em> an object does rather than <em>how</em> it does it.</p>
      <ul>
        <li><strong>Vs Encapsulation:</strong> Abstraction hides complexity; Encapsulation hides internal state.</li>
        <li><strong>Implementation:</strong> Done via Interfaces and Abstract Classes (which cannot be instantiated directly).</li>
      </ul>
    </div>
  </div>
</div>

<!-- Pros and Cons Section -->
<div class="section-card" id="toc2_">
  <h2>3. Advantages & Disadvantages of OOP</h2>
  <div class="container">
    <div class="card green">
      <h3>✅ Advantages</h3>
      <ul>
        <li><strong>Troubleshooting:</strong> Modularity allows you to focus on a specific object when a bug occurs.</li>
        <li><strong>Code Reuse:</strong> Inheritance saves time by avoiding repetitive code.</li>
        <li><strong>Flexibility:</strong> Polymorphism makes the system easily extendable.</li>
        <li><strong>Safety:</strong> Encapsulation prevents accidental data modification.</li>
        <li><strong>Simplicity:</strong> Abstraction models complex realities efficiently.</li>
      </ul>
    </div>
    <div class="card red">
      <h3>❌ Disadvantages</h3>
      <ul>
        <li><strong>Steep Learning Curve:</strong> Harder to grasp initially than procedural programming.</li>
        <li><strong>Slower Execution:</strong> The overhead of instantiating objects can reduce performance.</li>
        <li><strong>Larger Program Size:</strong> More lines of code and memory usage.</li>
        <li><strong>Not for Everything:</strong> Overkill for simple math scripts or tiny utility tasks.</li>
      </ul>
    </div>
    <div class="card blue">
      <h3>When to use it?</h3>
      <div style="display: grid; grid-template-columns: 1fr 1fr; gap: 15px; margin-top: 10px;">
        <div>
          <p><strong>🟢 Best Fit For:</strong></p>
          <ul>
            <li>Large, complex software systems.</li>
            <li>Applications with GUIs.</li>
            <li>Projects requiring high scalability.</li>
          </ul>
        </div>
        <div>
          <p><strong>🔴 Avoid If:</strong></p>
          <ul>
            <li>Writing simple utilities/scripts.</li>
            <li>Building strict real-time systems with ultra-low memory.</li>
          </ul>
        </div>
      </div>
    </div>
