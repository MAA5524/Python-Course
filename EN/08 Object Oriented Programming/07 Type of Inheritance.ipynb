<div class="section-card" id="toc1_">
  <h2>1. Single Inheritance: Basics and Benefits</h2>
  <div class="container">
    <div class="card green">
      <h3>Concept</h3>
      <p>The simplest form: One subclass inherits from one superclass.</p>
      <img src="../../images/single-inheritance.png" alt="Single Inheritance">
    </div>
    <div class="card green">
      <h3>Example: Person Class</h3>
      <p>Let's start with a base <code>Person</code> class.</p>

  </div>

<div class="section-card" id="toc2_">
  <h2>2. Multilevel Inheritance: Advantages and Challenges</h2>
  <div class="container">
    <div class="card orange">
      <h3>Concept</h3>
      <p>A chain of inheritance: Grandparent -> Parent -> Child. Each level extends the previous one.</p>
      <img src="../../images/multi-level-inheritance.png" alt="Multilevel Inheritance">
    </div>
    <div class="card orange">
      <h3>Example: Employee Hierarchy</h3>
      <p>Let's define a 3-level structure: <code>Person</code> -> <code>Employee</code> -> <code>Manager</code>.</p>
    </div>
  </div>
</div>

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_info(self):
        print(f"Name: {self.name}, Age: {self.age}")
```

```python
class Employee(Person):
    def __init__(self, name, age, employee_id):
        super().__init__(name, age)
        self.employee_id = employee_id

    def display_employee_info(self):
        print(f"Employee ID: {self.employee_id}")
        self.display_info()
```

```python
class Manager(Employee):
    def __init__(self, name, age, employee_id, department):
        super().__init__(name, age, employee_id)
        self.department = department

    def display_manager_info(self):
        print(f"Department: {self.department}")
        self.display_employee_info()
```

    <!-- Advantages Card -->
    <div class="card orange">
      <h3>✅ Advantages</h3>
      <ul>
        <li><strong>Organized Structure:</strong> Provides a clear and hierarchical organization. Each class is an extension of the one above it, mirroring real-world relationships intuitively.</li>
        <li><strong>Incremental Enhancement:</strong> Allows for incremental additions. Each class in the hierarchy focuses on extending or customizing its parent, making the code manageable.</li>
        <li><strong>Reusability and Efficiency:</strong> Functionalities defined in upper-level classes cascade down, reducing redundancy and increasing efficiency.</li>
      </ul>
    </div>
    <!-- Challenges Card -->
    <div class="card orange">
      <h3>⚠️ Challenges</h3>
      <ul>
        <li><strong>Increased Complexity:</strong> As the chain grows, maintaining the codebase can become challenging. Long chains introduce difficulties in tracing the flow of logic or debugging.</li>
        <li><strong>Coupling:</strong> Classes are tightly coupled. A change in a top-level parent can have rippling effects all the way down the hierarchy.</li>
        <li><strong>Risk of Overriding Issues:</strong> With multiple levels, there's an increased risk of accidentally overriding methods, which could break functionality unexpectedly.</li>
      </ul>
    </div>
  </div>

<div class="section-card" id="toc3_">
  <h2>3. Hierarchical Inheritance: Structure and Scenarios</h2>
  <div class="container">
    <div class="card purple">
      <h3>Concept</h3>
      <p>One parent, multiple children. Creates a tree-like structure.</p>
      <img src="../../images/hierarchical-inheritance.png" alt="Hierarchical Inheritance">
    </div>
    <div class="card purple">
      <h3>Example: Vehicles</h3>
      <p>A base <code>Vehicle</code> class extended by <code>Car</code>, <code>Truck</code>, and <code>Motorcycle</code>.</p>
    </div>
  </div>
</div>

```python
class Vehicle:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.make} {self.model}")
```

```python
class Car(Vehicle):
    def __init__(self, make, model, year, doors):
        super().__init__(make, model, year)
        self.doors = doors

class Truck(Vehicle):
    def __init__(self, make, model, year, payload):
        super().__init__(make, model, year)
        self.payload = payload

class Motorcycle(Vehicle):
    def __init__(self, make, model, year, cc):
        super().__init__(make, model, year)
        self.cc = cc
```

      <h3>Example: Hybrid Animal</h3>
      <p>An <code>Alligator</code> that inherits from both <code>Walker</code> and <code>Swimmer</code>.</p>

```python
class Walker:
    def walk(self):
        print("This animal walks")

class Swimmer:
    def swim(self):
        print("This animal swims")

class Alligator(Walker, Swimmer):
    pass
```

```python
# Creating an instance of Alligator
ally = Alligator()
```

```python
ally.walk()
```

```python
ally.swim()
```

        <li><strong>Single Inheritance:</strong> <code>Person</code> -> <code>UniversityMember</code>.</li>
        <li><strong>Multilevel Inheritance:</strong> <code>UniversityMember</code> -> <code>Student</code> / <code>Instructor</code>.</li>
        <li><strong>Hierarchical Inheritance:</strong> Both <code>Student</code> and <code>Instructor</code> inherit from <code>UniversityMember</code>.</li>
      </ol>
    </div>
    <div class="card green">
      <h3>Solution</h3>
      <p>Here is the complete implementation of the University Hierarchy.</p>

```python
# Task 1: Implement Single Inheritance
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce_self(self):
        print(f"Hello! My name is {self.name}, and I am {self.age} years old.")
```

```python
class UniversityMember(Person):
    def __init__(self, name, age, university_name):
        super().__init__(name, age)
        self.university_name = university_name

    def introduce_self(self):
        super().introduce_self()
        print(f"I am a member of {self.university_name}.")
```

```python
# Task 2: Implement Multilevel Inheritance
class Student(UniversityMember):
    def __init__(self, name, age, university_name, student_id):
        super().__init__(name, age, university_name)
        self.student_id = student_id

    def introduce_self(self):
        super().introduce_self()
        print(f"I am a student with ID: {self.student_id}.")
```

```python
class Instructor(UniversityMember):
    def __init__(self, name, age, university_name, employee_id):
        super().__init__(name, age, university_name)
        self.employee_id = employee_id

    def introduce_self(self):
        super().introduce_self()
        print(f"I am an instructor with ID: {self.employee_id}.")
```

```python
# Task 3: Implement Hierarchical Inheritance
# Instantiate objects with different universities
alice = Student(name="Alice", age=24, university_name="XYZ University", student_id="S123")
bob = Instructor(name="Bob", age=35, university_name="ABC University", employee_id="E456")
```

```python
# Demonstrate hierarchical inheritance outputs
alice.introduce_self()
bob.introduce_self()
```
