  <h2>The 3 Types of Methods</h2>
  <div class="container">
    <div class="card blue">
      <h3>Overview</h3>
      <p>Not all methods are created equal. Python has three primary types:</p>
      <ul>
        <li><strong>Instance Methods:</strong> Operate on an object (`self`). Access/modify instance state.</li>
        <li><strong>Static Methods:</strong> Independent utility functions (`@staticmethod`). No `self` or `cls`.</li>
        <li><strong>Class Methods:</strong> Operate on the class itself (`@classmethod`, `cls`). Modify class state.</li>
      </ul>
    </div>
  </div>
</div>

<div class="section-card" id="toc1_">
  <h2>1. Instance Methods</h2>
  <div class="container">
    <div class="card green">
      <h3>The Backbone of OOP</h3>
      <p>These methods act on an instance. They require <code>self</code> as the first parameter to access data unique to that object.</p>
    </div>
  </div>
</div>

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} says woof!"
```

```python
# Creating an instance of Dog
my_dog = Dog("Buddy", 4)
```

```python
# Calling an instance method
my_dog.bark()
```

<div class="section-card" id="toc2_">
  <h2>2. Static Methods</h2>
  <div class="container">
    <div class="card orange">
      <h3>Independent Utilities</h3>
      <p>Static methods belong to a class but don't need access to class or instance data. They are marked with <code>@staticmethod</code>.</p>
      <p><strong>Use Case:</strong> Utility functions (e.g., math calculations, conversions) that are logically related to the class.</p>
    </div>
  </div>
</div>

```python
class MathOperations:
    @staticmethod
    def calculate_square_root(number):
        return number ** 0.5
```

```python
result = MathOperations.calculate_square_root(9)
result
```

```python
class TemperatureConverter:
    # Class definition here
    pass
```

```python
class TemperatureConverter:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (celsius * 9/5) + 32
```

<div class="section-card" id="toc3_">
  <h2>3. Class Methods</h2>
  <div class="container">
    <div class="card purple">
      <h3>Managing Class State</h3>
      <p>Class methods take <code>cls</code> as the first parameter. They are marked with <code>@classmethod</code>. They can modify class variables that affect <strong>all</strong> instances.</p>
      <p><strong>Common Use Case:</strong> Factory methods (alternative constructors) and singletons.</p>
    </div>
  </div>
</div>

```python
class Singleton:

    _instance = None

    def __new__(cls):
        if cls._instance is None:
            print("Creating the Singleton instance")
            cls._instance = super(Singleton, cls).__new__(cls)
        return cls._instance

    @classmethod
    def get_instance(cls):
        if not cls._instance:
            cls()
        return cls._instance
```

```python
obj1 = Singleton()
```

```python
obj2 = Singleton()
```

```python
obj1 is obj2
```

<div class="section-card">
  <div class="container">
    <div class="card purple">
      <h3>Example: Modifying Class State</h3>
      <p>Let's use a class method to change the discount rate for <strong>all</strong> items in an inventory.</p>
    </div>
  </div>
</div>

```python
class Inventory:
    discount_rate = 10  # percent

    def __init__(self, product_name, product_price):
        self.product_name = product_name
        self.product_price = product_price
```

```python
class Inventory:
    discount_rate = 10  # percent

    def __init__(self, product_name, product_price):
        self.product_name = product_name
        self.product_price = product_price

    @classmethod
    def set_discount_rate(cls, new_rate):
        cls.discount_rate = new_rate
```

```python
# Initial discount rate
print("Initial discount rate:", Inventory.discount_rate)
```

```python
# Changing the discount rate with a class method
Inventory.set_discount_rate(15)
print("New discount rate:", Inventory.discount_rate)
```

<div class="section-card" id="toc4_">
  <h2>4. Comparison & Summary</h2>
  <div class="container">
    <div class="card green">
      <h3>Summary Table</h3>
      <p><strong>Instance Method:</strong> Uses <code>self</code>. Modifies object state. (Default)</p>
      <p><strong>Static Method:</strong> No <code>self/cls</code>. Independent logic. (Helper)</p>
      <p><strong>Class Method:</strong> Uses <code>cls</code>. Modifies class state. (Factory/Config)</p>
    </div>
    <div class="card green">
      <h3>Real-World Use Cases</h3>
      <ul>
        <li><strong>Bank Account (Instance):</strong> <code>deposit(self, amount)</code> - Updates specific balance.</li>
        <li><strong>Math Helper (Static):</strong> <code>area(radius)</code> - Pure calculation.</li>
        <li><strong>Employee (Class):</strong> <code>from_string(cls, str)</code> - Alternative constructor.</li>
      </ul>
    </div>
  </div>
</div>

```python
class BankAccount:
    def __init__(self, balance):
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
```

```python
class MathHelper:

    @staticmethod
    def area_of_circle(radius):
        return 3.14 * radius * radius
```

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def from_string(cls, employee_str):
        name, salary = employee_str.split('-')
        return cls(name, int(salary))
```

```python
emp = Employee.from_string("John-5000")
```

```python
emp.name, emp.salary
```
