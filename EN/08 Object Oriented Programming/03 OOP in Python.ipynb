<div class="section-card" id="toc1_">
  <h2>1. Basic Concepts of OOP in Python</h2>
  <div class="container">
    <div class="card purple">
      <h3>Classes and Objects</h3>
      <ul>
        <li><strong>Class:</strong> A blueprint or template. It encapsulates data (attributes) and methods to interact with that data.</li>
        <li><strong>Object:</strong> An instance of a class. It embodies the actual data and functionality defined by the class.</li>
      </ul>
      <img src="../../images/class-objects.png" alt="Class and Objects">
    </div>
    <div class="card purple">
      <h3>Simple Class Creation Example</h3>
      <p>Let’s create a simple <code>Dog</code> class with two attributes (<code>name</code>, <code>age</code>) and one method (<code>bark</code>).</p>
    </div>
  </div>
</div>

```python
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f"{self.name} says Woof!"
```

<div class="section-card" id="toc2_">
  <h2>2. Key Features of Python OOP (The 4 Pillars)</h2>
  <div class="container">
    <div class="card orange">
      <h3>1. Encapsulation</h3>
      <p><strong>Concept:</strong> Hiding the internal details and exposing only the necessary parts. It bundles the data and methods into a single unit, preventing accidental modification by restricting direct access.</p>
      <img src="../../images/oop-encapsulation.png" alt="Encapsulation">
      <p><strong>Example:</strong> A bank account where you use methods to deposit instead of directly changing the balance.</p>
    </div>
    <div class="card orange">
      <p>In the below example, the <code>__balance</code> is a private attribute (indicated by double underscores), encapsulated within the <code>Account</code> class. It cannot be accessed directly from outside the class.</p>
    </div>
  </div>
</div>

```python
class Account:
    def __init__(self, owner, amount=0):
        self.owner = owner
        self.__balance = amount  # Private attribute

    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print("Deposit Successful")

    def show_balance(self):
        print(f"Balance: {self.__balance}")
```

```python
acc = Account("John")
```

```python
acc.deposit(100)
```

```python
acc.show_balance()  # Output: Balance: 100
```

<div class="section-card">
  <div class="container">
    <div class="card blue">
      <h3>3. Polymorphism</h3>
      <p><strong>Concept:</strong> "Many forms". Allows objects of different classes to be treated as objects of a common superclass. Enabling methods to do different things based on the object.</p>
      <img src="../../images/oop-polymorphism.png" alt="Polymorphism">
      <p><strong>Example:</strong> A single <code>fly()</code> command acting differently on a Parrot vs. a Penguin.</p>
    </div>
    <div class="card blue">
      <p>In the below example, the <code>Bird</code> class function <code>fly</code> is overridden in both subclasses. When <code>test_flying</code> calls <code>bird.fly()</code>, it automatically knows which version to execute based on the object passed.</p>
    </div>
  </div>
</div>

```python
class Bird:
    def fly(self):
        print("Some birds can fly")

class Parrot(Bird):
    def fly(self):
        print("Parrots can fly")

class Penguin(Bird):
    def fly(self):
        print("Penguins do not fly")
```

```python
# Polymorphism in action
def test_flying(bird):
    bird.fly()
```

```python
test_flying(Parrot())  # Output: Parrots can fly
```

```python
test_flying(Penguin())  # Output: Penguins do not fly
```

<div class="section-card">
  <div class="container">
    <div class="card pink">
      <h3>4. Abstraction</h3>
      <p><strong>Concept:</strong> Hiding complex reality and exposing only necessary parts. Usually implemented using Abstract Base Classes (ABCs) in Python using the <code>abc</code> module.</p>
      <p>It enforces a contract: subclasses <em>must</em> implement specific methods (like <code>area</code> and <code>perimeter</code> for a <code>Shape</code>).</p>
    </div>
  </div>
</div>

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

    @abstractmethod
    def perimeter(self):
        pass
```

```python
class Square(Shape):
    def __init__(self, side):
        self.__side = side

    def area(self):
        return self.__side * self.__side

    def perimeter(self):
        return 4 * self.__side
```

```python
square = Square(5)
square.area()
```

        <li><strong><code>__init__</code>:</strong> A special method called automatically to initialize new objects.</li>
        <li><strong><code>self</code>:</strong> Refers to the current instance of the class.</li>
        <li><strong>Instance Attributes:</strong> Like <code>self.model</code>, stores specific info for that object.</li>
        <li><strong>Instance Methods:</strong> Like <code>honk()</code>, functions defined inside the class that only the object can call.</li>
