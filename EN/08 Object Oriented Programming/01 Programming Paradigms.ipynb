<div class="section-card" id="toc1_">
  <h2>1. Programming Paradigm Overview</h2>
  <div class="container">
    <div class="card green">
      <h3>Imperative vs. Declarative</h3>
      <p>Paradigms are broadly split into two camps:</p>
      <ul>
        <li><strong>Imperative:</strong> Focuses on <em>HOW</em> to solve a problem (step-by-step).</li>
        <li><strong>Declarative:</strong> Focuses on <em>WHAT</em> the result should be (logic-based).</li>
      </ul>
      <img src="../../images/programming-paradigms.jpg" alt="Paradigms Tree">
    </div>
    <div class="card green">
      <h3>It's Not a Language</h3>
      <p>A paradigm is a style, not a language. Python is a <strong>multi-paradigm</strong> language, meaning it supports OOP, Procedural, and Functional styles simultaneously.</p>
    </div>
  </div>
</div>

<!-- Imperative Section -->
<div class="section-card" id="toc2_">
  <h2>2. Imperative Paradigm</h2>
  <div class="container">
    <div class="card orange">
      <h3>The "How" Approach</h3>
      <p>Like a recipe, you give the computer a sequence of instructions. It uses loops, conditionals, and state changes.</p>
      <img src="../../images/how-what.webp" alt="How vs What">
    </div>
    <div class="card orange">
      <h3>Procedural Programming</h3>
      <p>Breaks tasks into functions (procedures). Best for linear, simple tasks.</p>
    </div>
  </div>
</div>

```python
def sum_of_evens(numbers):
    total = 0
    for num in numbers:
        if num % 2 == 0:
            total += num
    return total
```

```python
numbers = [1, 2, 3, 4, 5, 6]
result = sum_of_evens(numbers)
result
```

  <div class="container">
    <div class="card orange">
      <h3>Object-Oriented Programming (OOP)</h3>
      <p>Organizes code into objects that hold data and methods. Focuses on <strong>Encapsulation</strong>, <strong>Inheritance</strong>, and <strong>Polymorphism</strong>.</p>
    </div>
  </div>
</div>

```python
class Circle:
    pi = 3.14159

    def __init__(self, radius):
        self.radius = radius

    def calculate_area(self):
        return self.pi * (self.radius ** 2)
```

```python
circle1 = Circle(5)
circle2 = Circle(10)

print(f"The area of circle1 with radius {circle1.radius} is {circle1.calculate_area()}.")
print(f"The area of circle2 with radius {circle2.radius} is {circle2.calculate_area()}.")
```

<div class="section-card" id="toc3_">
  <h2>3. Declarative Paradigm</h2>
  <div class="container">
    <div class="card purple">
      <h3>The "What" Approach</h3>
      <p>Focuses on describing the desired result, abstracting away the control flow steps.</p>
      <img src="../../images/imperative-declarative.gif" alt="Imperative vs Declarative">
    </div>
    <div class="card purple">
      <h3>Functional Programming</h3>
      <p>Based on pure functions and immutability. Python supports this via <code>lambda</code>, <code>map</code>, <code>filter</code>.</p>
    </div>
  </div>
</div>

```python
numbers = [1, 2, 3, 4, 5, 6]
result = sum(filter(lambda x: x % 2 == 0, numbers))
result
```

<div class="section-card" id="toc4_">
  <h2>4. Comparison & Guidelines</h2>
  <div class="container">
    <div class="card red">
      <h3>Imperative</h3>
      <p><strong>✅ Pros:</strong> High control, familiar syntax.<br><strong>❌ Cons:</strong> Can become complex and hard to scale.</p>
      <img src="../../images/imperative-paradigm.png" alt="Imperative Pros Cons">
    </div>
    <div class="card red">
      <h3>Declarative</h3>
      <p><strong>✅ Pros:</strong> Concise, readable, maintainable.<br><strong>❌ Cons:</strong> Less control over performance, steeper learning curve.</p>
      <img src="../../images/declarative-paradigm.png" alt="Declarative Pros Cons">
    </div>
    <div class="card red">
      <h3>How to Choose?</h3>
      <p>Consider the problem domain, performance needs, and team expertise. Python allows you to mix paradigms (Multi-Paradigm), so use the best tool for each specific part of your code!</p>
    </div>
  </div>
</div>
